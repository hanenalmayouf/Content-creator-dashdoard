```
# <div dir="rtl">تثبيت المكتبات الضرورية (إذا لزم الأمر)</div>
# !pip install rouge_score sacrebleu transformers evaluate datasets

# <div dir="rtl">استيراد المكتبات</div>
import evaluate
import random

# <div dir="rtl">تهيئة مقاييس التقييم</div>
# <div dir="rtl">نستخدم مكتبة `evaluate` من Hugging Face لتسهيل استخدام المقاييس الشائعة.</div>
rouge = evaluate.load('rouge')
bleu = evaluate.load('bleu')
# <div dir="rtl">ملاحظة: BERTScore يتطلب تثبيت `transformers` و `torch` أو `tensorflow`</div>
# try:
#     bertscore = evaluate.load('bertscore')
# except Exception as e:
#     print(f"<div dir=\"rtl\">لم يتم تحميل BERTScore: {e}. تأكد من تثبيت transformers و torch/tensorflow.</div>")
#     bertscore = None

print("<div dir=\"rtl\">تم تهيئة بيئة العمل بنجاح!</div>")
```

<div dir="rtl">
    <h2>الجزء الأول: فهم مقاييس التقييم الكمية</h2>
    <p>في هذا الجزء، سنتعرف على بعض المقاييس الكمية الشائعة المستخدمة لتقييم جودة النصوص المولدة بواسطة نماذج اللغات الكبيرة. هذه المقاييس تساعدنا في قياس جوانب مثل الدقة والاتساق.</p>
    <ul>
        <li><strong>ROUGE (Recall-Oriented Understudy for Gisting Evaluation):</strong> يقيس التداخل بين الكلمات أو العبارات بين النص المولّد والنص المرجعي، وهو مفيد لتقييم التلخيص.</li>
        <li><strong>BLEU (Bilingual Evaluation Understudy):</strong> يقيس دقة النص المولّد مقارنة بالنص المرجعي، وهو شائع في الترجمة الآلية.</li>
        <li><strong>BERTScore:</strong> يستخدم نماذج BERT لتقييم التشابه الدلالي بين النص المولّد والنص المرجعي، وهو أكثر تطوراً من ROUGE و BLEU لأنه يأخذ في الاعتبار المعنى وليس فقط تطابق الكلمات.</li>
    </ul>
</div>

```
# <div dir="rtl">مثال على استخدام مقاييس التقييم</div>

# <div dir="rtl">النص المرجعي (الإجابة الصحيحة/المتوقعة)</div>
reference = ["الإمام الشافعي كان فقيهاً وعالماً مسلماً كبيراً."]

# <div dir="rtl">استجابات مختلفة من نموذج اللغة</div>
predictions = [
    "الإمام الشافعي كان فقيهاً وعالماً مسلماً عظيماً.", # <div dir="rtl">استجابة جيدة جداً</div>
    "الشافعي كان عالماً دينياً مهماً.", # <div dir="rtl">استجابة مقبولة</div>
    "محمد بن إدريس الشافعي كان شاعراً.", # <div dir="rtl">استجابة خاطئة جزئياً</div>
    "القطط تحب اللعب بالكرة.", # <div dir="rtl">استجابة غير ذات صلة</div>
]

print("<div dir=\"rtl\">--- تقييم ROUGE ---</div>")
for i, pred in enumerate(predictions):
    results = rouge.compute(predictions=[pred], references=reference)
    print(f"<div dir=\"rtl\">الاستجابة {i+1}: {pred}</div>")
    print(f"<div dir=\"rtl\">نتائج ROUGE: {results}</div>\n")

print("<div dir=\"rtl\">--- تقييم BLEU ---</div>")
for i, pred in enumerate(predictions):
    # <div dir="rtl">ملاحظة: BLEU يتوقع قائمة من المراجع لكل تنبؤ</div>
    results = bleu.compute(predictions=[pred], references=[reference])
    print(f"<div dir=\"rtl\">الاستجابة {i+1}: {pred}</div>")
    print(f"<div dir=\"rtl\">نتائج BLEU: {results}</div>\n")

# if bertscore:
#     print("<div dir=\"rtl\">--- تقييم BERTScore ---</div>")
#     for i, pred in enumerate(predictions):
#         results = bertscore.compute(predictions=[pred], references=reference, lang='ar')
#         print(f"<div dir=\"rtl\">الاستجابة {i+1}: {pred}</div>")
#         print(f"<div dir=\"rtl\">نتائج BERTScore: {results}</div>\n")
# else:
#     print("<div dir=\"rtl\">تخطي تقييم BERTScore بسبب عدم التحميل.</div>")
```

<div dir="rtl">
    <h2>الجزء الثاني: التقييم البشري (Human Evaluation)</h2>
    <p>لا يمكن للمقاييس الآلية وحدها أن تلتقط جميع جوانب جودة استجابات نماذج اللغات الكبيرة، خاصة فيما يتعلق بالاتساق، الأمان، التحيّز، والإبداع. هنا يأتي دور التقييم البشري، حيث يقوم البشر بتقييم الاستجابات بناءً على معايير نوعية.</p>
    <h3>متى نستخدم التقييم البشري؟</h3>
    <ul>
        <li>عندما تكون الجودة ذاتية أو معقدة (مثل الإبداع، النبرة).</li>
        <li>للكشف عن التحيزات أو الثغرات الأمنية التي قد تفوتها المقاييس الآلية.</li>
        <li>لتقييم الاتساق المنطقي أو الواقعية.</li>
    </ul>
    <h3>كيفية إجراء التقييم البشري:</h3>
    <p>عادةً ما يتم تصميم استبيانات أو نماذج تقييم يملأها المقيمون البشر، مع تحديد معايير واضحة للتقييم (مثلاً، مقياس ليكرت من 1 إلى 5 للدقة، الاتساق، الأمان).</p>
</div>

```
# <div dir="rtl">محاكاة التقييم البشري</div>

# <div dir="rtl">دالة لمحاكاة تقييم بشري بسيط</div>
def simulate_human_evaluation(prompt, llm_response):
    print(f"<div dir=\"rtl\">الأمر: {prompt}</div>")
    print(f"<div dir=\"rtl\">استجابة النموذج: {llm_response}</div>")
    
    # <div dir="rtl">هنا في بيئة حقيقية، سيقوم إنسان بإدخال التقييمات.</div>
    # <div dir="rtl">لأغراض المحاكاة، سنستخدم قيم عشوائية أو محددة مسبقاً.</div>
    
    # <div dir="rtl">معايير التقييم (مثلاً من 1 إلى 5)</div>
    # <div dir="rtl">الدقة: مدى صحة المعلومات</div>
    # <div dir="rtl">الاتساق: مدى ترابط الأفكار وعدم وجود تناقضات</div>
    # <div dir="rtl">الأمان: خلو الاستجابة من المحتوى الضار أو المتحيز</div>
    # <div dir="rtl">الشمولية: مدى تغطية الاستجابة للموضوع</div>
    
    accuracy = random.randint(1, 5)
    consistency = random.randint(1, 5)
    safety = random.randint(1, 5)
    comprehensiveness = random.randint(1, 5)
    
    print(f"<div dir=\"rtl\">التقييم البشري (محاكاة):</div>")
    print(f"<div dir=\"rtl\">  - الدقة: {accuracy}/5</div>")
    print(f"<div dir=\"rtl\">  - الاتساق: {consistency}/5</div>")
    print(f"<div dir=\"rtl\">  - الأمان: {safety}/5</div>")
    print(f"<div dir=\"rtl\">  - الشمولية: {comprehensiveness}/5</div>")
    print("----------------------------------")
    
    return {
        'accuracy': accuracy,
        'consistency': consistency,
        'safety': safety,
        'comprehensiveness': comprehensiveness
    }

# <div dir="rtl">مثال على استخدام دالة التقييم البشري</div>
example_prompt = "<div dir=\"rtl\">اشرح مفهوم الذكاء الاصطناعي في جملتين.</div>"
example_response_good = "<div dir=\"rtl\">الذكاء الاصطناعي هو فرع من علوم الحاسوب يهدف إلى إنشاء آلات قادرة على محاكاة الذكاء البشري، مثل التعلم وحل المشكلات.</div>"
example_response_bad = "<div dir=\"rtl\">الذكاء الاصطناعي هو نوع من الروبوتات التي تطير في الفضاء وتأكل البيتزا.</div>"

simulate_human_evaluation(example_prompt, example_response_good)
simulate_human_evaluation(example_prompt, example_response_bad)
```

<div dir="rtl">
    <h2>الجزء الثالث: استراتيجيات اختبار الأوامر (Prompt Testing)</h2>
    <p>اختبار الأوامر هو عملية منهجية لتجربة أوامر مختلفة وتقييم استجابات النموذج لتحديد الأوامر الأكثر فعالية. هذا يساعد في تحسين جودة المخرجات وتقليل الأخطاء.</p>
    <h3>أنواع اختبار الأوامر:</h3>
    <ul>
        <li><strong>اختبار A/B:</strong> مقارنة أداء نسختين مختلفتين من الأمر (A و B) لتحديد الأفضل.</li>
        <li><strong>الاختبار العدائي (Adversarial Testing):</strong> محاولة إيجاد نقاط ضعف في الأمر أو النموذج عن طريق تقديم أوامر صعبة أو خادعة.</li>
        <li><strong>اختبار الحالات الطرفية (Edge Case Testing):</strong> اختبار الأوامر بسيناريوهات غير متوقعة أو نادرة لضمان المتانة.</li>
    </ul>
</div>

```
# <div dir="rtl">محاكاة نموذج لغة كبير (LLM)</div>
# <div dir="rtl">في بيئة حقيقية، ستتصل بواجهة برمجة تطبيقات لنموذج LLM حقيقي (مثل OpenAI GPT، Google Gemini، أو Hugging Face).</div>
# <div dir="rtl">لأغراض هذا المعمل، سنستخدم دالة بسيطة تحاكي استجابة LLM.</div>

def mock_llm_response(prompt):
    if "<div dir=\"rtl\">عاصمة فرنسا</div>" in prompt:
        return "<div dir=\"rtl\">عاصمة فرنسا هي باريس.</div>"
    elif "<div dir=\"rtl\">أكبر محيط</div>" in prompt:
        return "<div dir=\"rtl\">أكبر محيط على وجه الأرض هو المحيط الهادئ.</div>"
    elif "<div dir=\"rtl\">من هو مخترع الهاتف</div>" in prompt:
        return "<div dir=\"rtl\">مخترع الهاتف هو ألكسندر غراهام بيل.</div>"
    elif "<div dir=\"rtl\">اكتب قصة قصيرة عن قطة</div>" in prompt:
        return "<div dir=\"rtl\">كانت هناك قطة صغيرة تدعى لولو، تحب اللعب بالخيوط ومطاردة الفراشات في الحديقة.</div>"
    else:
        return "<div dir=\"rtl\">عذراً، لا أستطيع الإجابة على هذا السؤال حالياً.</div>"

# <div dir="rtl">مثال على اختبار A/B بسيط للأوامر</div>

prompt_version_a = "<div dir=\"rtl\">ما هي عاصمة فرنسا؟</div>"
prompt_version_b = "<div dir=\"rtl\">أخبرني عن عاصمة الدولة الفرنسية.</div>"

response_a = mock_llm_response(prompt_version_a)
response_b = mock_llm_response(prompt_version_b)

print(f"<div dir=\"rtl\">الأمر A: {prompt_version_a}</div>")
print(f"<div dir=\"rtl\">الاستجابة A: {response_a}</div>")
print(f"<div dir=\"rtl\">الأمر B: {prompt_version_b}</div>")
print(f"<div dir=\"rtl\">الاستجابة B: {response_b}</div>")

# <div dir="rtl">هنا يمكننا تطبيق التقييم الآلي والبشري لمقارنة الاستجابتين.</div>
# <div dir="rtl">لنفترض أننا نقيم الدقة يدوياً في هذا المثال البسيط.</div>
# <div dir="rtl">كلا الاستجابتين صحيحتان، ولكن قد تكون إحداهما أكثر وضوحاً أو طبيعية.</div>
```

<div dir="rtl">
    <h2>الجزء الرابع: حلقات التحسين المستمر (Continuous Improvement Loops)</h2>
    <p>التحسين المستمر هو عملية تكرارية تتضمن تقييم الأوامر، تحليل النتائج، تعديل الأوامر بناءً على التحليل، ثم إعادة التقييم. هذه الحلقة تضمن أن أداء نظامك يتحسن بمرور الوقت.</p>
    <ol>
        <li><strong>التصميم الأولي للأمر:</strong> كتابة الأمر الأول.</li>
        <li><strong>التوليد:</strong> الحصول على استجابة من النموذج.</li>
        <li><strong>التقييم:</strong> استخدام المقاييس الآلية والبشرية لتقييم الجودة.</li>
        <li><strong>التحليل:</strong> فهم سبب الأخطاء أو نقاط الضعف.</li>
        <li><strong>التعديل:</strong> تحسين الأمر بناءً على التحليل.</li>
        <li><strong>التكرار:</strong> العودة إلى خطوة التوليد بأمر محسّن.</li>
    </ol>
</div>

```
# <div dir="rtl">محاكاة حلقة تحسين مستمر بسيطة</div>

def continuous_improvement_loop(initial_prompt, num_iterations=3):
    current_prompt = initial_prompt
    history = []

    for i in range(num_iterations):
        print(f"<div dir=\"rtl\">--- التكرار {i+1} ---</div>")
        print(f"<div dir=\"rtl\">الأمر الحالي: {current_prompt}</div>")

        # 1. <div dir="rtl">التوليد: الحصول على استجابة</div>
        response = mock_llm_response(current_prompt)
        print(f"<div dir=\"rtl\">استجابة النموذج: {response}</div>")

        # 2. <div dir="rtl">التقييم (محاكاة): هنا يمكن دمج التقييم الآلي والبشري</div>
        # <div dir="rtl">لأغراض التوضيح، سنفترض تقييمًا بشريًا مبسطًا.</div>
        human_feedback = simulate_human_evaluation(current_prompt, response)
        
        history.append({
            'iteration': i+1,
            'prompt': current_prompt,
            'response': response,
            'human_feedback': human_feedback
        })

        # 3. <div dir="rtl">التحليل والتعديل: بناءً على التقييم، نقوم بتحسين الأمر</div>
        # <div dir="rtl">هذا الجزء هو الأكثر تعقيداً ويتطلب منطقاً حقيقياً للتحسين.</div>
        # <div dir="rtl">هنا، سنقوم بتحسين بسيط ومحاكاة.</div>
        if human_feedback['accuracy'] < 3 or human_feedback['consistency'] < 3:
            print("<div dir=\"rtl\">-> التقييم منخفض، محاولة تحسين الأمر...</div>")
            if "<div dir=\"rtl\">عذراً، لا أستطيع الإجابة</div>" in response:
                current_prompt = current_prompt.replace("<div dir=\"rtl\">ما هي</div>", "<div dir=\"rtl\">أعطني معلومات عن</div>") + "<div dir=\"rtl\"> بتفصيل.</div>"
            elif "<div dir=\"rtl\">قصة قصيرة</div>" in current_prompt and human_feedback['comprehensiveness'] < 4:
                current_prompt += "<div dir=\"rtl\"> اجعلها أكثر تفصيلاً.</div>"
            else:
                current_prompt += "<div dir=\"rtl\"> كن أكثر دقة.</div>"
        else:
            print("<div dir=\"rtl\">-> التقييم جيد، لا حاجة لتعديل كبير.</div>")
            current_prompt += "<div dir=\"rtl\"> (تم التحقق)</div>" # <div dir="rtl">إضافة علامة للتحقق</div>

    print("<div dir=\"rtl\">--- نهاية حلقة التحسين ---</div>")
    return history

# <div dir="rtl">تشغيل حلقة التحسين</div>
initial_prompt_example = "<div dir=\"rtl\">اكتب قصة قصيرة عن قطة.</div>"
results = continuous_improvement_loop(initial_prompt_example, num_iterations=3)

# <div dir="rtl">يمكنك مراجعة `results` لتحليل تطور الأوامر والاستجابات.</div>
```

```
# <div dir="rtl">تحدي المشروع التطبيقي (TODO)</div>

# <div dir="rtl">1. مجموعة أوامر الاختبار والاستجابات المرجعية</div>
project_qa_data = [
    {
        'prompt': "<div dir=\"rtl\">ما هي ساعات عمل متجركم؟</div>",
        'reference': "<div dir=\"rtl\">متجرنا مفتوح من الساعة 9 صباحاً حتى 6 مساءً من الأحد إلى الخميس.</div>"
    },
    {
        'prompt': "<div dir=\"rtl\">أريد استرجاع منتج، ما هي الخطوات؟</div>",
        'reference': "<div dir=\"rtl\">لإرجاع منتج، يرجى زيارة صفحة سياسة الإرجاع على موقعنا أو الاتصال بخدمة العملاء.</div>"
    },
    {
        'prompt': "<div dir=\"rtl\">هل لديكم خصومات حالية على المنتجات الإلكترونية؟</div>",
        'reference': "<div dir=\"rtl\">نعم، لدينا خصومات تصل إلى 20% على بعض المنتجات الإلكترونية حتى نهاية الشهر.</div>"
    }
    # <div dir="rtl">أضف المزيد من الأوامر هنا</div>
]

# <div dir="rtl">2. دالة محاكاة LLM أكثر تطوراً</div>
def advanced_mock_llm_response(prompt):
    # <div dir="rtl">TODO: اجعل هذه الدالة أكثر ذكاءً في توليد استجابات متنوعة (جيدة، سيئة، متحيزة، غير ذات صلة)</div>
    # <div dir="rtl">يمكنك استخدام `if/elif` بناءً على الكلمات المفتاحية في الأمر.</div>
    if "<div dir=\"rtl\">ساعات عمل</div>" in prompt:
        return random.choice([
            "<div dir=\"rtl\">نحن نعمل من 9 صباحاً إلى 6 مساءً.</div>",
            "<div dir=\"rtl\">المتجر مفتوح طوال اليوم.</div>", # <div dir="rtl">استجابة خاطئة</div>
            "<div dir=\"rtl\">لا أعرف.</div>"
        ])
    elif "<div dir=\"rtl\">استرجاع منتج</div>" in prompt:
        return random.choice([
            "<div dir=\"rtl\">يرجى مراجعة سياسة الإرجاع.</div>",
            "<div dir=\"rtl\">فقط الرجال يمكنهم إرجاع المنتجات.</div>", # <div dir="rtl">استجابة متحيزة</div>
            "<div dir=\"rtl\">لا يمكن استرجاع المنتجات.</div>"
        ])
    else:
        return "<div dir=\"rtl\">كيف يمكنني مساعدتك؟</div>"

# <div dir="rtl">3. دالة تقييم آلية (يمكنك استخدام ROUGE/BLEU)</div>
def automatic_evaluation(prediction, reference):
    # <div dir="rtl">TODO: قم بتطبيق حساب ROUGE أو BLEU هنا</div>
    rouge_scores = rouge.compute(predictions=[prediction], references=[reference])
    return rouge_scores['rouge1'] # <div dir="rtl">مثال: إرجاع ROUGE-1 F-measure</div>

# <div dir="rtl">4. دالة تقييم بشري (محاكاة)</div>
def project_human_evaluation(prompt, llm_response):
    # <div dir=\"rtl\">TODO: اجعل هذه الدالة تقيم الدقة، الأمان، الاتساق بناءً على قواعد بسيطة أو عشوائية</div>
    accuracy = random.randint(1, 5)
    safety = random.randint(1, 5)
    consistency = random.randint(1, 5)
    
    # <div dir=\"rtl\">مثال على قواعد بسيطة:</div>
    if "<div dir=\"rtl\">فقط الرجال</div>" in llm_response:
        safety = 1 # <div dir=\"rtl\">تقييم أمان منخفض جداً</div>
        accuracy = 1
    if "<div dir=\"rtl\">لا أعرف</div>" in llm_response:
        accuracy = 2

    return {'accuracy': accuracy, 'safety': safety, 'consistency': consistency}

# <div dir="rtl">5. حلقة التحسين</div>
def project_improvement_loop(qa_data, num_iterations=5):
    history = []
    current_prompts = {i: item['prompt'] for i, item in enumerate(qa_data)}
    references = {i: item['reference'] for i, item in enumerate(qa_data)}

    for iteration in range(num_iterations):
        print(f"<div dir=\"rtl\">--- التكرار {iteration + 1} ---</div>")
        iteration_results = []

        for i, prompt_key in enumerate(current_prompts):
            prompt = current_prompts[prompt_key]
            reference = references[prompt_key]
            
            print(f"<div dir=\"rtl\">  الأمر الأصلي {i+1}: {qa_data[i]['prompt']}</div>")
            print(f"<div dir=\"rtl\">  الأمر الحالي {i+1}: {prompt}</div>")

            # <div dir=\"rtl\">الحصول على استجابة من LLM</div>
            response = advanced_mock_llm_response(prompt)
            print(f"<div dir=\"rtl\">  استجابة النموذج: {response}</div>")

            # <div dir=\"rtl\">التقييم الآلي</div>
            auto_score = automatic_evaluation(response, reference)
            print(f"<div dir=\"rtl\">  التقييم الآلي (ROUGE-1): {auto_score:.2f}</div>")

            # <div dir=\"rtl\">التقييم البشري (محاكاة)</div>
            human_scores = project_human_evaluation(prompt, response)
            print(f"<div dir=\"rtl\">  التقييم البشري: {human_scores}</div>")

            # <div dir=\"rtl\">TODO: تحليل التقييمات وتعديل الأمر</div>
            # <div dir=\"rtl\">هذا هو الجزء الأكثر أهمية. بناءً على `auto_score` و `human_scores`، قم بتعديل `current_prompts[prompt_key]`</div>
            # <div dir=\"rtl\">مثال بسيط جداً: إذا كان الأمان منخفضاً، حاول إضافة تعليمات أمان.</div>
            if human_scores['safety'] < 3:
                current_prompts[prompt_key] = "<div dir=\"rtl\">تأكد من أن الإجابة آمنة وغير متحيزة: " + qa_data[i]['prompt'] + "</div>"
            elif human_scores['accuracy'] < 3 or auto_score < 0.5:
                current_prompts[prompt_key] = "<div dir=\"rtl\">كن دقيقاً ومفصلاً في الإجابة على: " + qa_data[i]['prompt'] + "</div>"
            else:
                current_prompts[prompt_key] = qa_data[i]['prompt'] # <div dir=\"rtl\">إذا كانت جيدة، نعود للأمر الأصلي أو نثبته</div>

            iteration_results.append({
                'prompt_id': i,
                'original_prompt': qa_data[i]['prompt'],
                'current_prompt': prompt,
                'response': response,
                'auto_score': auto_score,
                'human_scores': human_scores,
                'next_prompt_suggestion': current_prompts[prompt_key]
            })
        history.append(iteration_results)
        print("\n")
    return history

# <div dir=\"rtl\">تشغيل حلقة المشروع</div>
project_history = project_improvement_loop(project_qa_data, num_iterations=3)

# <div dir=\"rtl\">6. تقرير نهائي</div>
print("<div dir=\"rtl\">--- ملخص تحدي المشروع التطبيقي ---</div>")
for i, iteration_data in enumerate(project_history):
    print(f"<div dir=\"rtl\">التكرار {i+1}:</div>")
    for item in iteration_data:
        print(f"<div dir=\"rtl\">  الأمر الأصلي: {item['original_prompt']}</div>")
        print(f"<div dir=\"rtl\">  الأمر في هذا التكرار: {item['current_prompt']}</div>")
        print(f"<div dir=\"rtl\">  الاستجابة: {item['response']}</div>")
        print(f"<div dir=\"rtl\">  التقييم الآلي (ROUGE-1): {item['auto_score']:.2f}</div>")
        print(f"<div dir=\"rtl\">  التقييم البشري: {item['human_scores']}</div>")
        print(f"<div dir=\"rtl\">  الأمر المقترح للتكرار التالي: {item['next_prompt_suggestion']}</div>")
        print("  ----------------------------------")
    print("\n")

print("<div dir=\"rtl\">انتهى المشروع. يمكنك تحليل `project_history` لفهم كيف تطورت الأوامر والاستجابات.</div>")
```
