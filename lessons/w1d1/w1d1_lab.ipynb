<div dir="rtl">## الفرق بين ETL و ELT

- **ETL (Extract, Transform, Load):** مثل طباخ يجهز المكونات في المطبخ قبل تقديمها للزبون. نعدل البيانات قبل تخزينها.
- **ELT (Extract, Load, Transform):** مثل بوفيه مفتوح، نضع كل شيء كما هو في المخزن، ثم نطبخ ما نحتاجه عند الطلب. هذا هو التوجه الحديث في السحابة (Cloud).</div>

```
# محاكاة بسيطة لعملية نقل البيانات
# لنفترض أن لدينا بيانات خام من مستشعرات حرارة

raw_data = [{"id": 1, "temp": 25.5}, {"id": 2, "temp": None}, {"id": 3, "temp": 28.0}]

# عملية التحويل (Transform): تنظيف البيانات من القيم المفقودة
def transform_data(data):
    # نفلتر البيانات التي لا تحتوي على درجة حرارة
    return [d for d in data if d['temp'] is not None]

cleaned_data = transform_data(raw_data)
print(f"البيانات بعد التنظيف: {cleaned_data}")
```
