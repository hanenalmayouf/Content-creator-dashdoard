<div dir="rtl"># معمل تصميم خطوط معالجة البيانات (ETL vs ELT)

أهلاً بكم يا شباب في معملنا اليوم. تخيّلوا أننا ندير متجراً إلكترونياً ضخماً، البيانات تأتينا من كل مكان: من تطبيق الجوال، من الموقع، ومن سجلات الدفع. كيف ننظم هذه البيانات لنستفيد منها؟ هذا ما نسميه "خط أنابيب البيانات" (Data Pipeline).</div>

<div dir="rtl">## الفرق بين ETL و ELT

1. **ETL (Extract, Transform, Load):** مثل طباخ يجهز المكونات في المطبخ (التحويل) قبل أن يقدمها للزبون (التحميل). نستخدمه عندما نحتاج بيانات نظيفة جداً ومحددة.
2. **ELT (Extract, Load, Transform):** مثل بوفيه مفتوح، نضع كل شيء (البيانات الخام) في المستودع (Load) ثم نطبخها ونحولها عند الحاجة. هذا هو التوجه الحديث مع السحابة (Cloud).</div>

```
# محاكاة بيانات خام من متجر إلكتروني
import pandas as pd

# بيانات خام غير منظمة (تاريخ، اسم المنتج، السعر)
raw_data = [
    {'date': '2023-10-01', 'item': 'تمر سكري', 'price': '50 SAR'},
    {'date': '2023-10-01', 'item': 'قهوة عربية', 'price': '30'},
    {'date': '02/10/2023', 'item': 'تمر خلاص', 'price': '60 SAR'}
]

# تحويل البيانات إلى DataFrame
df = pd.DataFrame(raw_data)
print("البيانات الخام قبل المعالجة:")
print(df)
```

<div dir="rtl">## تمرين عملي: تنظيف البيانات
لاحظوا أن الأسعار تحتوي على نصوص (SAR) وتواريخ بتنسيقات مختلفة. مهمتكم الآن هي تنظيف هذه البيانات لتصبح جاهزة للتحليل.</div>

```
# TODO: قم بكتابة كود لتنظيف عمود السعر (إزالة 'SAR' وتحويله لرقم) وتوحيد تنسيق التاريخ
# df['price'] = ...
# df['date'] = ...

print("البيانات بعد التنظيف:")
print(df)
```

```
# الحل النموذجي للتمرين:
# إزالة 'SAR' وتحويل السعر لرقم
df['price'] = df['price'].replace({' SAR': ''}, regex=True).astype(float)

# توحيد تنسيق التاريخ
df['date'] = pd.to_datetime(df['date'])

print("البيانات بعد التنظيف بنجاح:")
print(df)
print(df.dtypes)
```
