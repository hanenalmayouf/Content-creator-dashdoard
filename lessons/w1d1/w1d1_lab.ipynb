    ## 💡 ما هي الهندسة التوليدية (Generative Engineering)؟

    الهندسة التوليدية هي فرع من فروع الذكاء الاصطناعي والتعلم الآلي يركز على تصميم وتطوير نماذج قادرة على توليد بيانات جديدة تحاكي خصائص وتوزيع البيانات الأصلية التي تدربت عليها. على عكس النماذج التمييزية (Discriminative Models) التي تتعلم التمييز بين الفئات أو التنبؤ بقيم معينة، تتعلم النماذج التوليدية البنية الكامنة للبيانات نفسها.

    من منظور تقني، تسعى النماذج التوليدية إلى تقدير دالة الكثافة الاحتمالية (Probability Density Function) للبيانات التدريبية، مما يمكنها من إنشاء عينات جديدة من هذا التوزيع. هذا يفتح آفاقاً واسعة لتطبيقات تتجاوز مجرد التصنيف والتنبؤ، لتشمل الإبداع والتصميم والمحاكاة.

    ### أمثلة على البيانات التي يمكن توليدها:
    *   صور واقعية لوجوه بشرية غير موجودة.
    *   مقاطع صوتية أو موسيقية جديدة.
    *   نصوص إبداعية أو مقالات.
    *   تصميمات جزيئية أو مواد جديدة بخصائص محددة.
    *   بيانات طبية اصطناعية لأغراض البحث مع الحفاظ على الخصوصية.

    ### 1. الشبكات التنافسية التوليدية (GANs)

    تعتبر GANs من أكثر النماذج التوليدية تأثيراً، وقد قدمها إيان جودفيلو وزملاؤه في عام 2014. تقوم فكرتها على "لعبة" تنافسية بين شبكتين عصبيتين:

    *   **الشبكة المولدة (Generator):** مهمتها توليد بيانات جديدة (مثل الصور) تبدو حقيقية قدر الإمكان.
    *   **الشبكة التمييزية (Discriminator):** مهمتها التمييز بين البيانات الحقيقية (من مجموعة التدريب) والبيانات المزيفة (التي تولدها الشبكة المولدة).

    يتدرب النموذجان في وقت واحد في عملية تنافسية: تحاول الشبكة المولدة خداع الشبكة التمييزية، بينما تحاول الشبكة التمييزية أن تصبح أفضل في اكتشاف البيانات المزيفة. تستمر هذه "اللعبة" حتى تصبح الشبكة المولدة قادرة على إنتاج بيانات لا يمكن للشبكة التمييزية التمييز بينها وبين البيانات الحقيقية.

    #### 📊 البنية الرياضية والخوارزمية (مفهوم مبسط):

    يمكن التعبير عن عملية تدريب GAN كدالة قيمة (Value Function) يسعى المولد إلى تقليلها ويسعى المميز إلى زيادتها (لعبة minimax):

    $$ \min_G \max_D V(D, G) = \mathbb{E}_{x \sim p_{data}(x)}[\log D(x)] + \mathbb{E}_{z \sim p_z(z)}[\log(1 - D(G(z)))] $$

    حيث:
    *   $D(x)$ هو احتمال أن يكون $x$ حقيقياً وفقاً للمميز.
    *   $G(z)$ هو البيانات المولدة من ضوضاء $z$.
    *   $p_{data}(x)$ هو توزيع البيانات الحقيقية.
    *   $p_z(z)$ هو توزيع الضوضاء المدخلة للمولد (عادةً توزيع غاوسي).

    </div>

```
# مثال مبسط لبنية GAN باستخدام PyTorch

# تعريف الشبكة المولدة (Generator)
class Generator(nn.Module):
    def __init__(self, latent_dim, img_shape):
        super(Generator, self).__init__()
        self.img_shape = img_shape
        self.model = nn.Sequential(
            nn.Linear(latent_dim, 128),
            nn.LeakyReLU(0.2, inplace=True),
            nn.Linear(128, 256),
            nn.BatchNorm1d(256),
            nn.LeakyReLU(0.2, inplace=True),
            nn.Linear(256, 512),
            nn.BatchNorm1d(512),
            nn.LeakyReLU(0.2, inplace=True),
            nn.Linear(512, int(np.prod(img_shape))),
            nn.Tanh() # لضمان أن تكون قيم البكسل بين -1 و 1
        )

    def forward(self, z):
        img = self.model(z)
        return img.view(img.size(0), *self.img_shape)

# تعريف الشبكة التمييزية (Discriminator)
class Discriminator(nn.Module):
    def __init__(self, img_shape):
        super(Discriminator, self).__init__()
        self.model = nn.Sequential(
            nn.Linear(int(np.prod(img_shape)), 512),
            nn.LeakyReLU(0.2, inplace=True),
            nn.Linear(512, 256),
            nn.LeakyReLU(0.2, inplace=True),
            nn.Linear(256, 1),
            nn.Sigmoid() # لإخراج احتمال بين 0 و 1
        )

    def forward(self, img):
        img_flat = img.view(img.size(0), -1)
        validity = self.model(img_flat)
        return validity

# تهيئة الأبعاد
latent_dim = 100 # أبعاد متجه الضوضاء
img_shape = (1, 28, 28) # مثال: صورة أحادية اللون 28x28 (مثل MNIST)

# إنشاء النماذج
generator = Generator(latent_dim, img_shape)
discriminator = Discriminator(img_shape)

print("✅ تم تعريف بنية الشبكة المولدة والتمييزية لـ GAN بنجاح.")
print("\nبنية الشبكة المولدة:")
print(generator)
print("\nبنية الشبكة التمييزية:")
print(discriminator)

# مثال على توليد صورة
# z = torch.randn(1, latent_dim) # متجه ضوضاء عشوائي
# generated_img = generator(z)
# print(f"شكل الصورة المولدة: {generated_img.shape}")
# plt.imshow(generated_img.squeeze().detach().numpy(), cmap='gray')
# plt.title("صورة مولدة بواسطة Generator")
# plt.show()
```

    ### 2. الترميز التلقائي التغايري (Variational Autoencoders - VAEs)

    VAEs هي نوع آخر من النماذج التوليدية التي تجمع بين قوة الشبكات العصبية العميقة والمبادئ الإحصائية للاستدلال التغايري (Variational Inference). على عكس GANs التي تتعلم من خلال المنافسة، تتعلم VAEs عن طريق ترميز البيانات إلى فضاء كامن (Latent Space) ثم فك ترميزها لإعادة بناء البيانات الأصلية.

    تتكون VAEs من جزأين رئيسيين:

    *   **المرمز (Encoder):** يأخذ بيانات الإدخال (مثل صورة) ويحولها إلى توزيع احتمالي (عادةً توزيع غاوسي) في الفضاء الكامن. بدلاً من إخراج نقطة واحدة، يخرج المرمز متوسط (mean) وتباين (variance) لهذا التوزيع.
    *   **فاك الترميز (Decoder):** يأخذ عينة من الفضاء الكامن (التي تمثل تمثيلاً مضغوطاً للبيانات) ويعيد بناء البيانات الأصلية منها.

    #### 📊 البنية الرياضية والخوارزمية (مفهوم مبسط):

    الهدف الرئيسي لـ VAE هو تحسين حد أدنى منخفض من الأدلة (Evidence Lower Bound - ELBO)، والذي يمكن التعبير عنه بالمعادلة التالية:

    $$ \mathcal{L}(\theta, \phi; x) = \mathbb{E}_{q_\phi(z|x)}[\log p_\theta(x|z)] - D_{KL}(q_\phi(z|x) || p(z)) $$

    حيث:
    *   $x$ هي بيانات الإدخال.
    *   $z$ هو المتجه الكامن.
    *   $q_\phi(z|x)$ هو التوزيع التقريبي للفضاء الكامن الذي يتعلمه المرمز (Encoder).
    *   $p_\theta(x|z)$ هو التوزيع الذي يتعلمه فاك الترميز (Decoder) لإعادة بناء $x$ من $z$.
    *   $p(z)$ هو التوزيع المسبق (prior) للفضاء الكامن (عادةً توزيع غاوسي قياسي).
    *   $D_{KL}$ هو تباعد Kullback-Leibler، الذي يقيس مدى اختلاف توزيعين. هنا، يضمن أن يكون التوزيع الكامن الذي يتعلمه المرمز قريباً من التوزيع المسبق.

    الحد الأول ($\mathbb{E}_{q_\phi(z|x)}[\log p_\theta(x|z)]$) يمثل **خسارة إعادة البناء (Reconstruction Loss)**، ويسعى إلى جعل البيانات المعاد بناؤها مشابهة للبيانات الأصلية. الحد الثاني ($- D_{KL}(q_\phi(z|x) || p(z))$) يمثل **خسارة التغاير (KL Divergence Loss)**، ويسعى إلى تنظيم الفضاء الكامن ليكون سلساً ومستمراً، مما يسهل توليد عينات جديدة منه.

    </div>

```
# مثال مبسط لبنية VAE باستخدام PyTorch

class VAE(nn.Module):
    def __init__(self, input_dim, hidden_dim, latent_dim):
        super(VAE, self).__init__()

        # المرمز (Encoder)
        self.encoder = nn.Sequential(
            nn.Linear(input_dim, hidden_dim),
            nn.ReLU(),
            nn.Linear(hidden_dim, hidden_dim // 2),
            nn.ReLU()
        )
        self.fc_mu = nn.Linear(hidden_dim // 2, latent_dim) # لمتوسط التوزيع الكامن
        self.fc_logvar = nn.Linear(hidden_dim // 2, latent_dim) # للوغاريتم التباين (لضمان قيم موجبة)

        # فاك الترميز (Decoder)
        self.decoder = nn.Sequential(
            nn.Linear(latent_dim, hidden_dim // 2),
            nn.ReLU(),
            nn.Linear(hidden_dim // 2, hidden_dim),
            nn.ReLU(),
            nn.Linear(hidden_dim, input_dim),
            nn.Sigmoid() # لضمان قيم بكسل بين 0 و 1 للصور
        )

    def encode(self, x):
        h = self.encoder(x)
        mu = self.fc_mu(h)
        logvar = self.fc_logvar(h)
        return mu, logvar

    def reparameterize(self, mu, logvar):
        std = torch.exp(0.5 * logvar) # حساب الانحراف المعياري من لوغاريتم التباين
        eps = torch.randn_like(std) # عينة من توزيع غاوسي قياسي
        return mu + eps * std # إعادة التوزيع (sampling) من التوزيع الكامن

    def decode(self, z):
        return self.decoder(z)

    def forward(self, x):
        mu, logvar = self.encode(x.view(x.size(0), -1))
        z = self.reparameterize(mu, logvar)
        return self.decode(z), mu, logvar

# تهيئة الأبعاد
input_dim = 28 * 28 # مثال: صورة 28x28 مسطحة
hidden_dim = 400
latent_dim = 20 # أبعاد الفضاء الكامن

# إنشاء نموذج VAE
vae = VAE(input_dim, hidden_dim, latent_dim)

print("✅ تم تعريف بنية VAE بنجاح.")
print("\nبنية VAE:")
print(vae)

# مثال على تمرير بيانات وهمية
# dummy_data = torch.randn(64, 1, 28, 28) # 64 صورة 28x28
# reconstructed_data, mu, logvar = vae(dummy_data)
# print(f"شكل البيانات المعاد بناؤها: {reconstructed_data.shape}")
# print(f"شكل المتوسط (mu): {mu.shape}")
# print(f"شكل لوغاريتم التباين (logvar): {logvar.shape}")
```

    ### 3. نماذج الانتشار (Diffusion Models)

    نماذج الانتشار هي أحدث إضافة قوية إلى عائلة النماذج التوليدية، وقد أظهرت قدرات مذهلة في توليد صور عالية الجودة والواقعية. تعتمد هذه النماذج على فكرة محاكاة عملية فيزيائية تُعرف بالانتشار (Diffusion)، حيث يتم إضافة ضوضاء تدريجياً إلى البيانات حتى تتحول إلى ضوضاء عشوائية تماماً، ثم يتم تعلم عملية عكسية لإزالة هذه الضوضاء واستعادة البيانات الأصلية.

    تتكون نماذج الانتشار من عمليتين رئيسيتين:

    *   **العملية الأمامية (Forward Process - Diffusion):** يتم فيها إضافة ضوضاء غاوسية صغيرة تدريجياً إلى البيانات الأصلية على مدى عدد كبير من الخطوات الزمنية. في نهاية هذه العملية، تتحول البيانات الأصلية إلى ضوضاء نقية.
    *   **العملية العكسية (Reverse Process - Denoising):** هذه هي العملية التي يتعلمها النموذج. يحاول النموذج التنبؤ بالضوضاء التي تمت إضافتها في كل خطوة زمنية، ثم يطرحها من البيانات الحالية لإزالة الضوضاء تدريجياً واستعادة البيانات الأصلية. يتم تدريب شبكة عصبية (غالباً ما تكون من نوع U-Net) لتقدير هذه الضوضاء.

    #### 📊 البنية الرياضية والخوارزمية (مفهوم مبسط):

    العملية الأمامية يمكن وصفها بسلسلة ماركوف حيث يتم إضافة ضوضاء غاوسية:

    $$ q(x_t | x_{t-1}) = \mathcal{N}(x_t; \sqrt{1 - \beta_t} x_{t-1}, \beta_t I) $$

    حيث:
    *   $x_t$ هي البيانات عند الخطوة الزمنية $t$.
    *   $\beta_t$ هو جدول الضوضاء (noise schedule) الذي يحدد كمية الضوضاء المضافة في كل خطوة.

    الهدف هو تعلم العملية العكسية $p_\theta(x_{t-1} | x_t)$، والتي يتم تقريبها بواسطة شبكة عصبية (عادةً U-Net) تتنبأ بالضوضاء المضافة. يتم تدريب النموذج لتقليل الفرق بين الضوضاء الفعلية والضوضاء المتنبأ بها:

    $$ \mathcal{L} = \mathbb{E}_{t, x_0, \epsilon \sim \mathcal{N}(0, I)} [\|\epsilon - \epsilon_\theta(\sqrt{\bar{\alpha}_t} x_0 + \sqrt{1 - \bar{\alpha}_t} \epsilon, t)\|^2] $$

    حيث:
    *   $\epsilon$ هي الضوضاء الحقيقية المضافة.
    *   $\epsilon_\theta$ هي الضوضاء المتنبأ بها بواسطة النموذج.
    *   $t$ هي الخطوة الزمنية.
    *   $x_0$ هي البيانات الأصلية.
    *   $\bar{\alpha}_t$ هو معامل يعتمد على جدول الضوضاء.

    </div>

```
# مثال مبسط جداً لمفهوم Diffusion Model (بدون تدريب كامل)
# هذا الكود يوضح فقط بنية شبكة U-Net التي تستخدم عادة في نماذج الانتشار

class UNet(nn.Module):
    def __init__(self, in_channels, out_channels, time_embedding_dim=256):
        super().__init__()

        # تضمين الوقت (Time Embedding) - مهم لنماذج الانتشار
        self.time_mlp = nn.Sequential(
            nn.Linear(1, time_embedding_dim),
            nn.ReLU(),
            nn.Linear(time_embedding_dim, time_embedding_dim)
        )

        # المرمز (Encoder) - لتقليل الأبعاد واستخراج الميزات
        self.enc1 = self._conv_block(in_channels, 64)
        self.enc2 = self._conv_block(64, 128)
        self.pool = nn.MaxPool2d(2)

        # الطبقة الوسطى (Bottleneck)
        self.bottleneck = self._conv_block(128, 256)

        # فاك الترميز (Decoder) - لزيادة الأبعاد وإعادة بناء الصورة
        self.upconv1 = nn.ConvTranspose2d(256, 128, kernel_size=2, stride=2)
        self.dec1 = self._conv_block(256, 128) # 256 = 128 (من upconv) + 128 (من skip connection)
        self.upconv2 = nn.ConvTranspose2d(128, 64, kernel_size=2, stride=2)
        self.dec2 = self._conv_block(128, 64) # 128 = 64 (من upconv) + 64 (من skip connection)

        self.out_conv = nn.Conv2d(64, out_channels, kernel_size=1)

    def _conv_block(self, in_c, out_c):
        return nn.Sequential(
            nn.Conv2d(in_c, out_c, kernel_size=3, padding=1),
            nn.ReLU(),
            nn.Conv2d(out_c, out_c, kernel_size=3, padding=1),
            nn.ReLU()
        )

    def forward(self, x, t):
        # معالجة تضمين الوقت
        t_emb = self.time_mlp(t.unsqueeze(1).float() / 1000.0) # تحويل t إلى متجه
        # في تطبيق حقيقي، يتم دمج t_emb في طبقات الشبكة بطرق مختلفة (مثل FiLM conditioning)
        # هنا، سنبسط الأمر ونفترض أنه يؤثر على الطبقات بطريقة ما (ليس تطبيقاً مباشراً)

        # المرمز
        e1 = self.enc1(x)
        p1 = self.pool(e1)
        e2 = self.enc2(p1)
        p2 = self.pool(e2)

        # الطبقة الوسطى
        b = self.bottleneck(p2)

        # فاك الترميز مع وصلات التخطي (Skip Connections)
        d1 = self.upconv1(b)
        d1 = torch.cat((d1, e2), dim=1) # دمج مع مخرجات المرمز
        d1 = self.dec1(d1)

        d2 = self.upconv2(d1)
        d2 = torch.cat((d2, e1), dim=1) # دمج مع مخرجات المرمز
        d2 = self.dec2(d2)

        output = self.out_conv(d2)
        return output

# تهيئة الأبعاد
in_channels = 1 # صور أحادية اللون
out_channels = 1 # نتوقع إخراج ضوضاء بنفس عدد القنوات

# إنشاء نموذج U-Net
unet_model = UNet(in_channels, out_channels)

print("✅ تم تعريف بنية U-Net (المستخدمة في نماذج الانتشار) بنجاح.")
print("\nبنية U-Net:")
print(unet_model)

# مثال على تمرير بيانات وهمية
# dummy_image = torch.randn(1, 1, 32, 32) # صورة 32x32
# dummy_time = torch.tensor([500]) # خطوة زمنية
# predicted_noise = unet_model(dummy_image, dummy_time)
# print(f"شكل الضوضاء المتنبأ بها: {predicted_noise.shape}")
```

    ## 🚧 التحديات البحثية والفرص المستقبلية

    على الرغم من التقدم الهائل، لا تزال النماذج التوليدية تواجه تحديات كبيرة وتفتح آفاقاً واسعة للبحث والتطوير:

    ### التحديات البحثية الراهنة:

    *   **استقرار التدريب (Training Stability):** خاصة في GANs، يمكن أن يكون التدريب غير مستقر ويؤدي إلى انهيار النمط (Mode Collapse) حيث يولد المولد أنواعاً محدودة فقط من البيانات.
    *   **جودة التوليد والتحكم (Generation Quality and Control):** على الرغم من الجودة العالية، لا يزال التحكم الدقيق في سمات البيانات المولدة يمثل تحدياً (مثال: توليد صورة لوجه بابتسامة معينة وعمر محدد).
    *   **التقييم الكمي (Quantitative Evaluation):** قياس جودة وتنوع البيانات المولدة بشكل موضوعي لا يزال مجال بحث نشط، حيث أن المقاييس الحالية (مثل FID و Inception Score) لها قيودها.
    *   **التحيز (Bias):** يمكن للنماذج التوليدية أن تتعلم وتضخم التحيزات الموجودة في بيانات التدريب، مما يؤدي إلى توليد بيانات متحيزة أو غير عادلة.
    *   **المتطلبات الحاسوبية (Computational Requirements):** تدريب النماذج الكبيرة (خاصة نماذج الانتشار) يتطلب موارد حاسوبية هائلة.

    ### الفرص المستقبلية الواعدة:

    *   **النماذج التوليدية الشرطية (Conditional Generative Models):** تطوير نماذج يمكن التحكم فيها بشكل أفضل لتوليد بيانات بناءً على شروط محددة (مثل توليد صورة لقطة حمراء).
    *   **التعلم متعدد الوسائط (Multimodal Learning):** توليد بيانات تجمع بين أنواع مختلفة (مثل توليد فيديو مع صوت متزامن من وصف نصي).
    *   **النماذج التوليدية الفعالة (Efficient Generative Models):** تطوير نماذج تتطلب موارد حاسوبية أقل للتدريب والتوليد.
    *   **التطبيقات في العلوم والهندسة:** توسيع استخدام النماذج التوليدية في مجالات مثل اكتشاف المواد الجديدة، تصميم الروبوتات، والمحاكاة العلمية.
    *   **النماذج التوليدية الأخلاقية والشفافة (Ethical and Transparent Generative Models):** التركيز على تطوير نماذج يمكن فهمها وتفسيرها بشكل أفضل، ومعالجة قضايا التحيز والخصوصية.
