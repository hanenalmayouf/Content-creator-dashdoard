<div dir="rtl"># معمل تحليل البيانات الاستكشافي (EDA) ومعالجة البيانات المفقودة

أهلاً بك يا بطل البيانات! تخيّل أنك تاجر في سوق قديم، وصلتك بضاعة (بيانات) مختلطة، بعضها تالف، وبعضها مفقود، وبعضها يحتاج لفرز دقيق. مهمتك اليوم هي تنظيف هذه البضاعة وتجهيزها لتكون جاهزة للبيع (للتدريب).</div>

```
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# إنشاء مجموعة بيانات تجريبية تحاكي 'سوق البيانات'
# تخيل أننا نجمع بيانات عن مبيعات التمور
data = {'النوع': ['سكري', 'خلاص', np.nan, 'عجوة', 'سكري'],
        'الوزن_كجم': [10, 15, 12, np.nan, 20],
        'السعر': [50, 75, 60, 40, 100]}

df = pd.DataFrame(data)
print("نظرة على البيانات الخام:")
print(df.head())
```

<div dir="rtl">## القاعدة الأولى: التعامل مع البيانات المفقودة (Imputation)

لا نرمي البضاعة التالفة فوراً! أحياناً يمكننا تقدير القيمة المفقودة بناءً على المتوسط أو الوسيط. هذا يشبه تقدير سعر التمر بناءً على متوسط أسعار الأنواع الأخرى.</div>

```
# ملء القيم المفقودة في الوزن بمتوسط الأوزان
mean_weight = df['الوزن_كجم'].mean()
df['الوزن_كجم'] = df['الوزن_كجم'].fillna(mean_weight)

print("البيانات بعد معالجة القيم المفقودة:")
print(df)
```

<div dir="rtl">## تمرين عملي: تحدي المبرمج

الآن دورك! لدينا عمود 'النوع' يحتوي على قيمة مفقودة. استخدم مكتبة Pandas لملء القيمة المفقودة بأكثر نوع تكراراً (Mode).</div>

```
# TODO: قم بملء القيمة المفقودة في عمود 'النوع' بأكثر قيمة تكراراً (Mode)
# تلميح: استخدم df['النوع'].mode()[0]

# اكتب الكود الخاص بك هنا:
# df['النوع'] = ...
```

```
# الحل النموذجي:
most_frequent = df['النوع'].mode()[0]
df['النوع'] = df['النوع'].fillna(most_frequent)
print("البيانات بعد ملء النوع:")
print(df)
```

<div dir="rtl">## تقليل الأبعاد (PCA) لرفع الكفاءة

عندما يكون لديك الكثير من المتغيرات، يصبح التحليل صعباً. تقليل الأبعاد (PCA) يشبه النظر إلى خريطة الصحراء من الأعلى؛ نختصر التفاصيل الكثيرة في ملامح رئيسية تعبر عن الصورة كاملة.</div>

```
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler

# تحويل البيانات الرقمية فقط للتحليل
features = ['الوزن_كجم', 'السعر']
x = df[features]

# تحجيم البيانات (Standardization)
x_scaled = StandardScaler().fit_transform(x)

# تطبيق PCA لتقليل الأبعاد إلى بُعد واحد
pca = PCA(n_components=1)
principal_components = pca.fit_transform(x_scaled)

print("البيانات بعد تقليل الأبعاد:")
print(principal_components)
```
