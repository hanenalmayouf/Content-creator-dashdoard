<div dir="rtl">## القصة: نظام التوظيف الذكي
تخيل أن شركة كبرى قررت استخدام نظام ذكاء اصطناعي لفرز طلبات التوظيف. النظام تدرب على بيانات الموظفين الناجحين في الشركة خلال الـ 20 سنة الماضية. 

**المشكلة:** النظام بدأ يرفض طلبات النساء بشكل تلقائي! 

**لماذا؟** لأن الشركة تاريخياً كانت توظف الرجال أكثر. النظام "تعلّم" من الماضي أن "الرجل = موظف ناجح"، فصار يكرر هذا التحيز (Bias).</div>

```
# محاكاة بسيطة لنظام توظيف منحاز

# قائمة المتقدمين (الاسم، الخبرة، الجنس)
applicants = [
    {"name": "أحمد", "experience": 5, "gender": "male"},
    {"name": "سارة", "experience": 8, "gender": "female"},
    {"name": "خالد", "experience": 3, "gender": "male"},
    {"name": "نورة", "experience": 7, "gender": "female"}
]

def filter_applicants(candidate_list):
    accepted = []
    for person in candidate_list:
        # النظام هنا منحاز: يرفض النساء مهما كانت خبرتهن!
        if person['gender'] == 'male':
            accepted.append(person['name'])
    return accepted

print("المقبولون حسب النظام المنحاز:", filter_applicants(applicants))
```
