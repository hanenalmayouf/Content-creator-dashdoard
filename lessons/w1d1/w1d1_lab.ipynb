<div dir="rtl"># معمل تحليل البيانات الاستكشافي (EDA) ومعالجة البيانات المفقودة
أهلاً بكم في هذا المعمل التفاعلي. سنقوم اليوم بمحاكاة تحدي Kaggle حقيقي لمعالجة البيانات الملوثة وتحسين جودة النماذج.</div>

```
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler

# إنشاء مجموعة بيانات وهمية ملوثة للتدريب
np.random.seed(42)
data = pd.DataFrame(np.random.randn(100, 5), columns=['A', 'B', 'C', 'D', 'E'])
data.iloc[0:10, 1] = np.nan  # إدخال قيم مفقودة
print("تم تحميل البيانات بنجاح.")
```

<div dir="rtl">## الخطوة الأولى: استكشاف البيانات المفقودة
يجب علينا أولاً تحديد نسبة البيانات المفقودة في كل عمود لاتخاذ قرار بشأن استراتيجية المعالجة (الحذف أو التعويض).</div>

```
# حساب نسبة القيم المفقودة
missing_percentage = data.isnull().mean() * 100
print("نسبة البيانات المفقودة لكل عمود:\n", missing_percentage)
```

<div dir="rtl">## الخطوة الثانية: تمرين عملي (تحدي التعويض)
قم بتعبئة القيم المفقودة في العمود 'B' باستخدام الوسيط (Median) بدلاً من المتوسط الحسابي لتجنب تأثير القيم الشاذة.</div>

```
# TODO: قم بكتابة الكود لتعويض القيم المفقودة في العمود 'B' باستخدام الوسيط
# data['B'] = ...
```

```
# الحل النموذجي:
median_b = data['B'].median()
data['B'].fillna(median_b, inplace=True)
print("تم تعويض القيم المفقودة بنجاح.")
```

<div dir="rtl">## الخطوة الثالثة: تقليل الأبعاد (PCA)
بعد تنظيف البيانات، نستخدم PCA لتقليل عدد المتغيرات مع الحفاظ على التباين (Variance) لرفع كفاءة النموذج.</div>

```
# تحجيم البيانات قبل تطبيق PCA
scaler = StandardScaler()
data_scaled = scaler.fit_transform(data)

# تطبيق PCA
pca = PCA(n_components=2)
pca_result = pca.fit_transform(data_scaled)
print("تم تقليل الأبعاد بنجاح. شكل البيانات الجديد:", pca_result.shape)
```
