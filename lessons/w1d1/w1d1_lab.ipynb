<div dir="rtl"># معمل: مقدمة في هندسة الأوامر (Prompt Engineering)

أهلاً بكم في هذا المعمل التفاعلي. سنتعلم اليوم كيف نتحدث مع الذكاء الاصطناعي بفعالية من خلال هيكلة الأوامر.</div>

<div dir="rtl">## المفهوم الأساسي: هيكلية الأمر (Context - Task - Constraint - Format)

للحصول على أفضل النتائج، يجب أن يحتوي أمرك على:
1. **السياق (Context):** من هو الذكاء الاصطناعي؟ وما هو الموقف؟
2. **المهمة (Task):** ماذا تريد منه أن يفعل بالضبط؟
3. **القيد (Constraint):** ما هي الحدود أو الشروط؟
4. **التنسيق (Format):** كيف تريد شكل المخرجات؟</div>

```
# محاكاة بسيطة لهيكلة الأوامر في بايثون
def construct_prompt(context, task, constraint, format):
    # دمج المكونات لتكوين أمر احترافي
    prompt = f"السياق: {context}\nالمهمة: {task}\nالقيد: {constraint}\nالتنسيق: {format}"
    return prompt

# مثال على أمر مهيكل
my_prompt = construct_prompt(
    "أنت خبير تسويق رقمي",
    "اكتب وصفاً لمنتج قهوة جديد",
    "لا تزيد عن 50 كلمة",
    "قائمة نقطية"
)
print("الأمر المهيكل الناتج:\n", my_prompt)
```

<div dir="rtl">## المشروع التطبيقي (Project Challenge)

مهمتك هي بناء 'مولد أوامر ذكي' يقوم بطلب مدخلات من المستخدم ثم يخرج أمراً مهيكلاً جاهزاً للنسخ واللصق في ChatGPT. 

**المطلوب:**
1. استخدم دالة `input()` لجمع البيانات من المستخدم.
2. قم بتمريرها لدالة `construct_prompt`.
3. اطبع النتيجة النهائية للمستخدم.</div>

```
# هيكل المشروع التطبيقي
print("--- مولد الأوامر الذكي ---")
user_context = input("أدخل السياق (من هو الذكاء الاصطناعي؟): ")
user_task = input("أدخل المهمة المطلوبة: ")
user_constraint = input("أدخل القيود: ")
user_format = input("أدخل التنسيق المطلوب: ")

final_prompt = construct_prompt(user_context, user_task, user_constraint, user_format)
print("\nإليك الأمر الجاهز للاستخدام:\n", final_prompt)
```
