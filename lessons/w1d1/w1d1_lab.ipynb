<div dir="rtl">## النشاط العملي 2: بناء قصة البيانات 🗺️

**الهدف:** تتبع رحلة قطعة بيانات من لحظة إنشائها حتى استخدامها.

**كيفية النشاط:**
في مجموعات صغيرة، ستختارون "قطعة بيانات" معينة وتتبعون رحلتها. على سبيل المثال:

*   **سيناريو 1:** طلب شراء من متجر إلكتروني.
*   **سيناريو 2:** قراءة مستشعر في هاتف ذكي (مثل مستشعر الحركة أو الموقع).

**الخطوات:**
1.  **الإنشاء:** كيف ومتى تُنشأ هذه البيانات؟
2.  **الجمع:** كيف تُجمع هذه البيانات؟ (مثال: نموذج ويب، مستشعر).
3.  **التخزين:** أين تُخزن؟ (مثال: قاعدة بيانات، ملف).
4.  **المعالجة:** هل تحتاج إلى تنظيف أو تحويل؟ كيف؟
5.  **الاستخدام:** كيف تُستخدم هذه البيانات لتحقيق هدف معين؟ (مثال: إرسال تأكيد طلب، تحليل سلوك المستخدم).

كل مجموعة ستقدم قصتها المصورة أو المكتوبة، مما يعزز الفهم التعاوني لدورة حياة البيانات.</div>

<div dir="rtl">## التقييم التكويني المستمر: تحدي فكر سريع 🧠

**الهدف:** تقييم فهمكم للمفاهيم الأساسية وتقديم تغذية راجعة فورية.

**تحدي فكر سريع (Think-Pair-Share):**

1.  **فكر (Think):** ما هو الفرق بين البيانات النصية والبيانات الرقمية؟ أعطِ مثالاً لكل منهما.
2.  **شارك (Pair):** ناقش إجابتك مع زميلك.

**سؤال إضافي:**
إذا كان لديك صورة لمنتج، ما هو نوع البيانات التي تمثلها؟ وكيف يمكن أن تُستخدم هذه البيانات في متجر إلكتروني؟

**ملاحظة:** سيتم استخدام استطلاع رأي سريع في نهاية اليوم حول أكثر مفهوم أثار اهتمامكم أو وجدتم صعوبة فيه لضبط المحتوى المستقبلي.</div>

<div dir="rtl">## ما هي هندسة البيانات؟ 🛠️

**هندسة البيانات (Data Engineering)** هي تخصص يركز على تصميم وبناء وصيانة الأنظمة والبنية التحتية التي تسمح بجمع وتخزين ومعالجة البيانات على نطاق واسع. فكر فيها كبناء الطرق والجسور التي تسمح للبيانات بالتدفق بسلاسة من نقطة إلى أخرى.

**دور مهندس البيانات:**
مهندس البيانات هو المهندس المعماري والبناء لعالم البيانات. تشمل مهامه الرئيسية:

*   **تصميم خطوط أنابيب البيانات (Data Pipelines):** إنشاء أنظمة لجمع البيانات من مصادر مختلفة (مثل قواعد البيانات، واجهات برمجة التطبيقات، الملفات).
*   **تنظيف وتحويل البيانات (Data Cleaning & Transformation):** التأكد من أن البيانات دقيقة ومتسقة وجاهزة للتحليل.
*   **بناء مستودعات البيانات (Data Warehouses) وبحيرات البيانات (Data Lakes):** تصميم حلول تخزين فعالة للبيانات.
*   **تحسين الأداء (Performance Optimization):** التأكد من أن أنظمة البيانات تعمل بكفاءة وسرعة.

باختصار، مهندس البيانات يجعل البيانات متاحة وموثوقة وقابلة للاستخدام لعلماء البيانات ومحللي البيانات والتطبيقات.</div>

<div dir="rtl">## أهمية هندسة البيانات في عالمنا الحديث 🌍

تخيل عالماً لا تستطيع فيه البيانات أن تتدفق أو تُفهم. هذا هو العالم بدون هندسة البيانات! إليك لماذا هي مهمة للغاية:

1.  **دعم اتخاذ القرارات:** توفر هندسة البيانات بيانات نظيفة وموثوقة للمديرين والمحللين لاتخاذ قرارات مستنيرة، سواء كانت قرارات عمل أو قرارات حكومية.
2.  **تطوير التطبيقات اليومية:**
    *   **أنظمة التوصية:** هل تساءلت يوماً كيف يقترح عليك Netflix أفلاماً أو Amazon منتجات؟ هذا بفضل هندسة البيانات التي تجمع وتحلل سلوكك وتفضيلاتك.
    *   **السيارات ذاتية القيادة:** تعتمد على تدفق مستمر للبيانات من المستشعرات (الكاميرات، الرادارات) التي يجب معالجتها بسرعة فائقة لاتخاذ قرارات القيادة.
    *   **تطبيقات الصحة واللياقة البدنية:** تجمع بيانات عن نشاطك البدني، نومك، ومعدل ضربات قلبك، ثم تعالجها لتقديم رؤى صحية.
3.  **الذكاء الاصطناعي وتعلم الآلة:** هذه التقنيات تحتاج إلى كميات هائلة من البيانات عالية الجودة للتدريب. مهندسو البيانات هم من يجهزون هذه البيانات.
4.  **تحليل البيانات الضخمة (Big Data Analytics):** مع تزايد حجم البيانات، أصبحت هندسة البيانات ضرورية لإدارة ومعالجة هذه الكميات الهائلة بكفاءة.

بدون هندسة البيانات، ستكون البيانات مجرد فوضى غير منظمة، ولن نتمكن من استخلاص القيمة الحقيقية منها.</div>

<div dir="rtl">## تمرين تفاعلي: تحويل البيانات 🔄

**الهدف:** تطبيق مفهوم بسيط لتحويل البيانات باستخدام بايثون.

لنفترض أن لدينا قائمة بأسعار المنتجات كـ "نصوص" (strings)، ونريد تحويلها إلى أرقام عشرية (floats) لحساب متوسط السعر أو إجراء عمليات حسابية أخرى.</div>

````
# <div dir="rtl">تمرين: تحويل أسعار المنتجات</div>

# <div dir="rtl">لدينا قائمة بأسعار المنتجات كـ "نصوص".</div>
اسعار_المنتجات_النصية = ["120.50", "75.00", "299.99", "50.25", "150.00"]

# <div dir="rtl">المطلوب:</div>
# <div dir="rtl">1. قم بإنشاء قائمة جديدة تسمى `اسعار_المنتجات_الرقمية`.</div>
# <div dir="rtl">2. استخدم حلقة تكرارية (loop) لتحويل كل سعر من نص إلى رقم عشري (float).</div>
# <div dir="rtl">3. أضف الأسعار المحولة إلى القائمة الجديدة.</div>
# <div dir="rtl">4. اطبع القائمة الجديدة ونوع أول عنصر فيها للتأكد من التحويل الصحيح.</div>

اسعار_المنتجات_الرقمية = []

# TODO: <div dir="rtl">اكتب الكود الخاص بك هنا لتحويل الأسعار</div>
# <div dir="rtl">تلميح: استخدم دالة `float()` للتحويل.</div>

# <div dir="rtl">بعد الانتهاء، يجب أن يكون الناتج مشابهاً لما يلي:</div>
# <div dir="rtl">القائمة المحولة: [120.5, 75.0, 299.99, 50.25, 150.0]</div>
# <div dir="rtl">نوع أول عنصر: <class 'float'></div>
````

````
# <div dir="rtl">الحل النموذجي لتمرين تحويل البيانات</div>

اسعار_المنتجات_النصية = ["120.50", "75.00", "299.99", "50.25", "150.00"]
اسعار_المنتجات_الرقمية_الحل = []

for سعر_نصي in اسعار_المنتجات_النصية:
    سعر_رقمي = float(سعر_نصي)
    اسعار_المنتجات_الرقمية_الحل.append(سعر_رقمي)

print(f"<div dir=\"rtl\">القائمة المحولة: {اسعار_المنتجات_الرقمية_الحل}</div>")
if اسعار_المنتجات_الرقمية_الحل:
    print(f"<div dir=\"rtl\">نوع أول عنصر: {type(اسعار_المنتجات_الرقمية_الحل[0])}</div>")
else:
    print("<div dir=\"rtl\">القائمة فارغة، لم يتم التحويل بشكل صحيح.</div>")

# <div dir="rtl">يمكننا الآن إجراء عمليات حسابية، مثل حساب متوسط السعر:</div>
متوسط_السعر = sum(اسعار_المنتجات_الرقمية_الحل) / len(اسعار_المنتجات_الرقمية_الحل)
print(f"<div dir=\"rtl\">متوسط سعر المنتجات: {متوسط_السعر:.2f}</div>")
````

<div dir="rtl">## المشروع التطبيقي: تصميم مسار بيانات بسيط 🚀

**الهدف:** تطبيق المفاهيم الأساسية لهندسة البيانات لتصميم مسار بيانات مبسط.

**السيناريو:**
تخيل أنك مهندس بيانات في شركة تجارة إلكترونية صغيرة. مهمتك هي تصميم مسار بيانات مبسط لجمع ومعالجة بيانات "تقييمات المنتجات" التي يتركها العملاء.

**المطلوب (في شكل نقاط أو رسم تخطيطي):**

1.  **مصدر البيانات:** من أين ستأتي تقييمات المنتجات؟ (مثال: نموذج ويب، تطبيق جوال).
2.  **جمع البيانات:** كيف ستجمع هذه التقييمات؟ (مثال: API، قاعدة بيانات).
3.  **تخزين البيانات الأولية:** أين ستخزن التقييمات فور جمعها؟ (مثال: قاعدة بيانات مؤقتة، ملفات نصية).
4.  **خطوات المعالجة:** ما هي خطوات المعالجة التي قد تحتاجها هذه التقييمات قبل أن تصبح جاهزة للتحليل؟ (مثال: تنظيف النصوص من الرموز غير المرغوبة، تحويل التقييمات الرقمية، استخلاص المشاعر).
5.  **تخزين البيانات المعالجة:** أين ستخزن التقييمات بعد معالجتها لتكون جاهزة للتحليل؟ (مثال: مستودع بيانات).
6.  **الاستخدام النهائي:** كيف يمكن استخدام هذه البيانات المعالجة؟ (مثال: تحليل رضا العملاء، تحسين المنتجات، بناء نظام توصية).

**تحدي إضافي (اختياري):**
إذا كان لديك تقييمات نصية، كيف يمكنك استخدام بايثون لحساب طول كل تقييم؟ (فكر في دالة `len()`).
