<div dir="rtl">## القسم الأول: دور مهندس البيانات في منظومة الذكاء الاصطناعي

مهندس البيانات هو المعماري والباني للبنية التحتية التي تمكن علماء البيانات ومهندسي تعلم الآلة من الوصول إلى البيانات النظيفة والمنظمة والقابلة للتوسع. بدون هندسة بيانات قوية، لا يمكن لنماذج الذكاء الاصطناعي أن تعمل بفعالية أو على نطاق واسع.

### المهام الرئيسية:

*   تصميم وبناء وصيانة خطوط أنابيب البيانات (Data Pipelines).
*   تأمين جودة البيانات ونظافتها.
*   إدارة تخزين البيانات الموزعة.
*   تحسين أداء أنظمة البيانات.
*   التعاون مع فرق علم البيانات والتعلم الآلي لفهم احتياجاتهم من البيانات.
</div>

```
# مثال توضيحي: محاكاة بسيطة لدور مهندس البيانات في تحضير البيانات

import pandas as pd
import numpy as np

# 1. استيعاب البيانات الأولية (محاكاة قراءة من مصدر ما)
# لنفترض أن لدينا بيانات مبيعات خام تحتوي على بعض المشاكل
data = {
    'product_id': [1, 2, 3, 1, 2, 4, 5, 3, 1, 6],
    'customer_id': ['A', 'B', 'C', 'A', 'D', 'E', 'F', 'C', 'G', 'H'],
    'price': [10.5, 20.0, 5.2, 10.5, 20.0, 15.0, 25.0, 5.2, 'غير متوفر', 30.0],
    'quantity': [1, 2, 3, 1, 2, 1, 4, 3, 0, 2],
    'timestamp': pd.to_datetime([
        '2023-01-01 10:00:00', '2023-01-01 10:05:00', '2023-01-01 10:10:00',
        '2023-01-02 11:00:00', '2023-01-02 11:05:00', '2023-01-03 12:00:00',
        '2023-01-03 12:05:00', '2023-01-04 13:00:00', '2023-01-04 13:05:00',
        '2023-01-05 14:00:00'
    ])
}
raw_df = pd.DataFrame(data)
print("\nالبيانات الأولية:\n", raw_df)

# 2. تنظيف البيانات (معالجة القيم المفقودة أو غير الصحيحة)
# تحويل عمود 'price' إلى رقمي، مع تحويل القيم غير الصالحة إلى NaN
raw_df['price'] = pd.to_numeric(raw_df['price'], errors='coerce')

# ملء القيم المفقودة في 'price' بمتوسط السعر (أو استراتيجية أخرى)
raw_df['price'].fillna(raw_df['price'].mean(), inplace=True)

# إزالة الصفوف التي تحتوي على كمية (quantity) صفر أو سالبة (بيانات غير منطقية)
cleaned_df = raw_df[raw_df['quantity'] > 0].copy()

print("\nالبيانات بعد التنظيف:\n", cleaned_df)

# 3. تحويل البيانات (إنشاء ميزات جديدة)
cleaned_df['total_sales'] = cleaned_df['price'] * cleaned_df['quantity']

# 4. تجميع البيانات (لتحضيرها للتحليل أو نموذج الذكاء الاص0طناعي)
# حساب إجمالي المبيعات لكل منتج
product_sales = cleaned_df.groupby('product_id')['total_sales'].sum().reset_index()
print("\nإجمالي المبيعات لكل منتج:\n", product_sales)

# هذه البيانات النظيفة والمنظمة جاهزة الآن ليستخدمها عالم البيانات لبناء نموذج توصية أو تحليل سلوك العملاء.
```

<div dir="rtl">## القسم الثاني: أنماط استيعاب ومعالجة البيانات الضخمة (Batch و Streaming)

تعتبر كيفية استيعاب ومعالجة البيانات من أهم القرارات في هندسة البيانات، وتعتمد على طبيعة البيانات ومتطلبات العمل. هناك نمطان رئيسيان: معالجة الدفعات (Batch Processing) ومعالجة التدفق (Streaming Processing).

### 2.1 معالجة الدفعات (Batch Processing)

تتضمن معالجة الدفعات جمع البيانات على مدى فترة زمنية محددة (ساعات، أيام، أسابيع) ثم معالجتها دفعة واحدة. تُستخدم عادةً للبيانات التي لا تتطلب استجابة فورية، مثل التقارير اليومية، تحليلات نهاية الشهر، أو تدريب نماذج تعلم الآلة.

**المميزات:**
*   فعالة من حيث التكلفة للمجموعات الكبيرة من البيانات.
*   أسهل في التصميم والتنفيذ.
*   تسمح بمعالجة معقدة للبيانات.

**العيوب:**
*   تأخير في توفر البيانات (Latency).
*   غير مناسبة للتطبيقات التي تتطلب استجابة فورية.

**أمثلة على الأدوات:** Apache Hadoop, Apache Spark (في وضع الدفعات), SQL Databases.
</div>

```
# مثال على معالجة الدفعات باستخدام Pandas (محاكاة لعملية Spark Batch)

# لنفترض أن لدينا ملف CSV كبير يمثل بيانات المبيعات لشهر كامل
# سنقوم بإنشاء ملف وهمي لهذا الغرض
import os

# إنشاء بيانات وهمية لملف CSV
data_batch = {
    'order_id': range(1, 10001),
    'product_category': np.random.choice(['Electronics', 'Books', 'Clothing', 'Food'], 10000),
    'amount': np.random.uniform(10, 500, 10000),
    'order_date': pd.to_datetime('2023-01-01') + pd.to_timedelta(np.random.randint(0, 30, 10000), unit='D')
}
batch_df = pd.DataFrame(data_batch)
batch_file_path = 'monthly_sales.csv'
batch_df.to_csv(batch_file_path, index=False)

print(f"تم إنشاء ملف بيانات دفعات وهمي: {batch_file_path}")

# محاكاة عملية معالجة الدفعات:
# 1. قراءة البيانات من الملف (عادة ما يتم ذلك مرة واحدة في اليوم/الأسبوع)
print("\nبدء معالجة الدفعات...")
monthly_sales_df = pd.read_csv(batch_file_path)
monthly_sales_df['order_date'] = pd.to_datetime(monthly_sales_df['order_date'])

# 2. تنظيف وتحويل البيانات (مثلاً، حساب إجمالي المبيعات لكل فئة)
category_sales = monthly_sales_df.groupby('product_category')['amount'].sum().reset_index()
category_sales.rename(columns={'amount': 'total_revenue'}, inplace=True)

# 3. تخزين النتائج (مثلاً، في قاعدة بيانات أو ملف آخر للتقارير)
output_file_path = 'monthly_category_revenue.csv'
category_sales.to_csv(output_file_path, index=False)

print("\nنتائج معالجة الدفعات (إجمالي الإيرادات لكل فئة):\n", category_sales)
print(f"تم حفظ النتائج في: {output_file_path}")

# تنظيف الملفات الوهمية
os.remove(batch_file_path)
os.remove(output_file_path)
print("\nتم تنظيف الملفات الوهمية.")
```

<div dir="rtl">### 2.2 معالجة التدفق (Streaming Processing)

تتضمن معالجة التدفق معالجة البيانات فور وصولها، في الوقت الفعلي أو شبه الفعلي. تُستخدم للتطبيقات التي تتطلب استجابة فورية، مثل مراقبة الاحتيال، تحديث لوحات المعلومات في الوقت الفعلي، أو أنظمة التوصية الفورية.

**المميزات:**
*   استجابة فورية للبيانات الجديدة.
*   مثالية للتطبيقات الحساسة للوقت.
*   يمكنها التعامل مع كميات هائلة من البيانات المتدفقة باستمرار.

**العيوب:**
*   أكثر تعقيدًا في التصميم والتنفيذ.
*   تتطلب بنية تحتية قوية وموثوقة.
*   قد تكون أكثر تكلفة.

**أمثلة على الأدوات:** Apache Kafka, Apache Flink, Apache Spark Streaming, Amazon Kinesis.
</div>

```
# مثال على معالجة التدفق (محاكاة بسيطة باستخدام Python)

import time
import random
from collections import deque

# محاكاة مصدر بيانات متدفق (مثل مستشعرات IoT أو نقرات المستخدمين)
def generate_stream_data():
    product_ids = [101, 102, 103, 104, 105]
    user_ids = ['user_A', 'user_B', 'user_C', 'user_D']
    while True:
        data_point = {
            'timestamp': time.time(),
            'user_id': random.choice(user_ids),
            'product_id': random.choice(product_ids),
            'event_type': random.choice(['view', 'add_to_cart', 'purchase']),
            'value': round(random.uniform(1.0, 100.0), 2) if random.random() > 0.5 else None # قيمة اختيارية
        }
        yield data_point
        time.sleep(random.uniform(0.1, 0.5)) # محاكاة وصول البيانات بشكل غير منتظم

# محاكاة معالج تدفق بسيط:
# يقوم بتجميع آخر 10 أحداث 'purchase' وحساب متوسط قيمتها

def stream_processor():
    purchase_events = deque(maxlen=10) # قائمة دائرية لتخزين آخر 10 أحداث شراء
    data_generator = generate_stream_data()
    
    print("\nبدء معالج التدفق (سيتم معالجة 20 حدثًا)...")
    for i, event in enumerate(data_generator):
        if i >= 20: # معالجة 20 حدثًا فقط لأغراض العرض
            break
            
        print(f"\nحدث جديد: {event}")
        
        if event['event_type'] == 'purchase' and event['value'] is not None:
            purchase_events.append(event['value'])
            print(f"  -> تم إضافة حدث شراء بقيمة {event['value']} إلى المخزن المؤقت.")
            
            if len(purchase_events) > 0:
                avg_purchase_value = sum(purchase_events) / len(purchase_events)
                print(f"  -> متوسط قيمة آخر {len(purchase_events)} عملية شراء: {avg_purchase_value:.2f}")
        else:
            print(f"  -> حدث من نوع '{event['event_type']}' أو قيمة غير صالحة، تم تجاهله للمعالجة الحالية.")
            
        time.sleep(0.1) # محاكاة وقت المعالجة

# تشغيل معالج التدفق
stream_processor()
print("\nانتهت محاكاة معالج التدفق.")
```

<div dir="rtl">## القسم الثالث: تصميم بنى تخزين البيانات الموزعة

تعتبر بنى تخزين البيانات الموزعة أساسية للتعامل مع البيانات الضخمة. سنتناول ثلاثة مفاهيم رئيسية: بحيرات البيانات (Data Lakes)، مستودعات البيانات (Data Warehouses)، ومنازل البحيرة (Lakehouses).

### 3.1 بحيرات البيانات (Data Lakes)

بحيرة البيانات هي مستودع مركزي لتخزين كميات هائلة من البيانات الأولية (Raw Data) بتنسيقها الأصلي، بما في ذلك البيانات المهيكلة وغير المهيكلة وشبه المهيكلة. الهدف هو تخزين كل شيء أولاً ثم تحديد كيفية استخدامه لاحقًا.

**المميزات:**
*   مرونة عالية في تخزين أي نوع من البيانات.
*   تكلفة تخزين منخفضة (عادةً على تخزين الكائنات مثل S3 أو HDFS).
*   تدعم حالات استخدام متعددة مثل تعلم الآلة والتحليلات الاستكشافية.

**العيوب:**
*   قد تتحول إلى 'مستنقع بيانات' (Data Swamp) إذا لم تتم إدارتها بشكل صحيح.
*   تتطلب مهارات عالية لاستخراج القيمة منها.
*   تحديات في جودة البيانات وإدارة المخططات.

**أمثلة على الأدوات:** Apache HDFS, Amazon S3, Azure Data Lake Storage, Google Cloud Storage.
</div>

<div dir="rtl">### 3.2 مستودعات البيانات (Data Warehouses)

مستودع البيانات هو نظام تخزين مصمم خصيصًا للتحليلات وإعداد التقارير. يتم فيه تخزين البيانات المهيكلة والمنظمة بعد تنظيفها وتحويلها من مصادر متعددة، وتكون مُحسّنة للاستعلامات المعقدة.

**المميزات:**
*   أداء عالٍ للاستعلامات التحليلية.
*   جودة بيانات عالية ومخططات محددة جيدًا.
*   سهولة الاستخدام لأدوات BI والتقارير.

**العيوب:**
*   أقل مرونة في التعامل مع البيانات غير المهيكلة.
*   تكلفة أعلى مقارنة ببحيرات البيانات.
*   تتطلب عملية ETL (استخراج، تحويل، تحميل) معقدة.

**أمثلة على الأدوات:** Amazon Redshift, Google BigQuery, Snowflake, Teradata.
</div>

<div dir="rtl">### 3.3 منازل البحيرة (Lakehouses)

منزل البحيرة هو بنية جديدة تجمع بين أفضل ميزات بحيرات البيانات ومستودعات البيانات. يوفر مرونة بحيرة البيانات في تخزين البيانات الأولية، مع توفير هياكل البيانات وإدارة المخططات وميزات الأداء لمستودع البيانات، غالبًا باستخدام تنسيقات ملفات مفتوحة مثل Delta Lake أو Apache Iceberg أو Apache Hudi.

**المميزات:**
*   مرونة تخزين البيانات الأولية وغير المهيكلة.
*   دعم المعاملات (ACID transactions) على بحيرة البيانات.
*   أداء استعلامات عالي.
*   تبسيط بنية البيانات وتقليل الازدواجية.

**العيوب:**
*   تقنية حديثة نسبيًا وقد تتطلب خبرة متخصصة.
*   قد تكون أكثر تعقيدًا في الإعداد الأولي.

**أمثلة على الأدوات:** Databricks Lakehouse Platform (Delta Lake), Apache Iceberg, Apache Hudi.
</div>

```
# مثال توضيحي: محاكاة التفاعل مع بحيرة بيانات (قراءة أنواع مختلفة من الملفات)

# سنقوم بإنشاء بعض الملفات الوهمية لمحاكاة بحيرة بيانات
import os
import json
import pandas as pd

# إنشاء مجلد مؤقت للبيانات
if not os.path.exists('data_lake'):
    os.makedirs('data_lake')

# إنشاء ملف CSV
csv_data = "id,name,value\n1,itemA,100\n2,itemB,200"
with open('data_lake/structured_data.csv', 'w') as f:
    f.write(csv_data)

# إنشاء ملف JSON
json_data = [
    {"id": 3, "description": "eventC", "details": {"status": "success"}},
    {"id": 4, "description": "eventD", "details": {"status": "failed"}}
]
with open('data_lake/semi_structured_data.json', 'w') as f:
    json.dump(json_data, f)

# إنشاء ملف نصي (غير مهيكل)
text_data = "هذا نص غير مهيكل يمثل سجلات خادم أو بيانات نصية حرة."
with open('data_lake/unstructured_data.txt', 'w') as f:
    f.write(text_data)

print("تم إنشاء ملفات وهمية في 'data_lake/'")

# محاكاة قراءة البيانات من بحيرة البيانات
# في بيئة حقيقية، قد تستخدم Spark لقراءة هذه الملفات

# قراءة ملف CSV (بيانات مهيكلة)
print("\nقراءة بيانات CSV من بحيرة البيانات:")
csv_df = pd.read_csv('data_lake/structured_data.csv')
print(csv_df)

# قراءة ملف JSON (بيانات شبه مهيكلة)
print("\nقراءة بيانات JSON من بحيرة البيانات:")
json_df = pd.read_json('data_lake/semi_structured_data.json')
print(json_df)

# قراءة ملف نصي (بيانات غير مهيكلة)
print("\nقراءة بيانات نصية من بحيرة البيانات:")
with open('data_lake/unstructured_data.txt', 'r') as f:
    text_content = f.read()
print(text_content)

# تنظيف الملفات الوهمية والمجلد
os.remove('data_lake/structured_data.csv')
os.remove('data_lake/semi_structured_data.json')
os.remove('data_lake/unstructured_data.txt')
os.rmdir('data_lake')
print("\nتم تنظيف الملفات الوهمية والمجلد.")
```

### 4.1 Apache Spark

Apache Spark هو محرك تحليل موحد للبيانات الضخمة ومعالجة تعلم الآلة. يوفر واجهات برمجة تطبيقات عالية المستوى بلغات مثل Python (PySpark)، Java، Scala، و R. يمكنه معالجة البيانات في وضع الدفعات (Batch) والتدفق (Streaming) والرسوم البيانية (Graph Processing) والتعلم الآلي (Machine Learning).

**المميزات:**

```
# مثال على استخدام PySpark (يتطلب تثبيت PySpark)
# pip install pyspark

from pyspark.sql import SparkSession
from pyspark.sql.functions import col, sum

# إنشاء جلسة Spark
spark = SparkSession.builder \
    .appName("SparkDataProcessing") \
    .getOrCreate()

print("تم إنشاء جلسة Spark بنجاح.")

# إنشاء DataFrame بسيط (محاكاة لبيانات من مصدر كبير)
data = [
    ("productA", 100, 1),
    ("productB", 200, 2),
    ("productA", 150, 1),
    ("productC", 50, 3),
    ("productB", 220, 1)
]
columns = ["product_name", "price", "quantity"]
df = spark.createDataFrame(data, columns)

print("\nالبيانات الأصلية (Spark DataFrame):")
df.show()

# تحويل البيانات: حساب إجمالي الإيرادات لكل منتج
# 1. إضافة عمود 'total_revenue'
df_with_revenue = df.withColumn("total_revenue", col("price") * col("quantity"))

print("\nالبيانات مع عمود الإيرادات الكلية:")
df_with_revenue.show()

# 2. تجميع حسب اسم المنتج وحساب مجموع الإيرادات
product_summary = df_with_revenue.groupBy("product_name").agg(sum("total_revenue").alias("total_sales"))

print("\nملخص المبيعات لكل منتج:")
product_summary.show()

# إيقاف جلسة Spark
spark.stop()
print("تم إيقاف جلسة Spark.")
```

<div dir="rtl">### 4.2 Apache Kafka

Apache Kafka هو منصة توزيعية لتدفق الأحداث (Distributed Streaming Platform) مصممة للتعامل مع كميات كبيرة من البيانات المتدفقة في الوقت الفعلي. يُستخدم لنشر واشتراك وتخزين ومعالجة تدفقات الأحداث.

**المميزات:**
*   قابلية عالية للتوسع والأداء.
*   تحمل الأخطاء (Fault-tolerant).
*   معالجة الرسائل في الوقت الفعلي.
*   دعم إعادة تشغيل البيانات (Replayability).

**الاستخدامات الشائعة:** بناء خطوط أنابيب بيانات في الوقت الفعلي، تتبع نشاط المستخدم، سجلات التشغيل الموزعة، معالجة تدفق الأحداث.
</div>

```
# مثال على استخدام Kafka (مفاهيمي - يتطلب خادم Kafka فعليًا للتشغيل الكامل)
# pip install kafka-python (إذا كنت تريد تشغيل مثال حقيقي)
import time

# محاكاة منتج (Producer) يرسل رسائل إلى Kafka topic
def kafka_producer_mock(topic, messages):
    print(f"\nمحاكاة منتج Kafka لـ topic: {topic}")
    for i, msg in enumerate(messages):
        print(f"  -> إرسال رسالة {i+1}: {msg}")
        time.sleep(0.5) # محاكاة تأخير الإرسال
    print(f"تم إرسال جميع الرسائل إلى {topic}.")

# محاكاة مستهلك (Consumer) يستقبل رسائل من Kafka topic
def kafka_consumer_mock(topic, num_messages_to_consume, producer_messages_ref):
    print(f"\nمحاكاة مستهلك Kafka لـ topic: {topic}")
    received_messages = []
    for i in range(num_messages_to_consume):
        # في بيئة حقيقية، سيتم استدعاء consumer.poll() أو consumer.next()
        # هنا، سنفترض أننا نستقبل الرسائل التي تم إرسالها للتو
        if i < len(producer_messages_ref):
            msg = producer_messages_ref[i]
            received_messages.append(msg)
            print(f"  <- استلام رسالة {i+1}: {msg}")
        else:
            print("  <- لا توجد رسائل أخرى في الوقت الحالي.")
        time.sleep(0.7) # محاكاة تأخير الاستهلاك
    print(f"تم استلام {len(received_messages)} رسالة من {topic}.")
    return received_messages

# بيانات الرسائل التي سيرسلها المنتج
producer_messages = [
    "User A viewed product 101",
    "User B added product 102 to cart",
    "User A purchased product 101",
    "User C viewed product 103",
    "User B purchased product 102"
]

kafka_topic = "user_events"

# تشغيل المحاكاة
kafka_producer_mock(kafka_topic, producer_messages)
consumed_data = kafka_consumer_mock(kafka_topic, len(producer_messages), producer_messages)

print("\nالرسائل التي تم استهلاكها:\n", consumed_data)
```

<div dir="rtl">### 4.3 Apache Airflow

Apache Airflow هو منصة لإنشاء وجدولة ومراقبة خطوط أنابيب البيانات برمجياً (Programmatically Author, Schedule, and Monitor Workflows). يُستخدم لأتمتة مهام هندسة البيانات المعقدة، مثل استخراج البيانات، تحويلها، تحميلها، وتدريب نماذج تعلم الآلة.

**المميزات:**
*   تعريف سير العمل ككود (DAGs - Directed Acyclic Graphs).
*   واجهة مستخدم رسومية قوية للمراقبة.
*   قابلية التوسع والمرونة.
*   دعم واسع للموصلات (Connectors) للعديد من الأنظمة.

**الاستخدامات الشائعة:** ETL/ELT، أتمتة التقارير، تدريب نماذج تعلم الآلة، إدارة المهام الدورية.
</div>

```
# مثال على هيكل DAG في Apache Airflow (مفاهيمي - يتطلب بيئة Airflow للتشغيل)

from airflow import DAG
from airflow.operators.bash import BashOperator
from airflow.operators.python import PythonOperator
from datetime import datetime, timedelta
import time # لإضافة تأخيرات في الوظائف الوهمية

# تعريف الوظائف التي سيتم استدعاؤها بواسطة PythonOperator
def extract_data():
    print("استخراج البيانات من المصدر...")
    # هنا يمكن إضافة كود لقراءة البيانات من قاعدة بيانات أو API
    time.sleep(2)
    print("تم استخراج البيانات.")

def transform_data():
    print("تحويل البيانات وتنظيفها...")
    # هنا يمكن إضافة كود لمعالجة البيانات باستخدام Pandas أو Spark
    time.sleep(3)
    print("تم تحويل البيانات.")

def load_data():
    print("تحميل البيانات إلى مستودع البيانات...")
    # هنا يمكن إضافة كود لكتابة البيانات إلى Redshift أو Snowflake
    time.sleep(2)
    print("تم تحميل البيانات.")

def train_ml_model():
    print("تدريب نموذج تعلم الآلة...")
    # هنا يمكن إضافة كود لتدريب نموذج باستخدام scikit-learn أو TensorFlow
    time.sleep(5)
    print("تم تدريب النموذج.")

# تعريف الوسائط الافتراضية للـ DAG
default_args = {
    'owner': 'airflow',
    'depends_on_past': False,
    'email': ['airflow@example.com'],
    'email_on_failure': False,
    'email_on_retry': False,
    'retries': 1,
    'retry_delay': timedelta(minutes=5),
}

# تعريف الـ DAG
with DAG(
    'data_pipeline_for_ai',
    default_args=default_args,
    description='خط أنابيب بيانات ETL وتدريب نموذج AI',
    schedule_interval=timedelta(days=1), # تشغيل يومي
    start_date=datetime(2023, 1, 1),
    catchup=False,
    tags=['data_engineering', 'ai'],
) as dag:
    
    # تعريف المهام (Tasks)
    extract_task = PythonOperator(
        task_id='extract_raw_data',
        python_callable=extract_data,
    )

    transform_task = PythonOperator(
        task_id='transform_and_clean_data',
        python_callable=transform_data,
    )

    load_task = PythonOperator(
        task_id='load_to_data_warehouse',
        python_callable=load_data,
    )
    
    train_model_task = PythonOperator(
        task_id='train_ai_model',
        python_callable=train_ml_model,
    )

    # تحديد ترتيب المهام (Dependencies)
    extract_task >> transform_task >> load_task >> train_model_task

print("تم تعريف DAG 'data_pipeline_for_ai' بنجاح. يمكن رؤيته في واجهة مستخدم Airflow.")
# ملاحظة: هذا الكود لن يقوم بتشغيل Airflow فعليًا، بل يوضح كيفية تعريف DAG.
```

<div dir="rtl">## القسم الخامس: تحديات متقدمة في هندسة البيانات

تتجاوز هندسة البيانات مجرد نقل وتخزين البيانات لتشمل جوانب أكثر تعقيدًا تضمن موثوقية وقيمة البيانات.

### 5.1 جودة البيانات (Data Quality)

جودة البيانات هي مدى ملاءمة البيانات للاستخدام المقصود. البيانات ذات الجودة الرديئة يمكن أن تؤدي إلى قرارات خاطئة ونماذج ذكاء اصطناعي غير دقيقة. تشمل أبعاد جودة البيانات: الدقة، الاكتمال، الاتساق، التوقيت، والصلاحية.

**التحديات:**
*   البيانات المفقودة أو غير الصحيحة.
*   التنسيقات غير المتسقة.
*   القيم المكررة.
*   الانحرافات (Outliers).

**الحلول:**
*   فحوصات التحقق من صحة البيانات (Data Validation).
*   تنظيف البيانات (Data Cleansing).
*   المراقبة المستمرة لجودة البيانات.
*   تطبيق قواعد العمل (Business Rules).
</div>

```
# مثال على فحص جودة البيانات باستخدام Pandas

import pandas as pd
import numpy as np

# بيانات وهمية مع مشاكل في الجودة
data_quality = {
    'transaction_id': [101, 102, 103, 104, 105, 106],
    'product_name': ['Laptop', 'Mouse', 'Keyboard', 'Laptop', 'Monitor', 'Mouse'],
    'price': [1200.00, 25.50, 75.00, 1200.00, 'غير صالح', 25.50],
    'quantity': [1, 2, 0, 1, 1, -1], # كمية صفر أو سالبة غير منطقية
    'customer_email': ['a@example.com', 'b@example.com', 'c@example.com', 'a@example.com', np.nan, 'b@example.com']
}
df_quality = pd.DataFrame(data_quality)
print("\nالبيانات الأولية مع مشاكل الجودة:\n", df_quality)

# 1. فحص القيم المفقودة (Completeness)
print("\nالقيم المفقودة لكل عمود:\n", df_quality.isnull().sum())

# 2. فحص القيم المكررة (Uniqueness)
print("\nالصفوف المكررة:\n", df_quality[df_quality.duplicated(subset=['transaction_id'], keep=False)])

# 3. فحص الصلاحية (Validity) - تحويل 'price' إلى رقمي ومعالجة الأخطاء
df_quality['price'] = pd.to_numeric(df_quality['price'], errors='coerce')
print("\nالقيم غير الصالحة في 'price' بعد التحويل:\n", df_quality[df_quality['price'].isnull()])

# 4. فحص الاتساق (Consistency) - الكمية يجب أن تكون أكبر من صفر
print("\nالكميات غير المنطقية (أقل من أو تساوي صفر):\n", df_quality[df_quality['quantity'] <= 0])

# تطبيق بعض قواعد التنظيف:
# إزالة الصفوف ذات الكمية غير الصالحة
df_cleaned = df_quality[df_quality['quantity'] > 0].copy()

# ملء القيم المفقودة في 'customer_email' بقيمة افتراضية أو إزالتها
df_cleaned['customer_email'].fillna('unknown@example.com', inplace=True)

# إزالة الصفوف التي لا تزال تحتوي على قيم NaN في 'price' بعد التحويل (إذا لم يتمكن من إصلاحها)
df_cleaned.dropna(subset=['price'], inplace=True)

print("\nالبيانات بعد التنظيف الأولي:\n", df_cleaned)
```

```
# الحل النموذجي للتمرين العملي

from pyspark.sql import SparkSession
from pyspark.sql.functions import col, sum

# 1. إنشاء جلسة Spark
spark = SparkSession.builder \
    .appName("SimpleSparkPipelineSolution") \
    .getOrCreate()

# 2. إنشاء DataFrame بسيط (بيانات مبيعات وهمية)
data = [
    ("Laptop", 1200, 2),
    ("Mouse", 25, 10),
    ("Keyboard", 75, 5),
    ("Monitor", 300, 3),
    ("Webcam", 50, 8),
    ("Headphones", 150, 4)
]
columns = ["product_name", "price", "quantity"]
df_sales = spark.createDataFrame(data, columns)

print("\nالبيانات الأولية:\n")
df_sales.show()

# 3. أضف عمودًا جديدًا 'total_revenue' (السعر * الكمية)
df_with_revenue = df_sales.withColumn("total_revenue", col("price") * col("quantity"))

print("\nالبيانات مع عمود الإيرادات الكلية:\n")
df_with_revenue.show()

# 4. قم بتجميع البيانات حسب 'product_name' واحسب مجموع 'total_revenue'
product_revenue_summary = df_with_revenue.groupBy("product_name").agg(sum("total_revenue").alias("total_revenue"))

print("\nملخص الإيرادات لكل منتج:\n")
product_revenue_summary.show()

# 5. قم بتصفية المنتجات التي تجاوزت إيراداتها 500 وحدة نقدية
high_revenue_products = product_revenue_summary.filter(col("total_revenue") > 500)

print("\nالمنتجات ذات الإيرادات العالية (أكثر من 500):\n")
high_revenue_products.show()

# إيقاف جلسة Spark
spark.stop()
```

<div dir="rtl">## تحدي المشروع: تصميم نظام توصية للمنتجات في الوقت الفعلي

**السيناريو:** أنت جزء من فريق هندسة البيانات في شركة تجارة إلكترونية سريعة النمو. قررت الشركة بناء نظام توصية للمنتجات في الوقت الفعلي لزيادة المبيعات وتحسين تجربة المستخدم. يجب أن يكون النظام قادرًا على معالجة أحداث المستخدم (مثل مشاهدات المنتجات، الإضافات إلى سلة التسوق، عمليات الشراء) فور حدوثها، وتحديث التوصيات للعميل في غضون ثوانٍ.

**المهمة (في مجموعات صغيرة):**

1.  **تصميم معماري:** ارسم مخططًا معماريًا عالي المستوى (High-Level Architecture Diagram) يوضح المكونات الرئيسية لنظام التوصية هذا. يجب أن يشمل المخطط:
    *   مصادر البيانات (أحداث المستخدم، بيانات المنتجات، بيانات العملاء).
    *   طبقات الاستيعاب (Ingestion Layer).
    *   طبقات المعالجة (Processing Layer) - مع تحديد ما إذا كانت دفعات أو تدفق.
    *   طبقات التخزين (Storage Layer) - مع تحديد أنواع قواعد البيانات أو أنظمة الملفات.
    *   طبقة التقديم (Serving Layer) - كيف ستصل التوصيات إلى المستخدم.
    *   أدوات الأتمتة والمراقبة (Automation & Monitoring).

2.  **اختيار الأدوات:** لكل مكون رئيسي في مخططك، اقترح أداة أو تقنية محددة (مثل Apache Kafka، Apache Spark، Cassandra، Redis، S3، Airflow، إلخ) واشرح سبب اختيارك لها.

3.  **التحديات والحلول:** حدد تحديين تقنيين رئيسيين قد تواجههما في بناء هذا النظام (مثل جودة البيانات، قابلية التوسع، زمن الاستجابة، أمان البيانات) واقترح حلولاً موجزة لكل تحدي.
