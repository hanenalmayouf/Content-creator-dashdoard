## 1. ما هي نماذج اللغات الكبيرة (LLMs)؟

نماذج اللغات الكبيرة (Large Language Models) هي شبكات عصبية عميقة، غالبًا ما تكون من نوع المحولات (Transformers)، تم تدريبها على كميات هائلة من البيانات النصية. هدفها الأساسي هو فهم اللغة البشرية وتوليدها. يمكنها أداء مجموعة واسعة من المهام مثل الترجمة، التلخيص، الإجابة على الأسئلة، وحتى كتابة الشفرات البرمجية.

**كيف تعمل؟**

في جوهرها، تحاول LLMs التنبؤ بالكلمة التالية في تسلسل معين بناءً على الكلمات السابقة. هذا يبدو بسيطًا، لكن مع مليارات المعاملات (parameters) والتدريب على تيرابايت من البيانات، تكتسب هذه النماذج قدرة مذهلة على فهم السياق، القواعد النحوية، وحتى بعض جوانب المنطق.

*   **الترميز (Tokenization)**: تحويل النص إلى وحدات أصغر (كلمات، أجزاء كلمات، رموز) يمكن للنموذج معالجتها.
*   **الـ Embeddings**: تمثيل هذه الرموز كمتجهات رقمية في فضاء متعدد الأبعاد، حيث تكون الكلمات ذات المعاني المتشابهة قريبة من بعضها.
*   **آلية الانتباه (Attention Mechanism)**: تسمح للنموذج بالتركيز على الأجزاء الأكثر أهمية في النص المدخل عند توليد الاستجابة.
</div>

````
# مثال بسيط على تفاعل مع LLM (باستخدام النموذج الوهمي)

# لنفترض أننا نريد من النموذج تلخيص نص ما.
text_to_summarize = """
تعتبر نماذج اللغات الكبيرة (LLMs) ثورة في مجال الذكاء الاصطناعي، حيث أظهرت قدرات فائقة في فهم وتوليد اللغة البشرية. تعتمد هذه النماذج على بنية المحولات (Transformers) وتتدرب على مجموعات بيانات ضخمة جدًا من النصوص والتعليمات البرمجية. وقد أدى ظهورها إلى تطورات كبيرة في مجالات مثل معالجة اللغة الطبيعية، الترجمة الآلية، وإنشاء المحتوى، مما يفتح آفاقًا جديدة للابتكار والتطبيقات العملية في مختلف الصناعات.
"""

prompt_1 = f"لخص النص التالي في جملة واحدة: {text_to_summarize}"
response_1 = mock_llm.generate(prompt_1)
print(f"الاستجابة: {response_1.strip()}")

prompt_2 = "اكتب قصيدة قصيرة عن جمال الطبيعة."
response_2 = mock_llm.generate(prompt_2)
print(f"الاستجابة: {response_2.strip()}")
````

## 2. لماذا هندسة الأوامر (Prompt Engineering)؟

على الرغم من قوة LLMs، فإنها ليست سحرية. جودة استجاباتها تعتمد بشكل كبير على جودة التعليمات التي تتلقاها. هنا يأتي دور هندسة الأوامر.

**هندسة الأوامر** هي فن وعلم صياغة المدخلات (الأوامر) لـ LLMs بطريقة تمكنها من توليد الاستجابات الأكثر دقة، صلة، وفائدة للمهمة المطلوبة.

**لماذا هي حاسمة؟**

*   **فتح القدرات الكامنة**: الأمر المصاغ جيدًا يمكن أن يطلق العنان لقدرات النموذج التي قد لا تظهر بأمر سيء الصياغة.
*   **تقليل "الهلوسة" (Hallucinations)**: الأوامر الواضحة والمحددة تقلل من ميل النموذج لتوليد معلومات غير صحيحة أو غير ذات صلة.
*   **تحسين الدقة والصلة**: الحصول على الإجابات التي تحتاجها بالضبط، بدلاً من إجابات عامة أو غير مفيدة.
*   **توجيه سلوك النموذج**: يمكنك توجيه النموذج لتبني شخصية معينة، أو تنسيق إخراج محدد، أو التركيز على جوانب معينة من المهمة.
</div>

## 3. بنية الأمر الفعال (Prompt Structure)

الأمر الفعال ليس مجرد سؤال. إنه غالبًا ما يكون مزيجًا من عدة عناصر توجه النموذج نحو الاستجابة المطلوبة. المكونات الرئيسية تشمل:

1.  **التعليمات (Instruction)**: ما الذي يجب على النموذج فعله؟ (مثال: "لخص"، "اكتب"، "صنف"، "اشرح").
2.  **السياق (Context)**: أي معلومات خلفية ضرورية للنموذج لفهم المهمة. (مثال: النص المراد تلخيصه، البيانات المراد تحليلها).
3.  **بيانات الإدخال (Input Data)**: البيانات الفعلية التي سيعالجها النموذج. (مثال: المقال، الكود، المحادثة).
4.  **تنسيق الإخراج (Output Format)**: كيف يجب أن تبدو الاستجابة؟ (مثال: "في شكل قائمة نقطية"، "ككود بايثون"، "بصيغة JSON").
5.  **الأمثلة (Examples - Few-shot)**: تقديم أمثلة على المدخلات والمخرجات المطلوبة لمساعدة النموذج على فهم النمط.
6.  **الشخصية/الدور (Persona/Role)**: توجيه النموذج لتبني دور معين. (مثال: "أنت خبير تسويق"، "أنت مساعد برمجي").
</div>

## 4. تقنيات هندسة الأوامر: Zero-shot vs. Few-shot

هناك طريقتان رئيسيتان لتقديم الأوامر للنماذج اللغوية الكبيرة:

### أ. الأوامر الصفرية (Zero-shot Prompting)

في هذه التقنية، لا يتم تزويد النموذج بأي أمثلة على المهمة. يُتوقع من النموذج أن يفهم المهمة ويولد استجابة بناءً على تدريبه العام فقط. إنها مفيدة للمهام البسيطة أو عندما يكون النموذج قد تدرب جيدًا على المهمة المطلوبة.

**مثال:**
"صنف النص التالي إلى إيجابي أو سلبي: 'أنا أحب هذا المنتج، إنه رائع!'"
</div>

````
# مثال على Zero-shot Prompting

zero_shot_prompt = "صنف النص التالي إلى 'إيجابي' أو 'سلبي': 'خدمة العملاء كانت سيئة للغاية، لن أعود أبدًا.'"
response_zero_shot = mock_llm.generate(zero_shot_prompt)
print(f"الأمر (Zero-shot): {zero_shot_prompt}")
print(f"الاستجابة: {response_zero_shot.strip()}")

print("\n---")

zero_shot_prompt_2 = "استخرج الكلمات المفتاحية من الجملة التالية: 'الذكاء الاصطناعي يُحدث ثورة في معالجة البيانات الضخمة.'"
response_zero_shot_2 = mock_llm.generate(zero_shot_prompt_2)
print(f"الأمر (Zero-shot): {zero_shot_prompt_2}")
print(f"الاستجابة: {response_zero_shot_2.strip()}")
````

<div dir="rtl">
### ب. الأوامر القليلة اللقطات (Few-shot Prompting)

في هذه التقنية، يتم تزويد النموذج بعدد قليل من الأمثلة (عادة 1-5 أمثلة) للمدخلات والمخرجات المطلوبة قبل تقديم المهمة الفعلية. تساعد هذه الأمثلة النموذج على فهم النمط المطلوب وتوجيه استجابته بشكل أفضل، خاصة للمهام الأكثر تعقيدًا أو التي تتطلب تنسيقًا محددًا.

**مثال:**
"النص: 'الطقس جميل اليوم.' الشعور: إيجابي.
النص: 'أنا متعب جدًا.' الشعور: سلبي.
النص: 'هذا الفيلم ممل.' الشعور: "
</div>

````
# مثال على Few-shot Prompting

few_shot_prompt = """
النص: 'أنا أحب هذا الكتاب.' الشعور: إيجابي.
النص: 'كانت التجربة مخيبة للآمال.' الشعور: سلبي.
النص: 'هذا المنتج يعمل بشكل جيد.' الشعور: إيجابي.
النص: 'لم أستمتع بالوجبة على الإطلاق.' الشعور:
"""
response_few_shot = mock_llm.generate(few_shot_prompt)
print(f"الأمر (Few-shot): {few_shot_prompt}")
print(f"الاستجابة: {response_few_shot.strip()}")

print("\n---")

few_shot_prompt_2 = """
المدخل: 'Python is a great language.' الترجمة: 'بايثون لغة رائعة.'
المدخل: 'Hello world.' الترجمة: 'أهلاً بالعالم.'
المدخل: 'Prompt engineering is crucial.' الترجمة:
"""
response_few_shot_2 = mock_llm.generate(few_shot_prompt_2)
print(f"الأمر (Few-shot): {few_shot_prompt_2}")
print(f"الاستجابة: {response_few_shot_2.strip()}")
````

````
# الحل النموذجي لتحدي صياغة الأمر

# هذا هو مثال على أمر مصاغ جيدًا يمكن أن يُعطى لـ LLM حقيقي:

solution_prompt = """
أنت خبير في تطوير بايثون ومعالجة اللغة الطبيعية. مهمتك هي كتابة دالة بايثون لتحليل المشاعر.

**التعليمات:**
1.  اكتب دالة بايثون باسم `analyze_sentiment`.
2.  تأخذ الدالة وسيطًا واحدًا: `text` (سلسلة نصية تمثل الجملة أو الفقرة المراد تحليلها).
3.  يجب أن تُرجع الدالة سلسلة نصية واحدة من بين القيم التالية: 'إيجابي' (positive)، 'سلبي' (negative)، أو 'محايد' (neutral).
4.  استخدم مكتبة `textblob` لتحليل المشاعر. إذا لم تكن مثبتة، أضف تعليقًا يوضح كيفية تثبيتها.
5.  تأكد من أن الكود قابل للتشغيل ويحتوي على تعليقات توضيحية.
6.  أضف مثالين على الأقل لاستخدام الدالة.

**مثال على تنسيق الإخراج المطلوب:**
```python
# الكود هنا
```

الآن، اكتب دالة `analyze_sentiment`:
"""

# محاكاة استجابة النموذج (في الواقع، LLM حقيقي سيولد الكود)
# print(mock_llm.generate(solution_prompt, max_length=500))

# الكود المتوقع من LLM (الحل النموذجي):

print("--- الحل النموذجي للكود ---")

# قد تحتاج إلى تثبيت textblob: pip install textblob
# ثم تنزيل بياناتها: python -m textblob.download_corpora

from textblob import TextBlob

def analyze_sentiment(text):
    """
    تحلل مشاعر نص معين وتُرجع 'إيجابي'، 'سلبي'، أو 'محايد'.
    """
    analysis = TextBlob(text)
    
    # تحديد المشاعر بناءً على القطبية (polarity)
    if analysis.sentiment.polarity > 0:
        return 'إيجابي'
    elif analysis.sentiment.polarity < 0:
        return 'سلبي'
    else:
        return 'محايد'

# أمثلة على الاستخدام:
print(f"'أنا أحب هذا المنتج!' -> {analyze_sentiment('أنا أحب هذا المنتج!')}")
print(f"'كانت التجربة سيئة للغاية.' -> {analyze_sentiment('كانت التجربة سيئة للغاية.')}")
print(f"'الطقس اليوم عادي.' -> {analyze_sentiment('الطقس اليوم عادي.')}")
print(f"'هذا الفيلم رائع ومثير للإعجاب.' -> {analyze_sentiment('هذا الفيلم رائع ومثير للإعجاب.')}")
print(f"'لا أحب هذه الفكرة.' -> {analyze_sentiment('لا أحب هذه الفكرة.')}")
````

### 3. أسئلة سريعة (Quiz Bowl)

سيتم استخدام منصة تفاعلية لطرح أسئلة تقنية متعمقة حول بنية LLMs، أنواع الـ Embeddings، وتحديات الـ Tokenization، وذلك لضمان استيعاب المفاهيم الأساسية قبل الانتقال إلى المستويات الأكثر تقدماً.

**مثال على الأسئلة:**

*   ما هي الوحدة الأساسية التي تتعامل معها LLMs بعد عملية الـ Tokenization؟
*   اشرح باختصار آلية الانتباه (Attention Mechanism) في سياق نماذج المحولات (Transformers).
*   ما هو الفرق الرئيسي بين Zero-shot و Few-shot prompting؟

## مشروع تطبيقي للدرس (Project Challenge)

**المهمة**: بناء نظام بسيط لتصنيف رسائل البريد الإلكتروني كـ "مهم" أو "غير مهم" باستخدام LLM.

**الخطوات المقترحة:**

1.  **جمع البيانات**: قم بإنشاء 5-10 أمثلة لرسائل بريد إلكتروني (نص فقط) وصنفها يدويًا إلى "مهم" أو "غير مهم".
2.  **صياغة الأمر (Prompt Engineering)**: صغ أمرًا لـ LLM (باستخدام تقنية Few-shot Prompting) يتضمن الأمثلة التي جمعتها، ويوجه النموذج لتصنيف رسائل بريد إلكتروني جديدة.
3.  **الاختبار**: اختبر الأمر الخاص بك على 3-5 رسائل بريد إلكتروني جديدة لم يرها النموذج من قبل.
4.  **التحسين**: إذا لم تكن النتائج مرضية، قم بتحسين الأمر الخاص بك (تغيير الصياغة، إضافة المزيد من الأمثلة، تحديد السياق بشكل أفضل).

**الناتج المطلوب**: الأمر النهائي الذي صغته، مع 3 أمثلة لرسائل بريد إلكتروني جديدة وكيف قام النموذج بتصنيفها.
