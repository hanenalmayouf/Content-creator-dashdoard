<div dir="rtl"># معمل تصميم خطوط أنابيب البيانات القابلة للتوسع (Distributed Data Pipelines)

أهلاً بكم في هذا المعمل التفاعلي. سنقوم اليوم باستكشاف كيفية بناء أنظمة معالجة بيانات قوية باستخدام Apache Kafka و Apache Spark. الهدف هو فهم كيفية تقليل زمن التأخير (Latency) وضمان تناسق البيانات في الأنظمة الموزعة.</div>

```
# استيراد المكتبات الأساسية لمحاكاة معالجة البيانات
import time
import random

# محاكاة تدفق بيانات بسيط (Producer)
def simulate_data_stream(topic_name, count):
    print(f"بدء إرسال البيانات إلى الموضوع: {topic_name}")
    for i in range(count):
        data = {"id": i, "value": random.random()}
        # محاكاة زمن التأخير
        time.sleep(0.1)
        yield data

# عرض عينة من البيانات
for record in simulate_data_stream("sensor_data", 5):
    print(record)
```

<div dir="rtl">## تحدي المعمارية المنهارة
في هذا القسم، سنواجه مشكلة 'عنق الزجاجة'. الكود أدناه يعاني من تأخير عالٍ بسبب المعالجة التسلسلية للبيانات. مهمتك هي تحسين الكود ليكون أكثر كفاءة.</div>

```
# تمرين: تحسين معالجة البيانات (تحدي المعمارية المنهارة)
# TODO: قم بتعديل الدالة أدناه لاستخدام المعالجة المتوازية (Multiprocessing) بدلاً من المعالجة التسلسلية لتقليل زمن التأخير

def process_data_slow(data_list):
    results = []
    for item in data_list:
        # محاكاة عملية معالجة ثقيلة
        time.sleep(0.5)
        results.append(item['value'] * 2)
    return results

data = [{'value': i} for i in range(10)]
start = time.time()
process_data_slow(data)
print(f"زمن التنفيذ: {time.time() - start} ثانية")
```

```
# الحل النموذجي باستخدام المعالجة المتوازية
from multiprocessing import Pool

def heavy_task(item):
    time.sleep(0.5)
    return item['value'] * 2

def process_data_fast(data_list):
    with Pool(processes=4) as pool:
        return pool.map(heavy_task, data_list)

start = time.time()
process_data_fast(data)
print(f"زمن التنفيذ بعد التحسين: {time.time() - start} ثانية")
```
