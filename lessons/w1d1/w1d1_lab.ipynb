## 1. دور مهندس البيانات في منظومة الذكاء الاصطناعي

مهندس البيانات هو المعماري الذي يبني الأساس الذي تقوم عليه جميع تطبيقات الذكاء الاصطناعي. بدون بيانات نظيفة، منظمة، ومتاحة، لا يمكن لنموذج تعلم الآلة أن يعمل بفعالية. يضمن مهندس البيانات أن البيانات تتدفق بسلاسة من مصادرها إلى حيث تحتاجها نماذج الذكاء الاصطناعي.

**المهام الرئيسية:**
*   **استيعاب البيانات (Ingestion):** جلب البيانات من مصادر متنوعة (قواعد بيانات، ملفات، APIs، أجهزة استشعار).
*   **تحويل البيانات (Transformation):** تنظيف، توحيد، وإعداد البيانات للاستهلاك.
*   **تخزين البيانات (Storage):** تصميم وإدارة حلول التخزين الفعالة للبيانات الضخمة.
*   **أتمتة خطوط الأنابيب (Orchestration):** بناء وجدولة خطوط أنابيب البيانات.
*   **جودة البيانات (Data Quality):** ضمان دقة واكتمال واتساق البيانات.

دعونا نرى مثالاً بسيطاً لكيفية بدء مهندس البيانات في التعامل مع البيانات.
</div>

````
# استيراد المكتبات الأساسية لمعالجة البيانات
import pandas as pd
import numpy as np

# محاكاة بيانات خام من مصدر ما (على سبيل المثال، ملف CSV)
# في سيناريو حقيقي، قد تكون هذه البيانات قادمة من قاعدة بيانات أو API
data = {
    'user_id': [1, 2, 3, 4, 5, 6],
    'product_id': ['A1', 'B2', 'C3', 'A1', 'D4', 'B2'],
    'purchase_amount': [100.50, 25.75, 120.00, 99.99, 50.25, 26.00],
    'purchase_date': ['2023-01-01', '2023-01-01', '2023-01-02', '2023-01-02', '2023-01-03', '2023-01-03'],
    'status': ['completed', 'pending', 'completed', 'completed', 'failed', 'completed']
}
df_raw = pd.DataFrame(data)

print("\nالبيانات الخام المستوعبة:")
print(df_raw.head())

# مثال على خطوة تحويل بسيطة: حساب إجمالي المبيعات لكل منتج
# مهندس البيانات يقوم بتصميم هذه التحويلات لتجهيز البيانات للتحليل أو نماذج الذكاء الاصطناعي
df_transformed = df_raw.groupby('product_id')['purchase_amount'].sum().reset_index()
df_transformed.rename(columns={'purchase_amount': 'total_sales_amount'}, inplace=True)

print("\nالبيانات بعد التحويل (إجمالي المبيعات لكل منتج):")
print(df_transformed.head())

# مثال على خطوة جودة البيانات: التحقق من وجود قيم مفقودة
print("\nالتحقق من القيم المفقودة:")
print(df_raw.isnull().sum())

print("\nهذه مجرد لمحة بسيطة عن كيفية بدء مهندس البيانات في التعامل مع البيانات. في الأنظمة الموزعة، تكون هذه العمليات أكثر تعقيدًا وتتطلب أدوات متخصصة.")
````

## 2. أنماط استيعاب ومعالجة البيانات الضخمة (Batch و Streaming)

عند التعامل مع البيانات الضخمة، هناك نمطان رئيسيان لاستيعاب ومعالجة البيانات، ولكل منهما استخداماته وميزاته:

### 2.1 معالجة الدفعات (Batch Processing)

تتضمن معالجة الدفعات جمع البيانات على مدى فترة زمنية محددة (ساعات، أيام)، ثم معالجتها دفعة واحدة. هذا النمط مناسب للبيانات التي لا تتطلب استجابة فورية، مثل التقارير اليومية، تحليلات نهاية الشهر، أو تدريب نماذج التعلم الآلي التي لا تحتاج إلى تحديث مستمر.

**أمثلة:**
*   حساب إجمالي المبيعات الشهرية.
*   تحليل سلوك المستخدمين على مدى أسبوع.
*   إعادة تدريب نموذج توصية مرة واحدة يومياً.

**المميزات:**
*   كفاءة عالية في استخدام الموارد لمعالجة كميات كبيرة من البيانات.
*   أقل تعقيداً في التصميم والتنفيذ مقارنة بالتدفق.
*   مناسبة للعمليات التي يمكن أن تتحمل تأخيراً.

**العيوب:**
*   تأخير في توفر البيانات المعالجة.
*   قد لا تكون مناسبة للسيناريوهات التي تتطلب استجابة فورية.
</div>

````
# محاكاة معالجة الدفعات باستخدام Pandas
# لنفترض أن لدينا بيانات مبيعات كبيرة يتم جمعها يوميًا

# إنشاء بيانات دفعات (Batch Data)
np.random.seed(42)
batch_size = 100000 # حجم الدفعة
data_batch = {
    'transaction_id': range(batch_size),
    'customer_id': np.random.randint(1, 10000, batch_size),
    'amount': np.random.uniform(10, 1000, batch_size),
    'timestamp': pd.to_datetime('2023-01-01') + pd.to_timedelta(np.random.randint(0, 24*60*60, batch_size), unit='s')
}
df_batch = pd.DataFrame(data_batch)

print("\nأول 5 صفوف من بيانات الدفعة:")
print(df_batch.head())
print(f"حجم الدفعة: {len(df_batch)} سجل")

# مثال على معالجة الدفعة: حساب متوسط قيمة المعاملة لكل عميل
# هذه العملية تتم على الدفعة الكاملة بعد جمعها
print("\nبدء معالجة الدفعة... (قد يستغرق بعض الوقت للبيانات الكبيرة)")

average_amount_per_customer = df_batch.groupby('customer_id')['amount'].mean().reset_index()
average_amount_per_customer.rename(columns={'amount': 'average_transaction_amount'}, inplace=True)

print("\nمتوسط قيمة المعاملة لكل عميل (أول 5 نتائج):")
print(average_amount_per_customer.head())
print("\nتمت معالجة الدفعة بنجاح.")
````

### 2.2 معالجة التدفق (Streaming Processing)

تتضمن معالجة التدفق معالجة البيانات فور وصولها، في الوقت الفعلي أو شبه الفعلي. هذا النمط ضروري للسيناريوهات التي تتطلب استجابة فورية أو تحليلات مستمرة، حيث لا يمكن تحمل أي تأخير.

**أمثلة:**
*   اكتشاف الاحتيال في المعاملات المالية.
*   مراقبة أداء الأنظمة في الوقت الفعلي.
*   توصيات المنتجات الفورية على موقع التجارة الإلكترونية.
*   تحليل بيانات أجهزة الاستشعار في إنترنت الأشياء (IoT).

**المميزات:**
*   استجابة فورية للبيانات الجديدة.
*   تحديثات مستمرة للتحليلات والنتائج.
*   مناسبة للسيناريوهات الحساسة للوقت.

**العيوب:**
*   أكثر تعقيداً في التصميم والتنفيذ.
*   تتطلب موارد حوسبة مستمرة.
*   تحديات في ضمان اتساق البيانات وموثوقيتها.
</div>

````
# محاكاة معالجة التدفق (Streaming Processing)
# في بيئة حقيقية، قد نستخدم Apache Kafka أو Flink لاستيعاب ومعالجة التدفقات
# هنا، سنقوم بمحاكاة تدفق بسيط باستخدام مولد بيانات (generator)

import time
import random

def generate_stream_data(num_events=10):
    """مولد بيانات يحاكي تدفق الأحداث."""
    for i in range(num_events):
        event = {
            'event_id': i + 1,
            'sensor_id': random.randint(1, 5),
            'temperature': round(random.uniform(20.0, 30.0), 2),
            'timestamp': pd.Timestamp.now().isoformat()
        }
        yield event
        time.sleep(0.5) # محاكاة وصول الأحداث بفاصل زمني

print("\nبدء محاكاة تدفق البيانات...")

# معالجة كل حدث فور وصوله
processed_events = []
for event in generate_stream_data(num_events=5):
    print(f"\nحدث جديد وصل: {event}")
    # مثال على معالجة فورية: التحقق إذا كانت درجة الحرارة عالية بشكل غير عادي
    if event['temperature'] > 28.0:
        print(f"🚨 تنبيه: درجة حرارة عالية ({event['temperature']}°C) من المستشعر {event['sensor_id']}!")
    processed_events.append(event)
    
print("\nتمت معالجة جميع الأحداث في التدفق المحاكى.")
print(f"إجمالي الأحداث المعالجة: {len(processed_events)}")
````

## 3. تصميم بنى تخزين البيانات الموزعة (Data Lakes, Data Warehouses, Lakehouses)

تخزين البيانات الضخمة يتطلب بنى تحتية متخصصة. سنتعرف على ثلاثة أنماط رئيسية:

### 3.1 بحيرات البيانات (Data Lakes)

بحيرة البيانات هي مستودع مركزي يسمح لك بتخزين جميع بياناتك المنظمة وغير المنظمة على أي نطاق. يمكنك تخزين البيانات كما هي، دون الحاجة إلى هيكلة مسبقة. هذا يمنحك مرونة كبيرة لاستكشاف البيانات وتحليلها لاحقاً باستخدام أدوات مختلفة.

**المميزات:**
*   مرونة عالية في تخزين أنواع مختلفة من البيانات (نص، صور، فيديو، سجلات).
*   تخزين البيانات الخام (Raw Data) دون تحويل مسبق.
*   قابلية توسع كبيرة وبتكلفة منخفضة.
*   مناسبة لعلماء البيانات ومهندسي تعلم الآلة لاستكشاف البيانات.

**العيوب:**
*   قد تتحول إلى 'مستنقع بيانات' (Data Swamp) إذا لم تتم إدارتها بشكل صحيح.
*   تحديات في جودة البيانات واكتشاف المخططات (Schema Discovery).

### 3.2 مستودعات البيانات (Data Warehouses)

مستودع البيانات هو نظام تخزين مصمم خصيصاً للتحليلات وإعداد التقارير. يتم فيه تخزين البيانات المنظمة بعد تنظيفها وتحويلها وتجميعها في مخططات محددة (مثل مخطط النجمة Star Schema). يركز على توفير بيانات عالية الجودة ومحسنة للاستعلامات التحليلية السريعة.

**المميزات:**
*   أداء استعلامات تحليلي ممتاز.
*   جودة بيانات عالية ومخططات محددة جيداً.
*   مناسبة لتقارير الأعمال ولوحات المعلومات (Dashboards).
*   سهولة الاستخدام لمحللي الأعمال.

**العيوب:**
*   أقل مرونة في التعامل مع البيانات غير المنظمة.
*   تكلفة أعلى مقارنة ببحيرات البيانات.
*   يتطلب تحويلاً مسبقاً للبيانات قبل التخزين.

### 3.3 بيوت البحيرة (Lakehouses)

الـ Lakehouse هو بنية جديدة تجمع بين أفضل ميزات بحيرات البيانات ومستودعات البيانات. يوفر مرونة بحيرة البيانات في تخزين البيانات الخام وغير المنظمة، مع توفير هياكل البيانات والجودة والأداء التحليلي لمستودع البيانات. يتم تحقيق ذلك عادةً باستخدام تنسيقات ملفات مفتوحة مثل Delta Lake أو Apache Iceberg أو Apache Hudi فوق بحيرة البيانات.

**المميزات:**
*   يجمع بين مرونة بحيرة البيانات وأداء مستودع البيانات.
*   يدعم البيانات المنظمة وغير المنظمة.
*   يوفر خصائص ACID (الذرية، الاتساق، العزل، المتانة) على بحيرة البيانات.
*   يقلل من تعقيد البنية التحتية للبيانات.

**العيوب:**
*   تقنية حديثة نسبياً، قد تتطلب خبرة متخصصة.
*   قد تكون هناك تحديات في النضج البيئي للأدوات.
</div>

````
# محاكاة تفاعل مع بحيرة بيانات (Data Lake) باستخدام نظام الملفات المحلي
# في بيئة سحابية، قد يكون هذا S3 (AWS) أو ADLS (Azure) أو GCS (GCP)

import os
import json
import shutil

# إنشاء مجلد يمثل بحيرة البيانات
data_lake_path = 'my_data_lake'
if os.path.exists(data_lake_path):
    shutil.rmtree(data_lake_path)
os.makedirs(os.path.join(data_lake_path, 'raw_data', 'sales'), exist_ok=True)
os.makedirs(os.path.join(data_lake_path, 'processed_data', 'analytics'), exist_ok=True)

print(f"تم إنشاء بحيرة البيانات الافتراضية في: {data_lake_path}")

# 1. تخزين بيانات خام (غير منظمة) في بحيرة البيانات (مثلاً، سجلات JSON)
raw_log_data = [
    {'timestamp': '2023-01-01T10:00:00Z', 'event': 'user_login', 'user_id': 101, 'ip_address': '192.168.1.1'},
    {'timestamp': '2023-01-01T10:05:00Z', 'event': 'page_view', 'user_id': 101, 'page': '/home'},
    {'timestamp': '2023-01-01T10:10:00Z', 'event': 'user_logout', 'user_id': 101}
]

with open(os.path.join(data_lake_path, 'raw_data', 'logs_2023_01_01.json'), 'w') as f:
    json.dump(raw_log_data, f, indent=2)

print("\nتم تخزين بيانات سجلات خام في بحيرة البيانات.")

# 2. تخزين بيانات منظمة (CSV) في بحيرة البيانات
sales_data_csv = """product_id,quantity,price
A1,10,100.50
B2,5,25.75
C3,12,120.00
"""
with open(os.path.join(data_lake_path, 'raw_data', 'sales', 'sales_2023_01_01.csv'), 'w') as f:
    f.write(sales_data_csv)

print("تم تخزين بيانات مبيعات منظمة في بحيرة البيانات.")

# 3. قراءة البيانات من بحيرة البيانات (مثال على استكشاف البيانات)
print("\nقراءة البيانات من بحيرة البيانات:")

# قراءة ملف JSON
with open(os.path.join(data_lake_path, 'raw_data', 'logs_2023_01_01.json'), 'r') as f:
    logs = json.load(f)
print("سجلات JSON المقروءة:")
for log in logs:
    print(log)

# قراءة ملف CSV باستخدام Pandas
df_sales = pd.read_csv(os.path.join(data_lake_path, 'raw_data', 'sales', 'sales_2023_01_01.csv'))
print("\nبيانات المبيعات CSV المقروءة:")
print(df_sales)

# 4. محاكاة تخزين بيانات معالجة (مثلاً، في تنسيق Parquet لتحسين الأداء التحليلي)
# في بيئة حقيقية، قد نستخدم Spark لتحويل البيانات إلى Parquet
processed_sales_data = df_sales.groupby('product_id')['quantity'].sum().reset_index()
processed_sales_data.to_csv(os.path.join(data_lake_path, 'processed_data', 'analytics', 'total_quantity_per_product.csv'), index=False)

print("\nتم تخزين بيانات معالجة في بحيرة البيانات (مجلد processed_data).")

print("\nهذا يوضح مرونة بحيرة البيانات في تخزين أنواع مختلفة من البيانات الخام والمعالجة.")
````

## 4. أدوات رائدة في هندسة البيانات الموزعة

لتحقيق قابلية التوسع والكفاءة في معالجة البيانات الضخمة، نعتمد على مجموعة من الأدوات القوية:

### 4.1 Apache Spark

Apache Spark هو محرك حوسبة موحد لتحليل البيانات الضخمة ومعالجة تعلم الآلة. يوفر واجهات برمجة تطبيقات (APIs) بلغات متعددة (Python, Scala, Java, R) ويدعم معالجة الدفعات والتدفق، مما يجعله أداة متعددة الاستخدامات لمعظم مهام هندسة البيانات.

**الميزات الرئيسية:**
*   **معالجة في الذاكرة (In-memory processing):** أسرع بكثير من Hadoop MapReduce.
*   **واجهات برمجة تطبيقات سهلة الاستخدام:** مثل DataFrames و SQL.
*   **دعم متعدد اللغات:** Python (PySpark) هو الأكثر شيوعاً لعلماء ومهندسي البيانات.
*   **وحدات مدمجة:** Spark SQL, Spark Streaming, MLlib (تعلم آلة), GraphX (معالجة الرسوم البيانية).

### 4.2 Apache Kafka

Apache Kafka هو منصة تدفق موزعة عالية الأداء. يستخدم لاستيعاب البيانات في الوقت الفعلي، بناء خطوط أنابيب بيانات تدفقية، وتطبيقات تدفقية تتفاعل مع البيانات فور وصولها. إنه العمود الفقري للعديد من أنظمة البيانات التي تتطلب معالجة فورية.

**الميزات الرئيسية:**
*   **قابلية توسع عالية:** يمكنه التعامل مع آلاف الرسائل في الثانية.
*   **متانة عالية:** يخزن الرسائل بشكل دائم على القرص.
*   **وقت استجابة منخفض:** مثالي لمعالجة البيانات في الوقت الفعلي.
*   **نظام نشر/اشتراك (Publish/Subscribe):** يفصل بين منتجي البيانات ومستهلكيها.

### 4.3 Apache Airflow

Apache Airflow هو منصة لإنشاء، جدولة، ومراقبة خطوط أنابيب البيانات برمجياً. يسمح لك بتعريف مهام سير العمل (Workflows) كـ DAGs (Directed Acyclic Graphs) باستخدام كود بايثون، مما يجعله مرناً وقابلاً للتوسع. Airflow هو أداة أساسية لأتمتة مهام هندسة البيانات.

**الميزات الرئيسية:**
*   **تعريف سير العمل ككود (DAGs in Python):** يسهل الصيانة والتحكم في الإصدارات.
*   **واجهة مستخدم رسومية (UI):** لمراقبة وإدارة المهام.
*   **قابلية التوسع:** يمكنه جدولة آلاف المهام.
*   **إعادة المحاولة والتنبيهات:** يدعم آليات قوية للتعامل مع الفشل.

````
# محاكاة بسيطة لمفاهيم Apache Spark باستخدام Pandas (لأن Spark يتطلب بيئة خاصة)
# في بيئة Spark حقيقية، سنستخدم SparkSession و DataFrames

print("\nمحاكاة عملية Spark DataFrame باستخدام Pandas...")

# إنشاء DataFrame يمثل بيانات كبيرة
data_spark_sim = {
    'order_id': range(1, 10001),
    'customer_id': np.random.randint(1, 1000, 10000),
    'item_count': np.random.randint(1, 5, 10000),
    'price': np.random.uniform(5.0, 500.0, 10000)
}
df_spark_sim = pd.DataFrame(data_spark_sim)

print("أول 5 صفوف من DataFrame المحاكى:")
print(df_spark_sim.head())

# عملية تحويل تشبه Spark: حساب إجمالي الإيرادات لكل عميل
# في Spark، ستكون هذه العملية موزعة على عدة عقد
revenue_per_customer = df_spark_sim.groupby('customer_id').agg(
    total_items=('item_count', 'sum'),
    total_revenue=('price', 'sum')
).reset_index()

print("\nإجمالي الإيرادات لكل عميل (أول 5 نتائج):")
print(revenue_per_customer.head())

print("\nهذا يوضح كيف يمكن لـ Spark معالجة البيانات بشكل تجميعي وفعال.")

# ------------------------------------------------------------------

# محاكاة بسيطة لمفاهيم Apache Kafka (المنتج والمستهلك)
# في بيئة Kafka حقيقية، سنستخدم مكتبات مثل `confluent-kafka` أو `kafka-python`

import collections

class MockKafkaTopic:
    """فئة بسيطة لمحاكاة موضوع Kafka."""
    def __init__(self, name):
        self.name = name
        self._messages = collections.deque() # قائمة انتظار للرسائل

    def produce(self, message):
        """إضافة رسالة إلى الموضوع."""
        self._messages.append(message)
        print(f"[Kafka Producer] تم إنتاج رسالة إلى الموضوع '{self.name}': {message}")

    def consume(self):
        """استهلاك رسالة من الموضوع."""
        if self._messages:
            message = self._messages.popleft()
            print(f"[Kafka Consumer] تم استهلاك رسالة من الموضوع '{self.name}': {message}")
            return message
        else:
            print(f"[Kafka Consumer] لا توجد رسائل في الموضوع '{self.name}'.")
            return None

# إنشاء موضوع Kafka وهمي
orders_topic = MockKafkaTopic('new_orders')

# محاكاة منتج (Producer) يرسل رسائل
print("\nمحاكاة منتج Kafka...")
orders_topic.produce({'order_id': 1001, 'product': 'Laptop', 'quantity': 1})
orders_topic.produce({'order_id': 1002, 'product': 'Mouse', 'quantity': 2})

# محاكاة مستهلك (Consumer) يستقبل رسائل
print("\nمحاكاة مستهلك Kafka...")
orders_topic.consume()
orders_topic.consume()
orders_topic.consume() # محاولة استهلاك رسالة لا توجد

print("\nهذا يوضح كيف يمكن لـ Kafka تسهيل الاتصال غير المتزامن بين مكونات النظام.")

# ------------------------------------------------------------------

# محاكاة بسيطة لمفهوم Apache Airflow DAG (Directed Acyclic Graph)
# في بيئة Airflow حقيقية، يتم تعريف هذه في ملفات .py داخل مجلد DAGs

from datetime import datetime

def task_extract_data():
    print("\n[Airflow Task] استخراج البيانات من المصدر...")
    # كود لاستخراج البيانات
    return "data_extracted.csv"

def task_transform_data(input_file):
    print(f"[Airflow Task] تحويل البيانات من {input_file}...")
    # كود لتحويل البيانات
    return "data_transformed.parquet"

def task_load_data(input_file):
    print(f"[Airflow Task] تحميل البيانات المحولة {input_file} إلى الوجهة...")
    # كود لتحميل البيانات
    print("[Airflow Task] تم تحميل البيانات بنجاح.")

# تعريف DAG وهمي (في Airflow الحقيقي، سيكون هذا كائن DAG)
print("\nمحاكاة Airflow DAG...")
def run_data_pipeline():
    print(f"بدء خط أنابيب البيانات في {datetime.now()}")
    
    # تحديد ترتيب المهام (التبعيات)
    extracted_file = task_extract_data()
    transformed_file = task_transform_data(extracted_file)
    task_load_data(transformed_file)
    
    print("تم الانتهاء من خط أنابيب البيانات.")

# تشغيل خط الأنابيب المحاكى
run_data_pipeline()

print("\nهذا يوضح كيف يقوم Airflow بتنظيم وجدولة المهام في خط أنابيب البيانات.")
````

````
# مثال على التحقق من جودة البيانات باستخدام Pandas assertions
# في بيئة حقيقية، قد نستخدم مكتبات متخصصة مثل Great Expectations

print("\nالتحقق من جودة البيانات...")

# بيانات مثال قد تحتوي على مشاكل جودة
data_quality_check = {
    'product_id': ['A1', 'B2', 'C3', 'A1', 'D4', 'E5', 'F6'],
    'price': [100.50, 25.75, 120.00, 99.99, 50.25, -10.00, np.nan], # سعر سالب وقيمة مفقودة
    'category': ['Electronics', 'Food', 'Electronics', 'Electronics', 'Books', 'Food', 'Books']
}
df_quality = pd.DataFrame(data_quality_check)

print("البيانات الأصلية:")
print(df_quality)

# 1. التحقق من عدم وجود قيم مفقودة في عمود 'product_id'
try:
    assert df_quality['product_id'].isnull().sum() == 0, "يوجد قيم مفقودة في عمود product_id!"
    print("✅ لا توجد قيم مفقودة في product_id.")
except AssertionError as e:
    print(f"❌ فشل التحقق: {e}")

# 2. التحقق من أن الأسعار إيجابية
try:
    assert (df_quality['price'] >= 0).all(), "يوجد أسعار سالبة في عمود price!"
    print("✅ جميع الأسعار إيجابية.")
except AssertionError as e:
    print(f"❌ فشل التحقق: {e}")
    # معالجة القيم السالبة (مثلاً، تحويلها إلى NaN أو إزالتها)
    df_quality['price'] = df_quality['price'].apply(lambda x: x if x >= 0 else np.nan)
    print("تم تصحيح الأسعار السالبة إلى NaN.")

# 3. التحقق من أن جميع الفئات معروفة (مثلاً، قائمة محددة)
allowed_categories = ['Electronics', 'Food', 'Books', 'Home']
try:
    assert df_quality['category'].isin(allowed_categories).all(), "يوجد فئات غير معروفة!"
    print("✅ جميع الفئات معروفة.")
except AssertionError as e:
    print(f"❌ فشل التحقق: {e}")
    # معالجة الفئات غير المعروفة (مثلاً، تعيينها إلى 'Other')
    df_quality['category'] = df_quality['category'].apply(lambda x: x if x in allowed_categories else 'Other')
    print("تم تصحيح الفئات غير المعروفة إلى 'Other'.")

print("\nالبيانات بعد تطبيق تحسينات الجودة:")
print(df_quality)

print("\nهذا يوضح أهمية التحقق من جودة البيانات كجزء من خط أنابيب البيانات.")
````
