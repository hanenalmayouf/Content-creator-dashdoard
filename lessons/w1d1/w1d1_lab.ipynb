<div dir="rtl">## 1. مفاهيم المعمارية: Batch vs Stream
- **Batch Processing**: معالجة البيانات على دفعات (مثل ETL التقليدي).
- **Stream Processing**: معالجة البيانات لحظياً (مثل Kafka/Flink).
- **Lambda Architecture**: تجمع بين المعالجة الدفعية واللحظية.
- **Kappa Architecture**: تعتمد كلياً على المعالجة اللحظية (Stream-first).</div>

```
# محاكاة بسيطة لمعالجة البيانات
import pandas as pd

# بيانات خام (Raw Data)
data = {'id': [1, 2, 3], 'value': [100, 200, 300], 'timestamp': ['2023-10-01', '2023-10-01', '2023-10-01']}
df = pd.DataFrame(data)

# تحويل بسيط (Transformation)
def transform_data(df):
    df['value_taxed'] = df['value'] * 1.15
    return df

processed_df = transform_data(df)
print("البيانات بعد التحويل:")
print(processed_df)
```
