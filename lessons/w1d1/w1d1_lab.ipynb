<div dir="rtl">## 1. ما هي البيانات؟
البيانات هي حقائق خام (أرقام، نصوص، صور). عندما نقوم بمعالجتها وتنظيمها، تتحول إلى **معلومات** تساعدنا في فهم العالم من حولنا.</div>

```
# مثال توضيحي: الفرق بين البيانات الخام والمعلومات
# بيانات خام (قائمة درجات طلاب)
raw_data = [85, 92, 78, 95, 88]

# معالجة البيانات لاستخراج معلومة (المتوسط الحسابي)
average = sum(raw_data) / len(raw_data)

print(f"البيانات الخام: {raw_data}")
print(f"المعلومة المستخرجة (المتوسط): {average}")
```

<div dir="rtl">## 2. أنواع البيانات
- **بيانات منظمة (Structured):** مثل جداول Excel وقواعد البيانات SQL.
- **بيانات غير منظمة (Unstructured):** مثل الصور، الفيديوهات، والمنشورات النصية على وسائل التواصل.</div>

```
# تمرين عملي: تصنيف البيانات
# TODO: قم بتصنيف العناصر التالية في القائمة أدناه إلى 'منظمة' أو 'غير منظمة'

data_samples = {
    "جدول مبيعات": "...",
    "صورة شخصية": "...",
    "ملف CSV": "...",
    "مقطع فيديو": "..."
}

print(data_samples)
```

```
# الحل النموذجي للتمرين
data_samples = {
    "جدول مبيعات": "منظمة",
    "صورة شخصية": "غير منظمة",
    "ملف CSV": "منظمة",
    "مقطع فيديو": "غير منظمة"
}

for item, category in data_samples.items():
    print(f"العنصر: {item} -> النوع: {category}")
```

<div dir="rtl">## 4. المشروع التطبيقي (Project Challenge)
**المهمة:** قم بإنشاء قائمة تحتوي على 5 أنواع مختلفة من البيانات التي تتعامل معها يومياً، ثم اكتب كود بايثون يقوم بفلترة البيانات 'المنظمة' فقط وطباعتها.</div>

```
# هيكل المشروع التطبيقي
my_daily_data = [
    {"name": "سجل المكالمات", "type": "منظمة"},
    {"name": "صورة السيلفي", "type": "غير منظمة"},
    {"name": "جدول الحصص", "type": "منظمة"},
    {"name": "رسالة صوتية", "type": "غير منظمة"},
    {"name": "قائمة المهام", "type": "منظمة"}
]

# استخراج البيانات المنظمة فقط
structured_data = [d['name'] for d in my_daily_data if d['type'] == 'منظمة']

print("البيانات المنظمة في يومي:", structured_data)
```
