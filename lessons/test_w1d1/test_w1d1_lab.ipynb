<div dir="rtl">
  <h2>المتغيرات (Variables) في بايثون</h2>
  <p>المتغيرات هي أماكن في الذاكرة تستخدم لتخزين البيانات. يمكنك التفكير فيها كصناديق تحمل قيمًا مختلفة.</p>
  <p>في بايثون، لا تحتاج إلى تحديد نوع المتغير مسبقًا؛ بايثون تحدد النوع تلقائيًا بناءً على القيمة التي تخزنها.</p>
  
  <h3>قواعد تسمية المتغيرات:</h3>
  <ul>
    <li>يجب أن تبدأ بحرف أو بشرطة سفلية (<code>_</code>).</li>
    <li>لا يمكن أن تبدأ برقم.</li>
    <li>يمكن أن تحتوي على أحرف أبجدية رقمية (A-Z, a-z, 0-9) وشرطات سفلية.</li>
    <li>أسماء المتغيرات حساسة لحالة الأحرف (<code>name</code> تختلف عن <code>Name</code>).</li>
    <li>لا يمكن استخدام الكلمات المحجوزة في بايثون (مثل <code>if</code>, <code>for</code>, <code>while</code>).</li>
  </ul>
  
  <h3>أنواع البيانات الأساسية:</h3>
  <ul>
    <li><b>الأعداد الصحيحة (Integers):</b> أرقام كاملة مثل <code>10</code>, <code>-5</code>.</li>
    <li><b>الأعداد العشرية (Floats):</b> أرقام تحتوي على كسور عشرية مثل <code>3.14</code>, <code>-0.5</code>.</li>
    <li><b>السلاسل النصية (Strings):</b> نصوص محاطة بعلامات اقتباس مثل <code>"مرحباً"</code>, <code>'بايثون'</code>.</li>
    <li><b>القيم المنطقية (Booleans):</b> قيمتان فقط: <code>True</code> أو <code>False</code>.</li>
  </ul>
</div>

```
# مثال على تعريف المتغيرات واستخدامها

# متغير من نوع عدد صحيح (Integer)
العمر = 30
print("العمر:", العمر)

# متغير من نوع عدد عشري (Float)
درجة_الحرارة = 25.5
print("درجة الحرارة:", درجة_الحرارة)

# متغير من نوع سلسلة نصية (String)
الاسم = "سارة"
print("الاسم:", الاسم)

# متغير من نوع قيمة منطقية (Boolean)
هل_طالب = True
print("هل سارة طالبة؟", هل_طالب)

# يمكنك أيضاً معرفة نوع المتغير باستخدام دالة type()
print("نوع متغير العمر:", type(العمر))
print("نوع متغير الاسم:", type(الاسم))

# إعادة تعيين قيمة لمتغير
العمر = العمر + 1 # زيادة العمر بواحد
print("العمر بعد الزيادة:", العمر)
```

<div dir="rtl">
  <h2>العمليات الحسابية (Arithmetic Operations)</h2>
  <p>يمكنك إجراء عمليات حسابية بسيطة على المتغيرات والأرقام في بايثون:</p>
  <ul>
    <li>الجمع: <code>+</code></li>
    <li>الطرح: <code>-</code></li>
    <li>الضرب: <code>*</code></li>
    <li>القسمة: <code>/</code> (الناتج يكون دائمًا عشريًا)</li>
    <li>باقي القسمة (Modulo): <code>%</code></li>
    <li>الأس (Exponentiation): <code>**</code></li>
    <li>القسمة الصحيحة (Floor Division): <code>//</code> (الناتج يكون عددًا صحيحًا)</li>
  </ul>
</div>

```
# أمثلة على العمليات الحسابية

عدد_1 = 10
عدد_2 = 3

# الجمع
الناتج_جمع = عدد_1 + عدد_2
print(f"الجمع: {عدد_1} + {عدد_2} = {الناتج_جمع}") # استخدام f-string لتنسيق الطباعة

# الطرح
الناتج_طرح = عدد_1 - عدد_2
print(f"الطرح: {عدد_1} - {عدد_2} = {الناتج_طرح}")

# الضرب
الناتج_ضرب = عدد_1 * عدد_2
print(f"الضرب: {عدد_1} * {عدد_2} = {الناتج_ضرب}")

# القسمة (الناتج عشري)
الناتج_قسمة = عدد_1 / عدد_2
print(f"القسمة: {عدد_1} / {عدد_2} = {الناتج_قسمة}")

# باقي القسمة
الناتج_باقي = عدد_1 % عدد_2
print(f"باقي القسمة: {عدد_1} % {عدد_2} = {الناتج_باقي}")

# الأس
الناتج_أس = عدد_1 ** عدد_2 # 10 مرفوعة للقوة 3
print(f"الأس: {عدد_1} ** {عدد_2} = {الناتج_أس}")

# القسمة الصحيحة
الناتج_قسمة_صحيحة = عدد_1 // عدد_2
print(f"القسمة الصحيحة: {عدد_1} // {عدد_2} = {الناتج_قسمة_صحيحة}")
```

  <h2>الحصول على مدخلات من المستخدم: دالة <code>input()</code></h2>
  <p>دالة <code>input()</code> تسمح لبرنامجك بالتفاعل مع المستخدم عن طريق طلب إدخال بيانات منه.</p>
  <p>ملاحظة هامة: دالة <code>input()</code> دائمًا تُرجع قيمة من نوع سلسلة نصية (String)، حتى لو أدخل المستخدم أرقامًا. إذا كنت بحاجة إلى استخدام هذه الأرقام في عمليات حسابية، يجب تحويلها إلى أعداد صحيحة (<code>int()</code>) أو عشرية (<code>float()</code>).</p>
</div>

```
# مثال 1: الحصول على اسم المستخدم
اسم_المستخدم = input("من فضلك، أدخل اسمك: ")
print(f"أهلاً بك، {اسم_المستخدم}!")

# مثال 2: الحصول على رقمين وإجراء عملية حسابية
# لاحظ أننا نحتاج لتحويل المدخلات إلى أعداد صحيحة باستخدام int()
الرقم_الأول_نص = input("أدخل الرقم الأول: ")
الرقم_الثاني_نص = input("أدخل الرقم الثاني: ")

الرقم_الأول = int(الرقم_الأول_نص)
الرقم_الثاني = int(الرقم_الثاني_نص)

المجموع = الرقم_الأول + الرقم_الثاني
print(f"مجموع الرقمين هو: {المجموع}")

# مثال 3: تحويل إلى عدد عشري
الطول_نص = input("أدخل طولك بالمتر (مثال: 1.75): ")
الطول = float(الطول_نص)
print(f"طولك هو: {الطول} متر")
```
