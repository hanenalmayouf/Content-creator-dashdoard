### أنماط التصميم (Design Patterns)
أنماط التصميم هي حلول عامة وقابلة لإعادة الاستخدام لمشاكل شائعة تحدث في تصميم البرمجيات. إنها ليست تصميمات جاهزة يمكن تحويلها مباشرة إلى كود، بل هي قوالب لكيفية حل مشكلة معينة في سياقات مختلفة.
<br><br>
**مثال: نمط المصنع (Factory Pattern)**
يُستخدم نمط المصنع لإنشاء كائنات دون تحديد الفئة الدقيقة للكائن الذي سيتم إنشاؤه. هذا يوفر مرونة في إضافة أنواع جديدة من الكائنات دون تعديل الكود الحالي الذي يستخدم المصنع.
</div>

```
# مثال على نمط المصنع (Factory Pattern)

# 1. تعريف واجهة (أو فئة أساسية) للمنتجات
class Product:
    def operation(self):
        pass

# 2. تعريف منتجات ملموسة (Concrete Products)
class ConcreteProductA(Product):
    def operation(self):
        return "عملية المنتج A"

class ConcreteProductB(Product):
    def operation(self):
        return "عملية المنتج B"

# 3. تعريف المصنع (Factory)
class ProductFactory:
    def create_product(self, product_type):
        if product_type == "A":
            return ConcreteProductA()
        elif product_type == "B":
            return ConcreteProductB()
        else:
            raise ValueError("نوع المنتج غير مدعوم")

# استخدام المصنع لإنشاء المنتجات
factory = ProductFactory()

product1 = factory.create_product("A")
print(f"المنتج الأول: {product1.operation()}")

product2 = factory.create_product("B")
print(f"المنتج الثاني: {product2.operation()}")

# يمكننا بسهولة إضافة منتجات جديدة دون تعديل كود العميل الذي يستخدم المصنع
class ConcreteProductC(Product):
    def operation(self):
        return "عملية المنتج C"

# لتوسيع المصنع لدعم المنتج C، نحتاج فقط لتعديل دالة create_product
# (في أنماط مصنع أكثر تعقيدًا، يمكن أن يكون هذا التوسيع أكثر مرونة)
# لنفترض أننا قمنا بتحديث المصنع (هذا مثال توضيحي)
class UpdatedProductFactory:
    def create_product(self, product_type):
        if product_type == "A":
            return ConcreteProductA()
        elif product_type == "B":
            return ConcreteProductB()
        elif product_type == "C":
            return ConcreteProductC()
        else:
            raise ValueError("نوع المنتج غير مدعوم")

updated_factory = UpdatedProductFactory()
product3 = updated_factory.create_product("C")
print(f"المنتج الثالث: {product3.operation()}")
```

## القسم الثاني: هندسة المتطلبات (Requirements Engineering)
هندسة المتطلبات هي عملية تحديد وتوثيق وتحليل وإدارة المتطلبات اللازمة لنظام برمجي. إنها مرحلة حاسمة لضمان أن المنتج النهائي يلبي احتياجات المستخدمين وأصحاب المصلحة.
</div>

<div dir="rtl">
### مراحل هندسة المتطلبات
تتكون هندسة المتطلبات عادةً من عدة مراحل رئيسية:
1.  **استخلاص المتطلبات (Elicitation):** جمع المتطلبات من أصحاب المصلحة.
2.  **تحليل المتطلبات (Analysis):** فهم وتصنيف المتطلبات، وحل أي تعارضات.
3.  **تحديد المتطلبات (Specification):** توثيق المتطلبات بشكل واضح ومفصل.
4.  **التحقق من المتطلبات (Validation):** التأكد من أن المتطلبات صحيحة وكاملة وقابلة للتحقيق.
5.  **إدارة المتطلبات (Management):** التعامل مع التغييرات في المتطلبات طوال دورة حياة المشروع.
</div>

<div dir="rtl">
### أنواع المتطلبات
تنقسم المتطلبات بشكل عام إلى نوعين رئيسيين:
*   **المتطلبات الوظيفية (Functional Requirements):** تصف ما يجب أن يفعله النظام. على سبيل المثال: "يجب أن يتمكن المستخدم من تسجيل الدخول باستخدام اسم المستخدم وكلمة المرور."
*   **المتطلبات غير الوظيفية (Non-functional Requirements):** تصف كيفية أداء النظام لوظائفه، مثل الأداء، الأمان، قابلية الاستخدام، الموثوقية، وقابلية التوسع. على سبيل المثال: "يجب أن يستجيب النظام لطلبات تسجيل الدخول في أقل من ثانية واحدة."
</div>

<div dir="rtl">
### تقنيات استخلاص المتطلبات
هناك العديد من التقنيات المستخدمة لجمع المتطلبات من أصحاب المصلحة:
*   **المقابلات (Interviews):** محادثات فردية أو جماعية مع أصحاب المصلحة.
*   **ورش العمل (Workshops):** جلسات جماعية منظمة لجمع المتطلبات.
*   **الاستبيانات (Questionnaires):** لجمع المعلومات من عدد كبير من الأشخاص.
*   **تحليل المستندات (Document Analysis):** مراجعة المستندات الموجودة مثل أدلة التشغيل أو وثائق الأنظمة القديمة.
*   **الملاحظة (Observation):** مراقبة المستخدمين أثناء قيامهم بمهامهم.
*   **النماذج الأولية (Prototyping):** بناء نماذج مبسطة للنظام للحصول على ملاحظات.
</div>

<div dir="rtl">
### صياغة المتطلبات: قصص المستخدم (User Stories)
قصص المستخدم هي طريقة شائعة لوصف المتطلبات الوظيفية في منهجيات التطوير الرشيقة (Agile). تتبع عادةً التنسيق التالي:
<br>
**"بصفتي [نوع المستخدم]، أريد [وظيفة معينة]، لكي [قيمة أو فائدة]."**
<br><br>
**مثال:**
*   "بصفتي **عميلًا**، أريد **تصفح المنتجات حسب الفئة**، لكي **أجد ما أبحث عنه بسهولة**."
*   "بصفتي **مدير متجر**، أريد **إضافة منتجات جديدة إلى المخزون**، لكي **أحافظ على تحديث قائمة المنتجات**."
</div>

```
# مثال: تحليل قصة مستخدم بسيطة باستخدام بايثون

def parse_user_story(story_text):
    """
    تقوم بتحليل قصة مستخدم وتستخرج المكونات الرئيسية.
    تفترض أن القصة تتبع التنسيق: "بصفتي [دور]، أريد [وظيفة]، لكي [فائدة]."
    """
    parts = story_text.split('،')
    if len(parts) != 3:
        return {"error": "تنسيق قصة المستخدم غير صحيح."}

    as_a_part = parts[0].strip()
    i_want_part = parts[1].strip()
    so_that_part = parts[2].strip()

    role = as_a_part.replace("بصفتي", "").strip()
    function = i_want_part.replace("أريد", "").strip()
    benefit = so_that_part.replace("لكي", "").strip()

    return {
        "الدور": role,
        "الوظيفة": function,
        "الفائدة": benefit
    }

# أمثلة على قصص المستخدم
story1 = "بصفتي عميلًا، أريد تصفح المنتجات حسب الفئة، لكي أجد ما أبحث عنه بسهولة."
story2 = "بصفتي مدير متجر، أريد إضافة منتجات جديدة إلى المخزون، لكي أحافظ على تحديث قائمة المنتجات."
story3 = "بصفتي مستخدمًا، أريد تسجيل الدخول، لكي أصل إلى حسابي الشخصي."

print("تحليل قصة المستخدم 1:")
print(parse_user_story(story1))
print("\nتحليل قصة المستخدم 2:")
print(parse_user_story(story2))
print("\nتحليل قصة المستخدم 3:")
print(parse_user_story(story3))

# مثال على قصة مستخدم بتنسيق خاطئ
story_bad = "بصفتي عميلًا، أريد تصفح المنتجات."
print("\nتحليل قصة مستخدم خاطئة:")
print(parse_user_story(story_bad))
```

```
# الحل النموذجي لتمرين تصنيف المتطلبات

def classify_requirements_solution(requirements_list):
    """
    تصنف قائمة من المتطلبات إلى وظيفية وغير وظيفية.
    """
    functional_reqs = []
    non_functional_reqs = []
    
    non_functional_keywords = ["أداء", "أمان", "سرعة", "موثوقية", "قابلية التوسع", "استجابة", "واجهة المستخدم"]

    for req in requirements_list:
        is_non_functional = False
        for keyword in non_functional_keywords:
            if keyword in req: # تحقق مما إذا كانت الكلمة المفتاحية موجودة في المتطلب
                non_functional_reqs.append(req)
                is_non_functional = True
                break # بمجرد العثور على كلمة مفتاحية، نصنفه كغير وظيفي وننتقل للمتطلب التالي
        
        if not is_non_functional:
            functional_reqs.append(req)
    
    return {
        "functional": functional_reqs,
        "non_functional": non_functional_reqs
    }

# قائمة المتطلبات للاختبار
sample_requirements_solution = [
    "يجب أن يتمكن المستخدم من تسجيل الدخول.",
    "يجب أن يكون زمن استجابة النظام أقل من 3 ثوانٍ.",
    "يجب أن يوفر النظام تقارير مبيعات شهرية.",
    "يجب أن تكون واجهة المستخدم سهلة الاستخدام.",
    "يجب أن يدعم النظام 1000 مستخدم متزامن لضمان الأداء.",
    "يجب أن يتم تشفير بيانات المستخدم لضمان الأمان.",
    "يجب أن يتمكن المستخدم من إضافة منتجات إلى سلة التسوق."
]

# اختبر الدالة الخاصة بالحل
classified_results_solution = classify_requirements_solution(sample_requirements_solution)
print("المتطلبات الوظيفية (الحل):")
for req in classified_results_solution["functional"]:
    print(f"- {req}")

print("\nالمتطلبات غير الوظيفية (الحل):")
for req in classified_results_solution["non_functional"]:
    print(f"- {req}")
```
