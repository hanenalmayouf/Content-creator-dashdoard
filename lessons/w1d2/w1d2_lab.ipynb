## تهيئة بيئة العمل ومحاكاة النموذج اللغوي

لغرض هذا المعمل، سنقوم بإنشاء دالة بسيطة تحاكي استجابة نموذج لغوي كبير (LLM). هذه الدالة ستساعدنا على فهم كيفية تأثير السياق والتعليمات على المخرجات دون الحاجة إلى ربط فعلي بنموذج حقيقي.
</div>

````
# استيراد المكتبات اللازمة (إن وجدت)
import textwrap

# دالة محاكاة النموذج اللغوي (LLM)
# هذه الدالة بسيطة جداً وتستخدم قواعد أساسية لمحاكاة استجابات مختلفة بناءً على الكلمات المفتاحية في الأمر.
# في بيئة حقيقية، ستكون هذه الدالة عبارة عن استدعاء لواجهة برمجة تطبيقات (API) لنموذج لغوي حقيقي.
def simulate_llm_response(prompt: str) -> str:
    prompt_lower = prompt.lower()
    response = ""

    if "توليد كود بايثون" in prompt_lower:
        if "دالة" in prompt_lower and "جمع" in prompt_lower:
            response += "```python\ndef add(a, b):\n    return a + b\n```\n"
            if "شرح" in prompt_lower:
                response += "هذه الدالة تقوم بجمع عددين. "
            if "اختبار" in prompt_lower:
                response += "مثال للاستخدام: `add(5, 3)` ينتج `8`. "
        elif "قائمة" in prompt_lower and "فرز" in prompt_lower:
            response += "```python\ndef sort_list(data):\n    return sorted(data)\n```\n"
            if "تصاعدي" in prompt_lower:
                response += "هذه الدالة تفرز القائمة تصاعدياً. "
            if "تنازلي" in prompt_lower:
                response += "للفزر التنازلي، استخدم `sorted(data, reverse=True)`. "
        else:
            response += "يرجى تحديد نوع الكود المطلوب بشكل أوضح (مثال: دالة، فئة، خوارزمية). "
    elif "تلخيص" in prompt_lower:
        if "نص" in prompt_lower and "قصير" in prompt_lower:
            response += "ملخص موجز للنص المقدم. "
            if "نقاط" in prompt_lower:
                response += "أهم النقاط هي: [النقطة 1]، [النقطة 2]. "
            if "محايد" in prompt_lower:
                response += "تم الحرص على أن يكون الملخص محايداً. "
        else:
            response += "يرجى تقديم النص المراد تلخيصه. "
    elif "تحليل" in prompt_lower:
        if "بيانات" in prompt_lower:
            response += "لتحليل البيانات، نحتاج إلى مزيد من التفاصيل حول نوع البيانات والأهداف. "
        else:
            response += "يرجى تحديد ما الذي تريد تحليله. "
    else:
        response += "لم أفهم طلبك بوضوح. يرجى تقديم أمر أكثر تحديداً. "

    return textwrap.dedent(response).strip()

print("تم تهيئة دالة محاكاة النموذج اللغوي بنجاح.")
````

## 1. بناء السياق الفعال (Context Construction)

السياق هو الأساس الذي يبني عليه النموذج فهمه للمهمة. بدون سياق واضح ومحدد، قد يقدم النموذج استجابات عامة أو غير دقيقة. تخيل أنك تطلب من شخص ما "أعطني شيئاً" - سيحتاج إلى سياق ليعرف ما الذي تريده بالضبط. الأمر نفسه ينطبق على النماذج اللغوية.

**أمثلة على عناصر السياق:**
*   **الخلفية:** معلومات أساسية عن الموضوع.
*   **الدور:** تحديد دور النموذج (مثال: "أنت خبير تسويق...").
*   **الجمهور المستهدف:** لمن توجه الاستجابة.
*   **الأمثلة:** تقديم أمثلة للمدخلات والمخرجات المتوقعة (Few-shot prompting).
</div>

````
# مثال 1: أمر بدون سياق واضح
print("--- أمر بدون سياق ---")
prompt_v1 = "اكتب كود بايثون."
response_v1 = simulate_llm_response(prompt_v1)
print(f"الأمر: {prompt_v1}\nالاستجابة:\n{response_v1}\n")

# مثال 2: إضافة سياق بسيط
print("--- أمر بسياق بسيط ---")
prompt_v2 = "توليد كود بايثون لدالة جمع عددين."
response_v2 = simulate_llm_response(prompt_v2)
print(f"الأمر: {prompt_v2}\nالاستجابة:\n{response_v2}\n")

# مثال 3: إضافة سياق أكثر تفصيلاً (تحديد الدور والهدف)
print("--- أمر بسياق تفصيلي ---")
prompt_v3 = "أنت مبرمج بايثون خبير. توليد كود بايثون لدالة تقوم بجمع عددين، مع شرح موجز للدالة ومثال على كيفية استخدامها."
response_v3 = simulate_llm_response(prompt_v3)
print(f"الأمر: {prompt_v3}\nالاستجابة:\n{response_v3}\n")

# نلاحظ كيف أن إضافة السياق (الدور، الهدف، التفاصيل) أدت إلى استجابة أكثر دقة وفائدة.
````

## 2. تقنيات التوجيه (Instruction Tuning)

بعد بناء السياق، تأتي مرحلة توجيه النموذج بتعليمات دقيقة. التعليمات هي الأوامر المحددة التي تخبر النموذج بما يجب فعله بالضبط. كلما كانت التعليمات أكثر وضوحاً وتحديداً، كلما كانت الاستجابة أقرب لما تتوقعه.

**نصائح للتوجيه الفعال:**
*   **الوضوح:** تجنب الغموض والكلمات التي تحتمل أكثر من معنى.
*   **التحديد:** كن محدداً قدر الإمكان حول المخرجات المطلوبة (مثال: "في 3 نقاط"، "بصيغة JSON").
*   **القيود:** حدد أي قيود (مثال: "لا تستخدم مكتبات خارجية"، "يجب أن يكون الملخص محايداً").
*   **التعليمات السلبية:** أخبر النموذج بما لا يجب فعله (مثال: "لا تضمن أي مقدمات أو خاتمات").
</div>

````
# مثال 1: أمر بتعليمات عامة
print("--- أمر بتعليمات عامة ---")
prompt_i1 = "لخص النص التالي: 'الذكاء الاصطناعي هو مجال واسع في علوم الحاسوب يهدف إلى إنشاء آلات ذكية تعمل وتتفاعل مثل البشر.'"
response_i1 = simulate_llm_response(prompt_i1)
print(f"الأمر: {prompt_i1}\nالاستجابة:\n{response_i1}\n")

# مثال 2: أمر بتعليمات محددة
print("--- أمر بتعليمات محددة ---")
prompt_i2 = "لخص النص التالي في نقطتين فقط، مع التركيز على التعريف والأهداف الرئيسية، وتأكد أن الملخص قصير ومحايد: 'الذكاء الاصطناعي هو مجال واسع في علوم الحاسوب يهدف إلى إنشاء آلات ذكية تعمل وتتفاعل مثل البشر. يشمل التعلم الآلي ومعالجة اللغة الطبيعية والرؤية الحاسوبية.'"
response_i2 = simulate_llm_response(prompt_i2)
print(f"الأمر: {prompt_i2}\nالاستجابة:\n{response_i2}\n")

# نلاحظ كيف أن التعليمات المحددة (عدد النقاط، التركيز، الحيادية) أدت إلى استجابة أكثر دقة وتوافقاً مع المطلوب.
````

## 3. التعديل المتكرر (Iterative Refinement) وتقليل التحيّز (Bias Reduction)

نادراً ما تحصل على الاستجابة المثالية من المحاولة الأولى. هندسة الأوامر هي عملية تكرارية. تبدأ بأمر أولي، تحلل الاستجابة، ثم تعدل الأمر بناءً على ما تعلمته. هذه العملية تساعد أيضاً في تقليل التحيّزات غير المرغوبة في مخرجات النموذج.

**خطوات التعديل المتكرر:**
1.  **صياغة الأمر الأولي:** ابدأ بأمر بسيط يحدد المهمة.
2.  **الحصول على الاستجابة:** قم بتشغيل الأمر على النموذج.
3.  **تحليل الاستجابة:** هل هي دقيقة؟ كاملة؟ هل تحتوي على تحيزات؟ هل هناك معلومات مفقودة؟
4.  **تعديل الأمر:** أضف سياقاً، عدّل التعليمات، أضف قيوداً، أو أزل الغموض.
5.  **التكرار:** كرر الخطوات حتى تحصل على الاستجابة المطلوبة.

**تقليل التحيّز:**
*   **تحديد التحيّز:** هل الاستجابة متحيزة ضد فئة معينة؟ هل تعكس قوالب نمطية؟
*   **إضافة قيود:** اطلب من النموذج أن يكون محايداً، موضوعياً، أو أن يأخذ في الاعتبار وجهات نظر متعددة.
*   **تغيير السياق:** قدم سياقاً أكثر شمولية أو تنوعاً.

## 4. تأثير طول السياق ونوعيته

طول السياق ونوعيته يؤثران بشكل كبير على أداء النموذج. السياق الطويل جداً قد يؤدي إلى "ضياع" النموذج في التفاصيل، بينما السياق القصير جداً قد لا يوفر معلومات كافية. نوعية السياق (مدى صلته بالمهمة، دقته، خلوه من التناقضات) لا تقل أهمية.

*   **السياق ذو الصلة:** ركز على المعلومات الضرورية للمهمة.
*   **التوازن:** لا تفرط في التفاصيل ولا تقلل منها.
*   **الجودة:** تأكد من أن السياق خالٍ من الأخطاء أو المعلومات المضللة.
