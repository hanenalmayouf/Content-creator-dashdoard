```
# تهيئة بيئة Spark
from pyspark.sql import SparkSession

# إنشاء جلسة Spark مع ضبط الذاكرة لتحسين الأداء
spark = SparkSession.builder \
    .appName("DistributedProcessingLab") \
    .config("spark.executor.memory", "2g") \
    .getOrCreate()

print("تم تهيئة بيئة Spark بنجاح! أنت الآن جاهز للبدء.")
```

<div dir="rtl">## الجزء الأول: فهم الـ Partitioning
توزيع البيانات هو سر السرعة في Spark. إذا كانت البيانات غير موزعة بشكل جيد، سيحدث ما يسمى بـ Data Skew، مما يؤدي إلى بطء شديد. دعونا نرى كيف نتحكم في عدد الـ Partitions.</div>

```
# إنشاء DataFrame تجريبي
data = [(i, f"user_{i}") for i in range(10000)]
df = spark.createDataFrame(data, ["id", "name"])

# TODO: قم بإعادة توزيع البيانات (Repartition) إلى 8 أقسام لتحسين التوازي
# df_optimized = ...

# الحل:
df_optimized = df.repartition(8)
print(f"عدد الأقسام الحالي: {df_optimized.rdd.getNumPartitions()}")
```

<div dir="rtl">## الجزء الثاني: محاكاة تدفق البيانات (Kafka Simulation)
في بيئة الإنتاج، نستخدم Kafka لجلب البيانات لحظياً. هنا سنحاكي تدفق البيانات ونقوم بمعالجتها باستخدام Spark Structured Streaming.</div>

```
# محاكاة تدفق بيانات بسيط
# في الواقع، سنقوم بالاتصال بـ Kafka Broker
streaming_df = spark.readStream \
    .format("rate") \
    .option("rowsPerSecond", 100) \
    .load()

# معالجة البيانات: إضافة عمود جديد يمثل وقت المعالجة
query = streaming_df.selectExpr("value", "current_timestamp() as processing_time") \
    .writeStream \
    .format("console") \
    .outputMode("append") \
    .start()

# ملاحظة: سيتم إيقاف الاستعلام بعد فترة قصيرة في بيئة المعمل
# query.stop()
```
