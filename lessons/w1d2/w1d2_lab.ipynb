<div dir="rtl">## 1. تحليل زمن التأخير (Latency Analysis)
في الأنظمة الموزعة، زمن التأخير هو الفرق بين وقت وصول الحدث ووقت معالجته. سنقوم بمحاكاة تدفق بيانات بسيط لقياس هذا التأخير.</div>

```
# محاكاة تدفق بيانات بسيط وقياس التأخير
def simulate_latency(data_points):
    latencies = []
    for i in range(data_points):
        start_time = time.time()
        # محاكاة معالجة البيانات
        time.sleep(0.01) 
        end_time = time.time()
        latencies.append(end_time - start_time)
    return sum(latencies) / len(latencies)

avg_latency = simulate_latency(100)
print(f"متوسط زمن التأخير في المحاكاة: {avg_latency:.4f} ثانية")
```
