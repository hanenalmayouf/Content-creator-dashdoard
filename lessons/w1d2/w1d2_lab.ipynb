<div dir="rtl"># معمل النمذجة التنبؤية وهندسة الميزات

أهلاً بك يا بطل في معملنا اليوم. تخيّل أننا في سوق تجاري كبير، والتاجر الذكي ليس من يملك بضاعة أكثر، بل من يعرف 'كيف' يرتّب بضاعته ليبيعها بأفضل سعر. هندسة الميزات (Feature Engineering) هي فن ترتيب هذه البضاعة، والنمذجة التنبؤية هي قدرتنا على توقع حركة السوق.</div>

```
# استيراد المكتبات الأساسية
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from xgboost import XGBClassifier
from sklearn.metrics import f1_score, roc_auc_score

# تحميل بيانات تجريبية (محاكاة لبيانات مبيعات)
data = pd.DataFrame({'price': [100, 200, 150, 300, 250], 'quality': [1, 5, 3, 4, 2], 'target': [0, 1, 0, 1, 1]})
print("تم تحميل البيانات بنجاح!")
```

<div dir="rtl">## هندسة الميزات (Feature Engineering)

البيانات الخام مثل التمر غير المغسول؛ تحتاج تنظيفاً وتنسيقاً. سنقوم الآن بإنشاء ميزة جديدة تعبر عن 'القيمة مقابل السعر'.</div>

```
# إنشاء ميزة جديدة: النسبة بين الجودة والسعر
data['value_ratio'] = data['quality'] / data['price']
print(data.head())
```

<div dir="rtl">## تمرين عملي: تدريب النموذج

الآن دورك! استخدم خوارزمية XGBoost لتدريب نموذج يتنبأ بالهدف (target). أكمل الكود الناقص أدناه.</div>

```
# تقسيم البيانات
X = data.drop('target', axis=1)
y = data['target']
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)

# TODO: قم بإنشاء كائن XGBClassifier وتدريبه باستخدام X_train و y_train
model = None 
# model.fit(...)

print("تمت محاولة التدريب (بانتظار حلك)")
```

```
# الحل النموذجي
model = XGBClassifier(use_label_encoder=False, eval_metric='logloss')
model.fit(X_train, y_train)
preds = model.predict(X_test)
print(f"دقة النموذج (F1-Score): {f1_score(y_test, preds)}")
```
