<div dir="rtl">## الجزء الأول: فهم زمن التأخير (Latency)
في الأنظمة الموزعة، زمن التأخير هو الفرق الزمني بين وصول الحدث ومعالجته. سنقوم بمحاكاة تدفق بيانات بسيط وقياس الوقت المستغرق.</div>

```
# محاكاة تدفق البيانات وقياس التأخير
def simulate_stream():
    start_time = time.time()
    data = "Event_Payload_" + str(random.randint(1, 100))
    # محاكاة معالجة البيانات
    time.sleep(0.05) 
    end_time = time.time()
    return end_time - start_time

latency = simulate_stream()
print(f"زمن التأخير الحالي: {latency:.4f} ثانية")
```

### قواعد التحدي:

3. سيتم مراقبة "مستوى ضغطك الذهني" (عبر محاكاة سرعة الاستجابة)؛ إذا زاد الضغط، سيقوم النظام بزيادة صعوبة البيانات المدخلة.

```
from kafka import KafkaProducer
import json

# إعداد المنتج (Producer) للمشاركة في التحدي
producer = KafkaProducer(bootstrap_servers=['localhost:9092'],
                         value_serializer=lambda x: json.dumps(x).encode('utf-8'))

# إرسال بيانات التحدي
for i in range(1000):
    producer.send('challenge_topic', {'event_id': i, 'value': random.random()})

print("تم إرسال 1000 حدث بنجاح. ابدأ المعالجة الآن!")
```
