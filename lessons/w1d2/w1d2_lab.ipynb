<div dir="rtl"># معمل الحوسبة العنقودية: سباق البيانات الضخمة

أهلاً بك يا بطل في معملنا اليوم. تخيّل أننا في سوق كبير جداً، والبيانات هي بضائع تصل بالآلاف كل ثانية. لو حاول شخص واحد (معالج واحد) ترتيبها، سيتعطل السوق. الحل؟ أن نوزع العمل على فريق كامل (عنقود - Cluster). اليوم سنتعلم كيف نوزع المهام باستخدام Apache Spark و Kafka.</div>

```
# تهيئة البيئة: استيراد المكتبات الأساسية
from pyspark.sql import SparkSession

# إنشاء جلسة سبارك (Spark Session) - تخيلها كمدير الفريق الذي يوزع المهام
spark = SparkSession.builder \
    .appName("BigDataRace") \
    .getOrCreate()

print("تم تجهيز مدير الفريق (Spark Session) بنجاح!")
```

<div dir="rtl">## التحدي الأول: معالجة البيانات الموزعة

في هذا الجزء، سنقوم بمحاكاة معالجة بيانات ضخمة. تذكر: السرعة ليست في قوة المعالج الواحد، بل في تقسيم العمل (Partitioning).</div>

```
# إنشاء بيانات تجريبية (محاكاة لتدفق بيانات من Kafka)
data = [("تمر", 100), ("قهوة", 50), ("تمور", 200), ("قهوة", 30)]
rdd = spark.sparkContext.parallelize(data)

# تمرين: قم بحساب مجموع الكميات لكل صنف باستخدام التجميع (ReduceByKey)
# TODO: أكمل الكود التالي لجمع القيم بناءً على المفتاح
# result = rdd.reduceByKey(lambda a, b: ...)

# الحل النموذجي:
result = rdd.reduceByKey(lambda a, b: a + b).collect()
print("نتائج المعالجة الموزعة:", result)
```

<div dir="rtl">## إدارة الإحباط التقني

هل واجهت خطأً برمجياً؟ لا تقلق، هذا جزء من رحلة التعلم. في الأنظمة الموزعة، الأخطاء غالباً ما تكون بسبب 'تأخر الشبكة' أو 'عدم توازن البيانات'. خذ نفساً عميقاً، اقرأ رسالة الخطأ، وتذكر أن حتى كبار المهندسين يواجهون هذه التحديات يومياً.</div>

```
# محاكاة تدفق بيانات بسيط (Kafka Stream Simulation)
# في الواقع نستخدم Kafka-Python، هنا نستخدم قائمة كطابور بيانات
kafka_stream = ["سعر_تمر: 20", "سعر_قهوة: 40", "سعر_تمر: 22"]

def process_stream(stream):
    for item in stream:
        # معالجة كل رسالة قادمة من التدفق
        key, value = item.split(": ")
        print(f"تمت معالجة: {key} بقيمة {value}")

process_stream(kafka_stream)
```
