<div dir="rtl">## المشروع التطبيقي (Project Challenge)

**المهمة:** قم ببناء دالة `validate_feature_store` تأخذ إطار بيانات (DataFrame) وتقوم بالتحقق من:
1. خلو البيانات من القيم المفقودة (Nulls).
2. التأكد من أن القيم تقع ضمن نطاق منطقي (مثلاً بين -3 و 3).
3. إرجاع تقرير حالة (Status Report) يوضح جودة البيانات قبل إرسالها للنموذج.</div>

```
def validate_feature_store(df):
    # تحقق من القيم المفقودة
    null_count = df.isnull().sum().sum()
    # تحقق من النطاق
    out_of_range = ((df['feature_1'] < -3) | (df['feature_1'] > 3)).sum()
    
    return {"nulls": null_count, "out_of_range": out_of_range}

report = validate_feature_store(df_production)
print(f"تقرير جودة البيانات: {report}")
```
