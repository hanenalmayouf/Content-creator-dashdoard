## 1. فهم التحيّز في الذكاء الاصطناعي: تحدي كشف التحيّز (Gamified Challenge)

التحيّز في الذكاء الاصطناعي يشير إلى ميل النظام لإظهار تفضيل غير عادل أو تمييز ضد مجموعات معينة من الأفراد. يمكن أن ينشأ هذا التحيّز من مصادر متعددة خلال دورة حياة نظام الذكاء الاصطناعي.

### مصادر التحيّز الرئيسية:
1.  **تحيّز البيانات (Data Bias):**
    *   **التحيّز التاريخي:** البيانات تعكس التمييز الموجود في المجتمع (مثل بيانات التوظيف التي تفضل الذكور تاريخياً).
    *   **تحيّز التمثيل:** مجموعات معينة ممثلة بشكل غير كافٍ أو مبالغ فيه في البيانات.
    *   **تحيّز القياس:** الأخطاء في كيفية جمع البيانات أو تسميتها.
2.  **تحيّز الخوارزمية (Algorithmic Bias):**
    *   **تحيّز التصميم:** اختيار ميزات أو نماذج تفضل مجموعة على أخرى.
    *   **تحيّز التفاعل:** تعزيز التحيّز الموجود من خلال التفاعل المستمر مع المستخدمين.
3.  **تحيّز المستخدم/البشري (Human/User Bias):**
    *   التحيّزات البشرية التي تؤثر على كيفية تصميم الأنظمة أو تفسير مخرجاتها.

في هذا القسم، سنقوم بتحميل مجموعة بيانات شائعة تُستخدم لدراسة الإنصاف، وهي مجموعة بيانات "Adult Income"، والتي تحتوي على معلومات ديموغرافية ومالية، والهدف هو التنبؤ بما إذا كان دخل الفرد يتجاوز 50 ألف دولار سنويًا. سنركز على الكشف عن التحيّز المحتمل بناءً على سمات حساسة مثل الجنس والعرق.
</div>

```
# تحميل مجموعة بيانات Adult Income باستخدام AIF360
# هذه المجموعة تستخدم عادة لدراسة الإنصاف
# سنقوم بتحديد السمات الحساسة (الجنس والعرق) والسمة المستهدفة (الدخل)

dataset_orig = StandardDataset(
    df=pd.read_csv('https://raw.githubusercontent.com/IBM/AIF360/master/aif360/data/raw/adult.csv', na_values='?'),
    label_name='income',
    favorable_classes=['>50K'],
    protected_attribute_names=['sex', 'race'],
    privileged_classes=[['Male'], ['White']],
    instance_weights_name=None,
    categorical_features=['workclass', 'education', 'marital-status', 'occupation', 'skill', 'relationship', 'native-country'],
    features_to_drop=['fnlwgt', 'education-num'] # إسقاط بعض الميزات غير الضرورية
)

# تنظيف البيانات: إزالة الصفوف التي تحتوي على قيم مفقودة بعد التحميل
dataset_orig.df.dropna(inplace=True)

# عرض أول 5 صفوف من البيانات بعد التحميل والتنظيف
print("\nأول 5 صفوف من مجموعة البيانات:")
print(dataset_orig.df.head())

# عرض معلومات عن السمات الحساسة
print("\nمعلومات عن السمات الحساسة:")
print(dataset_orig.protected_attribute_names)
print("المجموعات المميزة (Privileged Groups):", dataset_orig.privileged_protected_attributes)
print("المجموعات غير المميزة (Unprivileged Groups):", dataset_orig.unprivileged_protected_attributes)

# تقسيم البيانات إلى مجموعات تدريب واختبار
dataset_train, dataset_test = dataset_orig.split([0.7], shuffle=True)

print(f"\nعدد الأمثلة في مجموعة التدريب: {len(dataset_train.df)}")
print(f"عدد الأمثلة في مجموعة الاختبار: {len(dataset_test.df)}")
```

## 2. مقاييس الإنصاف الكمية (Quantitative Fairness Metrics)

لقياس الإنصاف بشكل كمي، نستخدم مقاييس محددة تساعدنا على فهم مدى عدالة النموذج تجاه المجموعات المختلفة. هذه المقاييس يمكن تقسيمها إلى فئات بناءً على المرحلة التي يتم فيها تقييم الإنصاف (قبل التدريب، أثناء التدريب، بعد التدريب).

### مقاييس الإنصاف الشائعة:
*   **التكافؤ الإحصائي (Statistical Parity):** يعني أن النموذج يجب أن يعطي نتائج إيجابية (مثل قبول طلب قرض) بنفس المعدل لجميع المجموعات، بغض النظر عن السمة الحساسة. يُقاس بـ `Disparate Impact` أو `Statistical Parity Difference`.
*   **تكافؤ الفرص (Equal Opportunity):** يركز على أن النموذج يجب أن يكون له نفس معدل الإيجابيات الحقيقية (True Positive Rate) لجميع المجموعات. أي، إذا كان الأفراد مؤهلين بالفعل، فيجب أن يتم تصنيفهم بشكل صحيح بنفس المعدل.
*   **التكافؤ التنبؤي (Predictive Parity):** يعني أن النموذج يجب أن يكون له نفس معدل الدقة التنبؤية (Precision) لجميع المجموعات.

سنستخدم مكتبة AIF360 لحساب هذه المقاييس.
</div>

```
# تدريب نموذج تصنيف أساسي (Baseline Model)

# إعداد محول الأعمدة للتعامل مع الميزات الرقمية والفئوية
# تحديد الميزات الرقمية والفئوية من مجموعة البيانات
# (نحن نستخدم dataset_train.features لضمان التناسق مع AIF360)

# تحديد الميزات الرقمية والفئوية يدويًا بناءً على مجموعة بيانات Adult
numeric_features = ['age', 'capital-gain', 'capital-loss', 'hours-per-week']
categorical_features = ['workclass', 'education', 'marital-status', 'occupation', 'relationship', 'race', 'sex', 'native-country']

preprocessor = ColumnTransformer(
    transformers=[
        ('num', StandardScaler(), numeric_features),
        ('cat', OneHotEncoder(handle_unknown='ignore'), categorical_features)
    ])

# بناء خط أنابيب (Pipeline) للنموذج
model_pipeline = Pipeline(steps=[
    ('preprocessor', preprocessor),
    ('classifier', LogisticRegression(solver='liblinear', random_state=42))
])

# فصل الميزات (X) والهدف (y) من مجموعة التدريب والاختبار
X_train = dataset_train.features
y_train = dataset_train.labels.ravel()
X_test = dataset_test.features
y_test = dataset_test.labels.ravel()

# تدريب النموذج
model_pipeline.fit(X_train, y_train)

# التنبؤ على مجموعة الاختبار
y_pred = model_pipeline.predict(X_test)

# تقييم أداء النموذج
accuracy = accuracy_score(y_test, y_pred)
print(f"\nدقة النموذج الأساسي: {accuracy:.4f}")
print("\nتقرير التصنيف:")
print(classification_report(y_test, y_pred))

# تحويل التنبؤات إلى تنسيق AIF360 لتقييم الإنصاف
dataset_test_pred = dataset_test.copy()
dataset_test_pred.labels = y_pred.reshape(-1, 1)

# تحديد المجموعات المميزة وغير المميزة لتقييم الإنصاف
# سنركز على سمة 'sex' (الجنس) كمثال
privileged_groups = [{'sex': 1}] # Male (عادة ما يتم ترميز الذكور بـ 1 في هذه المجموعة) 
unprivileged_groups = [{'sex': 0}] # Female (عادة ما يتم ترميز الإناث بـ 0)

# ملاحظة: يجب التأكد من الترميز الصحيح للسمات الحساسة بعد OneHotEncoder
# AIF360 يتعامل مع هذا بشكل أفضل إذا تم تمرير dataset_test_pred مباشرة
# ولكن لغرض التوضيح، سنستخدم القيم الأصلية للسمات الحساسة من dataset_test

# حساب مقاييس الإنصاف باستخدام AIF360
metric_orig_test = ClassificationMetric(
    dataset_test,
    dataset_test_pred,
    unprivileged_groups=unprivileged_groups,
    privileged_groups=privileged_groups
)

print("\n--- مقاييس الإنصاف للنموذج الأساسي (الجنس) ---")
print(f"فرق التأثير المتباين (Disparate Impact): {metric_orig_test.disparate_impact():.4f}")
print(f"فرق التكافؤ الإحصائي (Statistical Parity Difference): {metric_orig_test.statistical_parity_difference():.4f}")
print(f"فرق تكافؤ الفرص (Equal Opportunity Difference): {metric_orig_test.equal_opportunity_difference():.4f}")
print(f"فرق معدل الإيجابيات الكاذبة (False Positive Rate Difference): {metric_orig_test.false_positive_rate_difference():.4f}")

# تفسير النتائج:
# - Disparate Impact: النسبة بين معدل النتائج الإيجابية للمجموعة غير المميزة والمجموعة المميزة. القيمة المثالية هي 1.
# - Statistical Parity Difference: الفرق بين معدل النتائج الإيجابية للمجموعة غير المميزة والمجموعة المميزة. القيمة المثالية هي 0.
# - Equal Opportunity Difference: الفرق بين معدل الإيجابيات الحقيقية (True Positive Rate) للمجموعة غير المميزة والمجموعة المميزة. القيمة المثالية هي 0.
# - False Positive Rate Difference: الفرق بين معدل الإيجابيات الكاذبة للمجموعة غير المميزة والمجموعة المميزة. القيمة المثالية هي 0.

# إذا كانت القيم بعيدة عن 1 (لـ Disparate Impact) أو 0 (للفروق الأخرى)، فهذا يشير إلى وجود تحيّز.
```

## 3. استراتيجيات تخفيف التحيّز: بناء جسور الإنصاف (Collaborative Lab Session)

بعد تحديد التحيّز، الخطوة التالية هي تخفيفه. هناك عدة استراتيجيات يمكن تطبيقها في مراحل مختلفة من دورة حياة التعلم الآلي:

1.  **معالجة مسبقة (Pre-processing):** تعديل البيانات قبل تدريب النموذج.
    *   **إعادة الوزن (Reweighing):** تعديل أوزان الأمثلة في مجموعة البيانات لضمان تمثيل عادل للمجموعات المختلفة.
    *   **إزالة التأثير المتباين (Disparate Impact Remover):** تعديل قيم السمات الحساسة لتقليل التحيّز.
2.  **معالجة أثناء التدريب (In-processing):** تعديل خوارزمية التعلم نفسها أثناء عملية التدريب.
    *   **التعلم المعاكس (Adversarial Debiasing):** تدريب شبكة عصبية لتصنيف الهدف، وشبكة أخرى معادية لمحاولة التنبؤ بالسمة الحساسة من مخرجات الشبكة الأولى، بهدف جعل النموذج الأول لا يميز.
3.  **معالجة لاحقة (Post-processing):** تعديل مخرجات النموذج بعد التدريب.
    *   **تصنيف خيار الرفض (Reject Option Classification):** تحديد منطقة "رفض" حول عتبة التصنيف، حيث يتم اتخاذ قرارات مختلفة للمجموعات المميزة وغير المميزة.

في هذا القسم، سنطبق استراتيجية "إعادة الوزن" (Reweighing) من مكتبة AIF360 لتخفيف التحيّز بناءً على سمة الجنس، ثم نقوم بتدريب نموذج جديد وتقييم الإنصاف والدقة.

```
# تطبيق استراتيجية تخفيف التحيّز: إعادة الوزن (Reweighing)

# 1. تهيئة خوارزمية إعادة الوزن
# سنستخدم نفس المجموعات المميزة وغير المميزة لسمة 'sex'
privileged_groups = [{'sex': 1}]
unprivileged_groups = [{'sex': 0}]

RW = Reweighing(
    unprivileged_groups=unprivileged_groups,
    privileged_groups=privileged_groups
)

# 2. تطبيق إعادة الوزن على مجموعة التدريب
dataset_train_reweighed = RW.fit_transform(dataset_train)

# 3. تدريب نموذج جديد باستخدام البيانات المعاد وزنها
# يجب أن نستخدم أوزان الأمثلة الجديدة عند تدريب النموذج

# فصل الميزات (X) والهدف (y) من مجموعة التدريب المعاد وزنها
X_train_reweighed = dataset_train_reweighed.features
y_train_reweighed = dataset_train_reweighed.labels.ravel()
sample_weights = dataset_train_reweighed.instance_weights

# تدريب النموذج مع الأوزان
# يجب أن نمرر أوزان العينات إلى طريقة fit في خط الأنابيب
model_pipeline.fit(X_train_reweighed, y_train_reweighed, classifier__sample_weight=sample_weights)

# 4. التنبؤ على مجموعة الاختبار (التي لم يتم إعادة وزنها)
y_pred_reweighed = model_pipeline.predict(X_test)

# 5. تقييم أداء النموذج المعاد تدريبه
accuracy_reweighed = accuracy_score(y_test, y_pred_reweighed)
print(f"\nدقة النموذج بعد إعادة الوزن: {accuracy_reweighed:.4f}")
print("\nتقرير التصنيف بعد إعادة الوزن:")
print(classification_report(y_test, y_pred_reweighed))

# 6. تقييم الإنصاف للنموذج المعاد تدريبه
dataset_test_pred_reweighed = dataset_test.copy()
dataset_test_pred_reweighed.labels = y_pred_reweighed.reshape(-1, 1)

metric_reweighed_test = ClassificationMetric(
    dataset_test,
    dataset_test_pred_reweighed,
    unprivileged_groups=unprivileged_groups,
    privileged_groups=privileged_groups
)

print("\n--- مقاييس الإنصاف للنموذج بعد إعادة الوزن (الجنس) ---")
print(f"فرق التأثير المتباين (Disparate Impact): {metric_reweighed_test.disparate_impact():.4f}")
print(f"فرق التكافؤ الإحصائي (Statistical Parity Difference): {metric_reweighed_test.statistical_parity_difference():.4f}")
print(f"فرق تكافؤ الفرص (Equal Opportunity Difference): {metric_reweighed_test.equal_opportunity_difference():.4f}")
print(f"فرق معدل الإيجابيات الكاذبة (False Positive Rate Difference): {metric_reweighed_test.false_positive_rate_difference():.4f}")

# مقارنة النتائج:
print("\n--- مقارنة مقاييس الإنصاف (الجنس) ---")
print(f"قبل إعادة الوزن - فرق التكافؤ الإحصائي: {metric_orig_test.statistical_parity_difference():.4f}")
print(f"بعد إعادة الوزن - فرق التكافؤ الإحصائي: {metric_reweighed_test.statistical_parity_difference():.4f}")

print(f"قبل إعادة الوزن - فرق تكافؤ الفرص: {metric_orig_test.equal_opportunity_difference():.4f}")
print(f"بعد إعادة الوزن - فرق تكافؤ الفرص: {metric_reweighed_test.equal_opportunity_difference():.4f}")

print("\n--- مقارنة الدقة ---")
print(f"دقة النموذج الأساسي: {accuracy:.4f}")
print(f"دقة النموذج بعد إعادة الوزن: {accuracy_reweighed:.4f}")

# هل تحسنت مقاييس الإنصاف؟ هل تأثرت الدقة؟
```

## 4. المفاضلات بين الدقة والإنصاف (Trade-offs between Accuracy and Fairness)

غالبًا ما يكون هناك مفاضلة معقدة بين دقة النموذج (Accuracy) ومقاييس الإنصاف (Fairness). تحسين الإنصاف قد يؤدي أحيانًا إلى انخفاض طفيف في الدقة الكلية للنموذج، والعكس صحيح. هذا ليس دائمًا هو الحال، ولكن من المهم فهم هذه الديناميكية.

### نقاط رئيسية للمناقشة:
*   **لماذا تحدث المفاضلة؟** غالبًا ما تكون البيانات التاريخية التي نتدرب عليها متحيزة، والنموذج الذي يتعلم هذه التحيّزات قد يكون "أكثر دقة" في التنبؤ بالواقع المتحيز. عندما نحاول إزالة التحيّز، فإننا نطلب من النموذج أن يتجاهل بعض الأنماط الموجودة في البيانات، مما قد يؤثر على قدرته على التنبؤ بدقة.
*   **أهمية السياق:** تعتمد أهمية الدقة مقابل الإنصاف بشكل كبير على سياق التطبيق. في بعض التطبيقات (مثل أنظمة العدالة الجنائية أو الرعاية الصحية)، قد تكون الأولوية للإنصاف حتى لو كان ذلك يعني تضحية طفيفة بالدقة. في تطبيقات أخرى، قد تكون الدقة هي الأهم.
*   **الهدف هو التوازن:** الهدف ليس دائمًا تحقيق إنصاف مثالي أو دقة مثالية، بل إيجاد توازن مقبول بينهما يتوافق مع القيم الأخلاقية ومتطلبات العمل.

في القسم السابق، رأينا كيف أن إعادة الوزن قد حسنت بعض مقاييس الإنصاف، وقد يكون لها تأثير طفيف على الدقة. هذا هو مثال عملي على هذه المفاضلة.
