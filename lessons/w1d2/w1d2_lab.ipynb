<div dir="rtl">## النمذجة العاطفية: تجربة الضياع
قبل أن نبدأ، فكر في لحظة فقدت فيها ملفاً مهماً أو صورة عزيزة بسبب سوء التنظيم. كيف شعرت؟ هذا الشعور هو المحرك الأساسي لأي مهندس بيانات لضمان الأمان والتنظيم.</div>

```
# محاكاة مستودع البيانات البسيط باستخدام القواميس (Dictionaries)
# كل مفتاح يمثل 'رفاً' في الخزانة، والقيمة هي قائمة البيانات

warehouse = {
    "صور_شخصية": ["صورة_التخرج.jpg", "صورة_العائلة.png"],
    "مستندات_عمل": ["عقد_العمل.pdf", "تقرير_مشروع.docx"]
}

def add_item(category, item):
    if category in warehouse:
        warehouse[category].append(item)
        print(f"تم إضافة {item} إلى {category} بنجاح!")
    else:
        print("خطأ: هذا الرف غير موجود!")

add_item("صور_شخصية", "رحلة_الصيف.jpg")
```

<div dir="rtl">## لعبة أمين المستودع (التلعيب)
الآن حان دورك! لديك مجموعة من البيانات المبعثرة، وعليك تصنيفها في المستودع الصحيح بأسرع وقت ممكن.</div>

```
# تمرين: أكمل الكود لتصنيف البيانات المبعثرة

scattered_data = ["فاتورة_كهرباء.pdf", "صورة_القطة.jpg", "بحث_جامعي.docx"]

# TODO: قم بإنشاء دالة تقوم بتوزيع البيانات بناءً على امتداد الملف (pdf, jpg, docx)
# إذا كان الملف ينتهي بـ .jpg ضعه في 'صور_شخصية'
# إذا كان ينتهي بـ .pdf أو .docx ضعه في 'مستندات_عمل'

print("ابدأ الحل هنا...")
```

```
# الحل النموذجي لتمرين أمين المستودع

warehouse = {"صور_شخصية": [], "مستندات_عمل": []}
scattered_data = ["فاتورة_كهرباء.pdf", "صورة_القطة.jpg", "بحث_جامعي.docx"]

for item in scattered_data:
    if item.endswith('.jpg'):
        warehouse['صور_شخصية'].append(item)
    elif item.endswith(('.pdf', '.docx')):
        warehouse['مستندات_عمل'].append(item)

print("المستودع بعد التنظيم:", warehouse)
```
