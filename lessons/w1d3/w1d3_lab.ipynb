<div dir="rtl">## القصة: نظام مراقبة مستشعرات التمر
تخيّل أن لدينا مخزناً ضخماً للتمور، وكل صندوق فيه مستشعر حرارة. إذا ارتفعت الحرارة، يجب أن نطلق تنبيهاً فوراً قبل أن يفسد التمر. 

في عالم التقنية، نستخدم **Kafka** كـ "ساعي بريد" سريع جداً ينقل الرسائل، و **Spark Streaming** كـ "عامل ذكي" يحلل هذه الرسائل فور وصولها.</div>

```
# محاكاة تدفق البيانات باستخدام بايثون
import time
import random

# محاكاة مستشعر حرارة يرسل بيانات كل ثانية
def sensor_stream():
    while True:
        temp = random.uniform(20.0, 50.0)
        yield {"sensor_id": "A1", "temp": temp, "timestamp": time.time()}
        time.sleep(1) # تأخير بسيط لمحاكاة الوقت الحقيقي
```

```
# الحل النموذجي
def process_data(data):
    if data['temp'] > 45:
        print(f"⚠️ تنبيه! حرارة عالية في المستشعر {data['sensor_id']}: {data['temp']:.2f} درجة")
    else:
        print(f"✅ الحرارة طبيعية: {data['temp']:.2f}")

stream = sensor_stream()
for _ in range(5):
    process_data(next(stream))
```
