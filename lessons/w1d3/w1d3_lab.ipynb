```
# استيراد المكتبات الأساسية لمحاكاة معالجة البيانات
import pandas as pd
import numpy as np

# إنشاء مجموعة بيانات تجريبية لمحاكاة خط إنتاج بيانات
data = {'user_id': [1, 2, 3, 4], 'sensitive_info': ['email@test.com', 'phone_123', 'email@demo.com', 'phone_456']}
df = pd.DataFrame(data)
print("بيانات خط الإنتاج الخام:")
print(df)
```

```
# الحل النموذجي للتمرين
def mask_email(email):
    if '@' in email:
        user, domain = email.split('@')
        return f"{user[0]}***@{domain}"
    return email

df['sensitive_info'] = df['sensitive_info'].apply(mask_email)
print("البيانات بعد تطبيق سياسة الحوكمة:")
print(df)
```

```
# محاكاة بيانات مخترقة
malicious_data = pd.DataFrame({'user_id': [999], 'sensitive_info': ['SQL_INJECTION_ATTACK']})
df = pd.concat([df, malicious_data])

# فحص الجودة: البحث عن قيم غير متوقعة
print("فحص جودة البيانات لاكتشاف الثغرات:")
print(df[df['sensitive_info'].str.contains('SQL')])
```
