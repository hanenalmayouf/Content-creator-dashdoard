<div dir="rtl"># معمل جودة البيانات وهندسة الميزات (Feature Engineering)

أهلاً بك يا بطل في معملنا اليوم. تخيّل أنك طباخ ماهر؛ هل يمكنك صنع وجبة فاخرة من خضروات تالفة؟ بالطبع لا. في عالم الذكاء الاصطناعي، البيانات هي 'المقادير'، وإذا كانت غير صالحة، فسيخرج النموذج بنتائج سيئة. اليوم سنتعلم كيف نكون 'مفتشي جودة' للبيانات.</div>

```
# استيراد المكتبات الأساسية
import pandas as pd
import numpy as np

# إنشاء مجموعة بيانات تجريبية تحتوي على أخطاء (بيانات مفقودة وقيم شاذة)
data = {'سعر_المنتج': [100, 120, 150, 10000, np.nan, 130], 'الكمية': [5, 10, 8, 7, 2, -1]}
df = pd.DataFrame(data)
print("نظرة على البيانات الخام:")
print(df)
```

<div dir="rtl">## 1. كشف الشذوذ (Anomaly Detection)

القيم الشاذة مثل '10000' كسعر لمنتج بسيط، أو كمية بالسالب، هي أخطاء يجب تنظيفها. سنقوم الآن ببناء دالة بسيطة لاكتشاف هذه القيم.</div>

```
# دالة بسيطة لكشف القيم غير المنطقية
def detect_anomalies(df):
    # كشف الأسعار غير المنطقية (أكبر من 500)
    anomalies = df[(df['سعر_المنتج'] > 500) | (df['الكمية'] < 0)]
    return anomalies

print("القيم الشاذة المكتشفة:")
print(detect_anomalies(df))
```

<div dir="rtl">## 2. تمرين عملي: أتمتة فحص الجودة

حان دورك الآن! نريد إضافة فحص للقيم المفقودة (NaN). أكمل الكود التالي لاكتشاف الصفوف التي تحتوي على قيم مفقودة.</div>

```
# TODO: اكتب كود لاكتشاف الصفوف التي تحتوي على قيم مفقودة (NaN) في أي عمود
# missing_data = ...
# print(missing_data)
```

```
# الحل النموذجي:
missing_data = df[df.isnull().any(axis=1)]
print("الصفوف التي تحتوي على بيانات مفقودة:")
print(missing_data)
```

<div dir="rtl">## 3. هندسة الميزات (Feature Engineering)

هندسة الميزات هي فن استخراج معلومات جديدة من البيانات الموجودة. مثلاً، بدلاً من 'السعر' و'الكمية'، يمكننا إنشاء ميزة 'إجمالي_القيمة' (السعر * الكمية).</div>

```
# تنظيف البيانات أولاً (ملء القيم المفقودة بمتوسط العمود)
df['سعر_المنتج'] = df['سعر_المنتج'].fillna(df['سعر_المنتج'].mean())

# إنشاء ميزة جديدة (Feature Engineering)
df['إجمالي_القيمة'] = df['سعر_المنتج'] * df['الكمية']
print("البيانات بعد هندسة الميزات:")
print(df)
```
