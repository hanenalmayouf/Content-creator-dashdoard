<div dir="rtl"># معمل: توسيع البيانات وتكامل الذكاء الاصطناعي (MLOps)

أهلاً بكم في هذا المعمل التفاعلي. سنركز اليوم على بناء خطوط أنابيب بيانات (Data Pipelines) قوية، وضمان جودة البيانات، ونشر النماذج في بيئات الإنتاج.</div>

```
# استيراد المكتبات الأساسية لمعالجة البيانات والتحقق منها
import pandas as pd
import numpy as np
from sklearn.ensemble import RandomForestRegressor
from sklearn.model_selection import train_test_split

# إنشاء مجموعة بيانات تجريبية لمحاكاة تدفق البيانات
data = pd.DataFrame({'feature_1': np.random.rand(100), 'target': np.random.rand(100) * 10})
print("تم تحميل البيانات بنجاح.")
```

<div dir="rtl">## التحقق من جودة البيانات (Data Quality)
قبل إدخال البيانات للنموذج، يجب التأكد من خلوها من القيم المفقودة أو الشاذة إحصائياً.</div>

```
# دالة للتحقق من جودة البيانات
def validate_data(df):
    # التحقق من وجود قيم مفقودة
    if df.isnull().sum().sum() > 0:
        return False, "تحذير: توجد قيم مفقودة!"
    return True, "البيانات سليمة."

status, message = validate_data(data)
print(message)
```

<div dir="rtl">## تمرين عملي: أتمتة خط البيانات
قم بإكمال الكود التالي لتدريب نموذج تنبئي بسيط.</div>

```
# TODO: قم بتقسيم البيانات إلى تدريب واختبار ثم تدريب نموذج RandomForest
X = data[['feature_1']]
y = data['target']

# اكتب الكود هنا:
# X_train, X_test, y_train, y_test = ...
# model = ...
# model.fit(...)
```

```
# الحل النموذجي:
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)
model = RandomForestRegressor()
model.fit(X_train, y_train)
print("تم تدريب النموذج بنجاح.")
```
