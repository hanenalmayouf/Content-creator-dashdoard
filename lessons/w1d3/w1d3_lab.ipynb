## 1. مقدمة إلى السلسلة الفكرية (Chain-of-Thought - CoT)

السلسلة الفكرية (CoT) هي تقنية هندسة أوامر تطلب من النموذج اللغوي أن يولد سلسلة من خطوات التفكير الوسيطة قبل إعطاء الإجابة النهائية. بدلاً من طلب الإجابة مباشرة، نطلب من النموذج أن 'يفكر بصوت عالٍ' أو 'يشرح خطواته'. هذا يحاكي طريقة تفكير البشر عند حل المشكلات المعقدة، حيث نقوم بتقسيم المشكلة إلى خطوات أصغر وأكثر قابلية للإدارة.

**لماذا هي قوية؟**

*   **تحسين الدقة:** تساعد النماذج على معالجة المشكلات المعقدة التي تتطلب التفكير المنطقي متعدد الخطوات.
*   **الشفافية:** تجعل عملية اتخاذ القرار للنموذج أكثر وضوحاً، مما يسهل فهم الأخطاء وتصحيحها.
*   **المرونة:** يمكن تطبيقها على مجموعة واسعة من المهام، من حل المسائل الرياضية إلى التفكير المنطقي والبرمجة.

هناك طريقتان رئيسيتان لتطبيق CoT:

*   **Zero-shot CoT:** حيث لا نقدم أي أمثلة، فقط نطلب من النموذج أن يفكر خطوة بخطوة.
*   **Few-shot CoT:** حيث نقدم للنموذج بضعة أمثلة توضح كيفية التفكير خطوة بخطوة.
</div>

<div dir="rtl">
### 1.1. Zero-shot CoT: التفكير التلقائي

في هذه الطريقة، لا نقدم للنموذج أي أمثلة على كيفية حل المشكلة. بدلاً من ذلك، نضيف ببساطة عبارة مثل "لنحل هذه المسألة خطوة بخطوة." أو "فكر بصوت عالٍ." إلى الأمر. هذا يحفز النموذج على توليد سلسلة من الأفكار الوسيطة تلقائياً.
</div>

````
# مثال على Zero-shot CoT
print("\n--- مثال على Zero-shot CoT ---")
problem_zero_shot = "ما هو حاصل ضرب 12 في 9؟ لنحل هذه المسألة خطوة بخطوة."
response_zero_shot = mock_llm_response(problem_zero_shot)
print(f"الأمر:\n{problem_zero_shot}\n")
print(f"استجابة النموذج:\n{response_zero_shot}")

print("\n--- مثال آخر على Zero-shot CoT لمشكلة أكثر تعقيداً ---")
problem_zero_shot_complex = "صمم خوارزمية لفرز قائمة من الأعداد. فكر بصوت عالٍ واشرح خطواتك بالتفصيل."
response_zero_shot_complex = mock_llm_response(problem_zero_shot_complex)
print(f"الأمر:\n{problem_zero_shot_complex}\n")
print(f"استجابة النموذج:\n{response_zero_shot_complex}")
````

<div dir="rtl">
### 1.2. Few-shot CoT: التعلم من الأمثلة

في Few-shot CoT، نقدم للنموذج بضعة أمثلة (عادةً 2-5 أمثلة) توضح كيفية حل المشكلة خطوة بخطوة. هذه الأمثلة تساعد النموذج على فهم النمط المطلوب للتفكير الاستنتاجي وتطبيق نفس المنطق على المشكلة الجديدة. هذه الطريقة تكون فعالة بشكل خاص للمهام التي تتطلب تنسيقاً أو هيكلاً محدداً في التفكير.
</div>

````
# مثال على Few-shot CoT
print("\n--- مثال على Few-shot CoT ---")

# بناء الأمر مع أمثلة CoT
few_shot_prompt = """
المثال 1:
السؤال: إذا كان لديك 3 تفاحات وأضفت 2، كم تفاحة لديك؟
التفكير: لدي 3 تفاحات في البداية. أضفت 2 تفاحة. إذن، 3 + 2 = 5. 
الجواب: 5

المثال 2:
السؤال: إذا كان لديك 10 أقلام وأعطيت 4 لأصدقائك، كم قلم بقي لديك؟
التفكير: لدي 10 أقلام في البداية. أعطيت 4 أقلام. إذن، 10 - 4 = 6.
الجواب: 6

المثال 3:
السؤال: إذا كان لديك 5 تفاحات وأكلت 2، كم تفاحة بقيت لديك؟
التفكير: لنحل هذه المشكلة خطوة بخطوة.
"""

response_few_shot = mock_llm_response(few_shot_prompt)
print(f"الأمر:\n{few_shot_prompt}\n")
print(f"استجابة النموذج:\n{response_few_shot}")
````

## 2. البرمجة بالتعليمات (Instruction-based Programming)

البرمجة بالتعليمات هي فن صياغة الأوامر بطريقة تجعل النموذج اللغوي يتصرف كـ 'مترجم' أو 'منفذ' لتعليمات محددة. بدلاً من مجرد طرح سؤال، فإننا نقدم للنموذج مجموعة من القواعد، الخطوات، أو الشروط التي يجب أن يتبعها لإنتاج مخرجات معينة. هذا يشبه كتابة كود برمجي، ولكن باللغة الطبيعية.

**كيف تعمل؟**

*   **تحديد الدور:** يمكن أن تطلب من النموذج أن يتصرف كخبير في مجال معين (مثلاً: "أنت خبير في الأمن السيبراني...").
*   **تحديد المهام:** تقسيم المهمة إلى خطوات واضحة ومحددة ("أولاً، قم بكذا. ثانياً، قم بكذا...").
*   **تحديد القيود:** وضع شروط على الإجابة (مثلاً: "يجب أن تكون الإجابة في 3 نقاط فقط.").
*   **تحديد التنسيق:** طلب تنسيق معين للمخرجات (مثلاً: "أرجع الإجابة بصيغة JSON.").

عند دمجها مع CoT، تصبح البرمجة بالتعليمات أداة قوية لتوجيه النموذج ليس فقط للتفكير، بل للتفكير بطريقة منظمة وموجهة نحو هدف محدد.
</div>

````
# مثال على البرمجة بالتعليمات مع CoT
print("\n--- مثال على البرمجة بالتعليمات مع CoT ---")

instruction_prompt = """
أنت مطور بايثون محترف. مهمتك هي كتابة دالة بايثون تقوم بقلب سلسلة نصية.
قبل كتابة الكود، فكر بصوت عالٍ واشرح خطواتك المنطقية لتصميم الدالة، مع الأخذ في الاعتبار أفضل الممارسات.
ثم قدم الكود النهائي مع أمثلة للاستخدام.
"""

response_instruction = mock_llm_response(instruction_prompt)
print(f"الأمر:\n{instruction_prompt}\n")
print(f"استجابة النموذج:\n{response_instruction}")
````

## 3. الآليات الكامنة وراء التفكير الاستنتاجي (Reasoning)

عندما نطبق تقنيات CoT، فإننا في الواقع نستغل قدرة نماذج اللغات على محاكاة التفكير الاستنتاجي. النماذج لا 'تفكر' بالمعنى البشري، ولكنها تتعلم من كميات هائلة من البيانات النصية كيف تبدو عملية التفكير المنطقي. عندما نطلب منها "التفكير خطوة بخطوة"، فإننا نوجهها لتوليد تسلسل من الرموز (tokens) التي تتوافق مع نمط التفكير المنطقي الذي رأته في بيانات التدريب.

**كيف تساعد CoT في Reasoning؟**

*   **تفكيك المشكلة:** تجبر النموذج على تقسيم المشكلة الكبيرة إلى أجزاء أصغر وأكثر قابلية للإدارة.
*   **توليد السياق:** كل خطوة وسيطة تولد سياقاً إضافياً للخطوات التالية، مما يقلل من الغموض ويحسن من دقة التنبؤ.
*   **تصحيح الأخطاء الذاتي:** في بعض الحالات، يمكن للنموذج اكتشاف وتصحيح الأخطاء في خطواته الوسيطة قبل الوصول إلى الإجابة النهائية.
*   **الاستفادة من المعرفة الضمنية:** تساعد النموذج على استدعاء وتطبيق المعرفة الضمنية التي اكتسبها أثناء التدريب بطريقة منظمة.

هذه القدرة على محاكاة التفكير هي ما يجعل CoT أداة قوية جداً في التعامل مع المهام المعقدة التي تتجاوز مجرد استرجاع المعلومات.
