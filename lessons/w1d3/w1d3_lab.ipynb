### 2.2. التعامل مع القيم المفقودة (Missing Values)

القيم المفقودة (عادة ما تظهر كـ `NaN` في Pandas) هي مشكلة شائعة. يمكننا التعامل معها بعدة طرق:

*   **إزالتها**: حذف الصفوف أو الأعمدة التي تحتوي على قيم مفقودة (يجب الحذر لعدم فقدان الكثير من البيانات).
*   **ملؤها**: استبدال القيم المفقودة بقيمة معينة (مثل المتوسط، الوسيط، المنوال، أو قيمة ثابتة).
</div>

````
# لنرى عدد القيم المفقودة في كل عمود
print("عدد القيم المفقودة قبل المعالجة:")
print(df.isnull().sum())

# مثال: ملء القيم المفقودة في عمود 'تقييم_العميل' بالمتوسط
# أولاً، نحسب متوسط التقييمات الموجودة
متوسط_التقييم = df['تقييم_العميل'].mean()
print(f"\nمتوسط تقييم العميل: {متوسط_التقييم:.2f}")

# ثم نملأ القيم المفقودة بهذا المتوسط
df['تقييم_العميل'].fillna(متوسط_التقييم, inplace=True)

# لنرى الآن عدد القيم المفقودة بعد المعالجة
print("\nعدد القيم المفقودة بعد ملء عمود 'تقييم_العميل':")
print(df.isnull().sum())

print("\nنظرة على البيانات بعد ملء القيم المفقودة في 'تقييم_العميل':")
print(df.head())
````

<div dir="rtl">
### 2.3. إزالة البيانات المكررة (Duplicate Data)

البيانات المكررة يمكن أن تؤدي إلى تحليلات خاطئة. من المهم تحديدها وإزالتها.
</div>

````
# لنرى الصفوف المكررة بناءً على جميع الأعمدة
print("عدد الصفوف المكررة قبل الإزالة:", df.duplicated().sum())

# إزالة الصفوف المكررة والاحتفاظ بالظهور الأول فقط
df_cleaned = df.drop_duplicates()

print("عدد الصفوف بعد إزالة التكرارات:", len(df_cleaned))
print("\nنظرة على البيانات بعد إزالة التكرارات:")
print(df_cleaned.head())
````

<div dir="rtl">
### 2.4. توحيد التنسيقات وتصحيح الأخطاء (Standardizing Formats & Correcting Errors)

قد تكون هناك اختلافات في التنسيق لنفس البيانات (مثل 'الرياض' و 'الرياض '). أو أخطاء إملائية. يمكننا توحيدها.
</div>

````
# لنفترض أن هناك خطأ إملائي في عمود 'المنطقة' أو اختلاف في حالة الأحرف (لو كانت بالإنجليزية)
# سنقوم بتوحيد تنسيق عمود 'المنطقة' (إزالة المسافات الزائدة مثلاً)
df_cleaned['المنطقة'] = df_cleaned['المنطقة'].str.strip()

# لنفترض أن 'لعبة سيارة' قد تظهر أحيانًا كـ 'سيارة لعبة'. يمكننا توحيدها.
# هنا سنقوم بتحويل كل أسماء المنتجات إلى أحرف صغيرة (لو كانت بالإنجليزية) أو التأكد من التوحيد.
# في حالتنا العربية، يمكننا البحث عن أخطاء إملائية وتصحيحها يدوياً أو باستخدام قاموس.
# مثال: توحيد 'لعبة دمية' و 'دمية لعبة' (إذا كانت موجودة)
df_cleaned['اسم_المنتج'] = df_cleaned['اسم_المنتج'].replace({'دمية لعبة': 'لعبة دمية'})

print("القيم الفريدة في عمود 'المنطقة' بعد التوحيد:")
print(df_cleaned['المنطقة'].unique())

print("\nالقيم الفريدة في عمود 'اسم_المنتج' بعد التوحيد:")
print(df_cleaned['اسم_المنتج'].unique())
````

### 2.5. تحويل أنواع البيانات (Data Type Conversion)

التأكد من أن الأعمدة لها أنواع البيانات الصحيحة أمر بالغ الأهمية للتحليل. على سبيل المثال، يجب أن يكون 'تاريخ_البيع' من نوع تاريخ وليس نصًا.
</div>

````
# لنرى أنواع البيانات الحالية
print("أنواع البيانات قبل التحويل:")
print(df_cleaned.dtypes)

# تحويل عمود 'تاريخ_البيع' إلى نوع تاريخ (datetime)
df_cleaned['تاريخ_البيع'] = pd.to_datetime(df_cleaned['تاريخ_البيع'])

print("\nأنواع البيانات بعد تحويل 'تاريخ_البيع':")
print(df_cleaned.dtypes)

# يمكننا الآن استخراج معلومات إضافية من التاريخ، مثل الشهر أو اليوم من الأسبوع
df_cleaned['الشهر'] = df_cleaned['تاريخ_البيع'].dt.month
df_cleaned['اليوم_من_الأسبوع'] = df_cleaned['تاريخ_البيع'].dt.day_name()

print("\nنظرة على البيانات بعد إضافة عمود الشهر واليوم من الأسبوع:")
print(df_cleaned.head())
````

## 3. مفهوم مسار البيانات (Data Pipeline)

تخيل أنك تقوم ببناء مصنع. المواد الخام تدخل من جهة، وتمر عبر عدة محطات (تنظيف، تجميع، طلاء، فحص جودة)، وفي النهاية يخرج المنتج النهائي الجاهز. هذا هو بالضبط ما يفعله مسار البيانات.

**مسار البيانات (Data Pipeline)** هو سلسلة من الخطوات المتتالية التي تمر بها البيانات من لحظة جمعها (المصدر الخام) حتى تصبح جاهزة للاستخدام النهائي (التحليل، التقارير، نماذج الذكاء الاصطناعي).

**مكونات مسار البيانات البسيط:**

1.  **الجمع (Ingestion)**: الحصول على البيانات من مصادرها (قواعد بيانات، ملفات، APIs).
2.  **التخزين (Storage)**: حفظ البيانات في مكان ما (قاعدة بيانات، مستودع بيانات).
3.  **المعالجة/التحويل (Processing/Transformation)**: تطبيق عمليات التنظيف، التوحيد، هندسة الميزات (كما فعلنا للتو).
4.  **التحميل (Loading)**: نقل البيانات المعالجة إلى وجهتها النهائية (مثل مستودع بيانات للتحليل).

**الاستراتيجية التعليمية: التعلم التشاركي والتعاوني - نشاط 'بناء مسار بيانات مبسط'**

في مجموعات، يتم تزويد الطلاب بسيناريو (مثل: بيانات مبيعات متجر ألعاب) ويُطلب منهم تصميم 'مسار' مبسط يوضح الخطوات التي تمر بها البيانات من لحظة جمعها حتى تصبح جاهزة لإعداد تقرير مبيعات أسبوعي، باستخدام رسومات بسيطة أو بطاقات. يتم عرض المسارات ومناقشة أوجه التشابه والاختلاف.

**مثال على مسار بيانات بسيط في الكود:**
يمكننا تجميع الخطوات التي قمنا بها في دالة واحدة أو سلسلة من العمليات.
</div>

````
def clean_and_transform_sales_data(df_raw):
    """
    دالة لتنظيف وتحويل بيانات المبيعات الخام.
    تأخذ DataFrame خام وتعيد DataFrame نظيفًا وجاهزًا للتحليل.
    """
    # 1. التعامل مع القيم المفقودة: ملء تقييم العميل بالمتوسط
    df_processed = df_raw.copy() # نعمل على نسخة لتجنب تعديل البيانات الأصلية
    متوسط_التقييم = df_processed['تقييم_العميل'].mean()
    df_processed['تقييم_العميل'].fillna(متوسط_التقييم, inplace=True)

    # 2. إزالة البيانات المكررة
    df_processed = df_processed.drop_duplicates()

    # 3. توحيد التنسيقات (إزالة المسافات الزائدة)
    df_processed['المنطقة'] = df_processed['المنطقة'].str.strip()

    # 4. تحويل أنواع البيانات (تاريخ البيع)
    df_processed['تاريخ_البيع'] = pd.to_datetime(df_processed['تاريخ_البيع'])

    # 5. هندسة الميزات: حساب إجمالي الإيرادات
    df_processed['إجمالي_الإيرادات'] = df_processed['السعر_الوحدة'] * df_processed['الكمية_المباعة']
    
    # 6. إضافة أعمدة إضافية من التاريخ
    df_processed['الشهر'] = df_processed['تاريخ_البيع'].dt.month
    df_processed['اليوم_من_الأسبوع'] = df_processed['تاريخ_البيع'].dt.day_name()

    return df_processed

# لنقم بإنشاء بيانات خام جديدة لتجربة المسار
new_raw_data = {
    'معرف_المنتج': [201, 202, 201, 203, 204],
    'اسم_المنتج': ['لعبة روبوت', 'لعبة بازل', 'لعبة روبوت', 'لعبة سيارة', 'لعبة طائرة'],
    'الفئة': ['روبوتات', 'بازل', 'روبوتات', 'سيارات', 'طائرات'],
    'السعر_الوحدة': [110.0, 35.0, 110.0, 55.0, 95.0],
    'الكمية_المباعة': [1, 2, 1, 1, 1],
    'تاريخ_البيع': ['2023-02-01', '2023-02-02', '2023-02-01', '2023-02-03', '2023-02-04'],
    'المنطقة': ['جدة ', 'الرياض', 'جدة', 'الدمام', 'الرياض'],
    'تقييم_العميل': [5, np.nan, 5, 4, 3]
}
new_df_raw = pd.DataFrame(new_raw_data)

print("\nالبيانات الخام الجديدة:")
print(new_df_raw)

# تطبيق مسار البيانات
df_final = clean_and_transform_sales_data(new_df_raw)

print("\nالبيانات بعد تطبيق مسار التنظيف والتحويل:")
print(df_final)
print("\nمعلومات عن البيانات النهائية:")
df_final.info()
````

## 4. تحدي تنظيف البيانات (تمرين عملي)

لديك مجموعة بيانات جديدة تحتوي على بعض المشاكل. مهمتك هي تطبيق ما تعلمته لتنظيفها وجعلها جاهزة للتحليل.

**البيانات الخام:**

| 1           | أحمد        | الرياض  | 30    | 2022-01-15    | 150.5       |
| 2           | سارة        | جدة     | 25    | 2022-02-20    | 200.0       |
| 3           | علي         | الرياض  | 35    | 2022-01-15    | 150.5       |
| 1           | أحمد        | الرياض  | 30    | 2022-01-15    | 150.5       |
| 4           | ليلى        | الدمام  | NaN   | 2022-03-10    | 75.25       |
| 5           | خالد        | جدة     | 40    | 2022-04-01    | 300.0       |

**المشاكل التي يجب حلها:**

1.  **القيم المكررة**: يوجد صفوف مكررة بالكامل.
2.  **القيم المفقودة**: يوجد قيم مفقودة في عمودي 'العمر' و 'قيمة_الشراء'.
    *   املأ القيم المفقودة في 'العمر' بمتوسط الأعمار الموجودة.
    *   املأ القيم المفقودة في 'قيمة_الشراء' بالقيمة 0.
3.  **توحيد التنسيقات**: عمود 'المدينة' قد يحتوي على مسافات زائدة.
4.  **تحويل أنواع البيانات**: عمود 'تاريخ_التسجيل' يجب أن يكون من نوع تاريخ.
5.  **هندسة الميزات**: أنشئ عمودًا جديدًا باسم 'العمر_بالسنوات' (إذا كان العمر بالأشهر مثلاً، لكن هنا هو بالسنوات بالفعل، لذا يمكننا إنشاء عمود 'فئة_العمر' بناءً على العمر: 'صغير' (<30)، 'متوسط' (30-40)، 'كبير' (>40)).
