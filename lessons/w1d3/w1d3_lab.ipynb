<div dir="rtl"># معمل: توسيع البيانات وتكامل الذكاء الاصطناعي (MLOps)

أهلاً بكم في هذا المعمل التفاعلي. سنقوم باستكشاف كيفية بناء خطوط أنابيب بيانات (Data Pipelines) قوية، وضمان جودة البيانات، ونشر النماذج في بيئات الإنتاج.</div>

```
# استيراد المكتبات الأساسية لمعالجة البيانات والتحقق منها
import pandas as pd
import numpy as np
from sklearn.ensemble import RandomForestRegressor

# إنشاء مجموعة بيانات تجريبية لمحاكاة تدفق البيانات
data = pd.DataFrame({'feature_1': np.random.rand(100), 'target': np.random.rand(100) * 10})
print("تم تحميل البيانات بنجاح.")
```

<div dir="rtl">## التحقق من صحة البيانات (Data Validation)
قبل إدخال البيانات إلى النموذج، يجب التأكد من خلوها من القيم الشاذة أو المفقودة لضمان جودة التنبؤات.</div>

```
# دالة للتحقق من جودة البيانات
def validate_data(df):
    # التحقق من وجود قيم مفقودة
    if df.isnull().sum().sum() > 0:
        return False, "تحذير: توجد قيم مفقودة في البيانات!"
    return True, "البيانات سليمة."

status, message = validate_data(data)
print(message)
```

```
# الحل النموذجي:
X = data[['feature_1']]
y = data['target']
model = RandomForestRegressor()
model.fit(X, y)
print("تم تدريب النموذج بنجاح باستخدام الحل النموذجي.")
```
