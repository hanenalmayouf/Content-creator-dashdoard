<div dir="rtl"># مشروع البيانات المصغر: اتخاذ القرار
أهلاً بكم في هذا المعمل التفاعلي. سنقوم اليوم بدور 'خبير البيانات الجامعي' لحل مشكلة واقعية: **تحسين تجربة الازدحام في كافتيريا الجامعة.**</div>

```
# استيراد المكتبات اللازمة لتحليل البيانات
import pandas as pd
import matplotlib.pyplot as plt

# إنشاء بيانات تجريبية حول أوقات الازدحام (عدد الطلاب في الساعة)
data = {'الساعة': ['8:00', '10:00', '12:00', '14:00', '16:00'],
        'عدد_الطلاب': [50, 120, 300, 200, 80]}

df = pd.DataFrame(data)
print("تم تحميل بيانات الازدحام بنجاح:")
print(df)
```

<div dir="rtl">## الخطوة الأولى: تحليل البيانات بصرياً
سنقوم الآن برسم مخطط بياني يوضح ذروة الازدحام لاتخاذ قرار مبني على الأرقام.</div>

```
# رسم مخطط بياني يوضح أوقات الازدحام
plt.figure(figsize=(8, 5))
plt.plot(df['الساعة'], df['عدد_الطلاب'], marker='o', color='b')
plt.title('معدل الازدحام في الكافتيريا')
plt.xlabel('الوقت')
plt.ylabel('عدد الطلاب')
plt.grid(True)
plt.show()
```

<div dir="rtl">## تمرين عملي: تحدي خبير البيانات
بناءً على الرسم البياني، نلاحظ أن الساعة 12:00 هي وقت الذروة. مهمتك الآن هي كتابة كود يقوم بحساب 'متوسط عدد الطلاب' في اليوم الدراسي.</div>

```
# TODO: قم بحساب متوسط عدد الطلاب باستخدام دالة mean() من مكتبة pandas
# متوسط_الطلاب = ...
# print(f"متوسط عدد الطلاب هو: {متوسط_الطلاب}")
```

```
# الحل النموذجي:
متوسط_الطلاب = df['عدد_الطلاب'].mean()
print(f"متوسط عدد الطلاب هو: {متوسط_الطلاب}")
```

<div dir="rtl">## المشروع التطبيقي (Project Challenge)
الآن، بصفتك خبير بيانات، قدم توصية لإدارة الجامعة بناءً على النتائج:
1. إذا كان متوسط الازدحام يتجاوز 150 طالباً، اقترح 'فتح نافذة تقديم إضافية'.
2. اكتب كوداً برمجياً بسيطاً (if statement) يطبع التوصية بناءً على المتغير 'متوسط_الطلاب'.</div>

```
# كود اتخاذ القرار
if متوسط_الطلاب > 150:
    print("التوصية: يجب فتح نافذة تقديم إضافية خلال وقت الذروة.")
else:
    print("التوصية: الوضع الحالي مقبول، لا حاجة لتغييرات فورية.")
```
