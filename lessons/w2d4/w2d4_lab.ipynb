## 1. توليد الأوامر الذاتي (Automatic Prompt Generation)

في هذا القسم، سنتعلم كيف يمكن للأنظمة أن تُنشئ أوامرها الخاصة ديناميكيًا بدلاً من الاعتماد على أوامر ثابتة ومحددة مسبقًا. هذا يفتح الباب أمام مرونة هائلة وقدرة على التكيف مع سيناريوهات مختلفة.
</div>

```
# <div dir="rtl"> مثال بسيط: توليد أمر بناءً على مدخلات المستخدم </div>

def generate_simple_prompt(user_query, context="عام"):
    """
    <div dir="rtl">
    تولد أمرًا بسيطًا بناءً على استعلام المستخدم والسياق.
    </div>
    """
    if context == "برمجة":
        return f"اكتب كود بايثون لـ: {user_query}. تأكد من تضمين التعليقات والشرح."
    elif context == "وصف منتج":
        return f"اكتب وصفًا تسويقيًا جذابًا لمنتج: {user_query}. ركز على الفوائد الرئيسية."
    else:
        return f"أجب على السؤال التالي: {user_query}."

# <div dir="rtl"> تجربة توليد الأوامر </div>
query1 = "كيفية فرز قائمة في بايثون"
prompt1 = generate_simple_prompt(query1, context="برمجة")
print(f"الأمر الأول: {prompt1}\n")

query2 = "هاتف ذكي جديد بميزات كاميرا متقدمة"
prompt2 = generate_simple_prompt(query2, context="وصف منتج")
print(f"الأمر الثاني: {prompt2}\n")

query3 = "ما هي عاصمة فرنسا؟"
prompt3 = generate_simple_prompt(query3)
print(f"الأمر الثالث: {prompt3}\n")
```

```
# <div dir="rtl"> مثال متقدم: استخدام نموذج لغوي كبير (LLM) لتوليد الأوامر </div>
# <div dir="rtl"> هنا، يمكن للنموذج نفسه أن يساعد في صياغة الأمر الأولي. </div>

from langchain.prompts import PromptTemplate
from langchain.chains import LLMChain

# <div dir="rtl"> قالب الأمر لتوليد الأوامر </div>
prompt_template_for_prompt_generation = PromptTemplate.from_template(
    """
    <div dir="rtl">
    أنت خبير في صياغة الأوامر. بناءً على طلب المستخدم التالي والسياق المحدد،
    قم بإنشاء أمر مفصل وموجه لنموذج لغوي آخر ليقوم بتنفيذ المهمة.
    الطلب: {user_request}
    السياق: {context}
    الأمر الذي يجب توليده:
    </div>
    """
)

# <div dir="rtl"> إنشاء سلسلة (Chain) لتوليد الأوامر </div>
prompt_generation_chain = LLMChain(llm=llm, prompt=prompt_template_for_prompt_generation)

# <div dir="rtl"> تجربة توليد أمر باستخدام LLM </div>
user_request_llm = "اكتب مقالًا قصيرًا عن فوائد تعلم البرمجة"
context_llm = "مقالات تعليمية، جمهور مبتدئ"

generated_prompt_llm = prompt_generation_chain.run(user_request=user_request_llm, context=context_llm)
print(f"الأمر الذي تم توليده بواسطة LLM:\n{generated_prompt_llm}\n")

# <div dir="rtl"> الآن يمكننا استخدام هذا الأمر مع LLM آخر (أو نفس LLM) </div>
print("--- تنفيذ الأمر الذي تم توليده ---")
response_from_generated_prompt = llm.invoke(generated_prompt_llm)
print(f"استجابة LLM للأمر الذي تم توليده:\n{response_from_generated_prompt.content}")
```

## 2. التكيف الذاتي للأوامر (Adaptive Prompting)

التكيف الذاتي للأوامر يعني أن الأمر لا يكون ثابتًا، بل يتغير ويتطور استجابةً للمعلومات الجديدة، التفاعلات السابقة، أو التغيرات في السياق. هذا يسمح للأنظمة بأن تكون أكثر ديناميكية وذكاءً.
</div>

```
# <div dir="rtl"> مثال: أمر يتكيف بناءً على تاريخ التفاعل </div>

def get_adaptive_prompt(user_name, conversation_history=None):
    """
    <div dir="rtl">
    تولد أمرًا يتكيف بناءً على اسم المستخدم وتاريخ المحادثة.
    </div>
    """
    base_prompt = f"مرحبًا {user_name}. كيف يمكنني مساعدتك اليوم؟"
    if conversation_history:
        # <div dir="rtl"> إذا كان هناك تاريخ للمحادثة، قم بتضمينه لتوفير سياق إضافي </div>
        history_str = "\\n".join([f"المستخدم: {h['user']}\\nالمساعد: {h['assistant']}" for h in conversation_history])
        return f"{base_prompt}\\n\\nبناءً على محادثتنا السابقة:\\n{history_str}\\n\\nما هو استفسارك الحالي؟"
    else:
        return base_prompt

# <div dir="rtl"> تجربة الأمر التكيفي </div>
user = "أحمد"
print(f"الأمر الأولي لأحمد:\n{get_adaptive_prompt(user)}\n")

# <div dir="rtl"> محادثة سابقة </div>
history = [
    {"user": "ما هي أفضل لغة برمجة للمبتدئين؟", "assistant": "بايثون ممتازة للمبتدئين."},
    {"user": "وماذا عن تطوير الويب؟", "assistant": "JavaScript و Python (مع Django/Flask) خيارات رائعة."}
]

print(f"الأمر التكيفي لأحمد بعد المحادثة:\n{get_adaptive_prompt(user, history)}\n")

# <div dir="rtl"> يمكن استخدام هذا الأمر التكيفي مع LLM </div>
# <div dir="rtl"> (لن نقوم بتنفيذه هنا لتجنب استدعاء LLM إضافي، لكن الفكرة واضحة) </div>
# response_from_adaptive_prompt = llm.invoke(get_adaptive_prompt(user, history))
# print(response_from_adaptive_prompt.content)
```

## 3. دمج هندسة الأوامر مع الأدوات والوكلاء (Tools & Agents)

هنا يكمن جوهر بناء الأنظمة الذكية المعقدة. سنستخدم مكتبات مثل LangChain لبناء وكلاء (Agents) يمكنهم التفاعل مع أدوات خارجية (Tools) لتنفيذ مهام تتجاوز قدرات النموذج اللغوي وحده.
</div>

```
# <div dir="rtl"> إعداد أدوات وهمية (Mock Tools) للمحاكاة </div>
# <div dir="rtl"> في التطبيقات الحقيقية، ستكون هذه الأدوات عبارة عن استدعاءات API حقيقية </div>

from langchain.tools import tool

@tool
def get_current_weather(location: str) -> str:
    """
    <div dir="rtl">
    الحصول على حالة الطقس الحالية لموقع معين.
    </div>
    """
    if "القاهرة" in location:
        return "مشمس ودافئ، درجة الحرارة 30 مئوية."
    elif "لندن" in location:
        return "غائم جزئيًا، درجة الحرارة 15 مئوية مع احتمال هطول أمطار خفيفة."
    else:
        return f"لا تتوفر بيانات الطقس لـ {location}."

@tool
def search_wikipedia(query: str) -> str:
    """
    <div dir="rtl">
    البحث عن معلومات في ويكيبيديا حول موضوع معين.
    </div>
    """
    if "بايثون" in query:
        return "بايثون هي لغة برمجة عالية المستوى، متعددة الأغراض، ومفسرة. تم إنشاؤها بواسطة جيدو فان روسم."
    elif "الذكاء الاصطناعي" in query:
        return "الذكاء الاصطناعي هو فرع من علوم الحاسوب يهدف إلى إنشاء آلات ذكية."
    else:
        return f"لم يتم العثور على معلومات حول '{query}' في ويكيبيديا الوهمية."

# <div dir="rtl"> قائمة الأدوات التي سيستخدمها الوكيل </div>
tools = [get_current_weather, search_wikipedia]
print(f"تم إعداد الأدوات: {[tool.name for tool in tools]}")
```

```
# <div dir="rtl"> بناء وكيل بسيط باستخدام LangChain </div>

from langchain.agents import AgentExecutor, create_react_agent
from langchain import hub

# <div dir="rtl"> جلب الأمر الافتراضي لـ ReAct Agent من LangChain Hub </div>
# <div dir="rtl"> هذا الأمر يوجه الوكيل حول كيفية التفكير واستخدام الأدوات </div>
prompt = hub.pull("hwchase17/react")

# <div dir="rtl"> إنشاء الوكيل </div>
agent = create_react_agent(llm, tools, prompt)

# <div dir="rtl"> إنشاء منفذ الوكيل (Agent Executor) </div>
agent_executor = AgentExecutor(agent=agent, tools=tools, verbose=True, handle_parsing_errors=True)

# <div dir="rtl"> تجربة الوكيل </div>
print("\n--- تجربة الوكيل: سؤال عن الطقس ---")
response1 = agent_executor.invoke({"input": "ما هو الطقس في القاهرة اليوم؟"})
print(f"استجابة الوكيل: {response1['output']}\n")

print("\n--- تجربة الوكيل: سؤال عن ويكيبيديا ---")
response2 = agent_executor.invoke({"input": "من هو جيدو فان روسم؟"})
print(f"استجابة الوكيل: {response2['output']}\n")

print("\n--- تجربة الوكيل: سؤال لا يتطلب أداة ---")
response3 = agent_executor.invoke({"input": "ما هي عاصمة اليابان؟"})
print(f"استجابة الوكيل: {response3['output']}\n")
```

```
# <div dir="rtl"> حل التمرين العملي: بناء وكيل يتكيف مع نوع الاستعلام </div>

# <div dir="rtl"> الأدوات المتاحة (نفس الأدوات السابقة) </div>
# <div dir="rtl"> الوكيل ReAct مصمم ليكون قادرًا على اختيار الأداة المناسبة أو الإجابة مباشرة </div>
# <div dir="rtl"> لذلك، لا نحتاج لتعديل كبير في بنية الوكيل نفسه، بل نعتمد على قدرة LLM على التفكير </div>

from langchain.agents import AgentExecutor, create_react_agent
from langchain import hub

# <div dir="rtl"> جلب الأمر الافتراضي لـ ReAct Agent </div>
prompt_solution = hub.pull("hwchase17/react")

# <div dir="rtl"> إنشاء الوكيل باستخدام نفس الأدوات والنموذج اللغوي </div>
agent_solution = create_react_agent(llm, tools, prompt_solution)

# <div dir="rtl"> إنشاء منفذ الوكيل </div>
agent_executor_solution = AgentExecutor(agent=agent_solution, tools=tools, verbose=True, handle_parsing_errors=True)

# <div dir="rtl"> اختبار الوكيل ببعض الاستعلامات </div>
print("\n--- اختبار الوكيل (الحل): سؤال عن الرياضيات ---")
response_math_solution = agent_executor_solution.invoke({"input": "ما هو حاصل جمع 15 و 27؟"})
print(f"استجابة الوكيل: {response_math_solution['output']}\n")

print("\n--- اختبار الوكيل (الحل): سؤال عن ويكيبيديا ---")
response_wiki_solution = agent_executor_solution.invoke({"input": "ما هي نظرية النسبية؟"})
print(f"استجابة الوكيل: {response_wiki_solution['output']}\n")

print("\n--- اختبار الوكيل (الحل): سؤال عن الطقس ---")
response_weather_solution = agent_executor_solution.invoke({"input": "ما هو الطقس في لندن؟"})
print(f"استجابة الوكيل: {response_weather_solution['output']}\n")

print("<div dir=\"rtl\"> **ملاحظة:** الوكيل ReAct (Reasoning and Acting) مصمم خصيصًا لتقييم الاستعلام وتحديد ما إذا كان يحتاج إلى أداة أو يمكنه الإجابة مباشرة. لذا، فإن الوكيل الافتراضي غالبًا ما يكون كافيًا لهذا النوع من التكيف. التحدي يكمن في صياغة الأدوات والأوامر الأولية بشكل جيد. </div>")
```

## 4. تحدي المشروع التطبيقي (Project Challenge)

لتجهيزكم لبناء أنظمة معقدة، سنركز على التطبيق العملي للتقنيات المتقدمة من خلال التحديات التالية:

### 1. ورشة عمل 'بناء وكيل ذكي (Agent)'

*   **المهمة**: سيعمل الطلاب في مجموعات على مشروع متكامل لإنشاء وكيل بسيط باستخدام LangChain أو LlamaIndex.
*   **المتطلبات**: يجب على الوكيل التفاعل مع أدوات خارجية (مثل API محرك بحث مخصص، أو قاعدة بيانات - يمكن محاكاتها كما فعلنا في الأمثلة) لتنفيذ مهمة معقدة.
*   **مثال للمهمة**: 'التخطيط لرحلة عمل مع مراعاة القيود والميزانية' (يتطلب البحث عن فنادق، رحلات طيران، أسعار، إلخ).
*   **التركيز**: تصميم الأوامر التي توجه الوكيل بفعالية، وكيفية جعله يتخذ قرارات بناءً على المعلومات المسترجعة من الأدوات.

### 2. تحدي بناء المساعد الشخصي

*   **السيناريو**: سيتم إعطاء فرق الطلاب سيناريو لمساعد شخصي يتطلب توليد أوامر تلقائياً (مثل 'مساعد خدمة عملاء يتعامل مع استفسارات متنوعة ومعقدة').
*   **المنافسة**: تتنافس الفرق على بناء النموذج الأكثر فعالية في التعامل مع مجموعة واسعة من الاستفسارات، مع التركيز على كفاءة توليد الأوامر والتكيف مع سياق المحادثة.

### 3. عروض مشاريع مصغرة سريعة

*   **العرض**: كل فريق يقدم مشروع الوكيل أو المساعد الخاص به في 5 دقائق.
*   **التركيز**: تسليط الضوء على تحديات تصميم الأوامر التوليدية والتكيفية وكيف تغلبوا عليها، مما يعزز مهارات العرض والتحليل النقدي.
